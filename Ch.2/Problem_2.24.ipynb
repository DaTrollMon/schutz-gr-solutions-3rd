# Chapter 2, Problem 24 — pair annihilation needs two photons

**Problem.** Prove that conservation of four-momentum forbids a reaction in which an electron and positron annihilate and produce a single photon ($\gamma$-ray). Prove that the production of two photons is not forbidden.

## One photon is impossible

Let the electron and positron (each of rest mass $m > 0$) have four-momenta $\vec p_-$ and $\vec p_+$. The total is $\vec P = \vec p_- + \vec p_+$.

**Step 1 — the total is timelike.**
$$
\vec P\cdot\vec P = \vec p_-^{\,2} + \vec p_+^{\,2} + 2\,\vec p_-\cdot\vec p_+ = -m^2 - m^2 + 2\,\vec p_-\cdot\vec p_+ .
$$
Evaluate $\vec p_-\cdot\vec p_+$ in the electron's rest frame: $\vec p_- \to (m, \mathbf 0)$ and $\vec p_+ \to (E_+, \mathbf p_+)$ with $E_+ \geq m > 0$. So $\vec p_-\cdot\vec p_+ = -mE_+ \leq -m^2$. Therefore
$$
\vec P\cdot\vec P \leq -2m^2 - 2m^2 = -4m^2 < 0 .
$$
(Equivalently: in the CM frame $\vec P \to (E_{CM}, \mathbf 0)$ with $E_{CM} \geq 2m$, so $\vec P^{\,2} = -E_{CM}^2 \leq -4m^2$.)

**Step 2 — a single photon is null.** A photon's four-momentum $\vec k$ has $\vec k\cdot\vec k = 0$ (Eq. 2.37).

**Step 3 — contradiction.** Conservation would require $\vec k = \vec P$, hence $\vec k\cdot\vec k = \vec P\cdot\vec P$, i.e. $0 \leq -4m^2$. Impossible. $\blacksquare$

**Physical picture.** In the CM frame the pair has zero total momentum but energy $\geq 2m$. A single photon always carries momentum equal to its energy ($|\mathbf k| = k^0$), so it can't have nonzero energy and zero momentum. The magnitude of a four-vector is frame-invariant, so no frame can fix this.

## Two photons are allowed

We need two null vectors that add up to $\vec P$. Work in the CM frame, $\vec P \to (E_{CM}, \mathbf 0)$. Take two photons back to back, each with energy $E_{CM}/2$, along any direction $\hat{\mathbf n}$:
$$
\vec k_1 \to \tfrac{E_{CM}}2(1, \hat{\mathbf n}),\qquad \vec k_2 \to \tfrac{E_{CM}}2(1, -\hat{\mathbf n}).
$$
Each is null ($-1 + |\hat{\mathbf n}|^2 = 0$), and $\vec k_1 + \vec k_2 \to (E_{CM}, \mathbf 0) = \vec P$ ✓. Since four-momentum conservation is a vector equation, it then holds in every frame (Problem 8b). So two-photon annihilation is permitted, and is what's observed: e.g. two 511 keV photons for an electron and positron annihilating at rest (used in PET scanners).

**Consistency of the magnitudes.** $(\vec k_1 + \vec k_2)^2 = 2\,\vec k_1\cdot\vec k_2 = 2\cdot\tfrac{E_{CM}^2}{4}(-1 - 1) = -E_{CM}^2$ ✓. Two null vectors *can* add to a timelike one, provided they aren't parallel; one null vector never can.

```python
import numpy as np
from schutz_gr import minkowski_dot, four_velocity, boost_general
rng = np.random.default_rng(24)
m = 0.511   # MeV
worst = -np.inf
for _ in range(1000):
    pm = m * four_velocity(rng.uniform(-0.55, 0.55, 3)); pp = m * four_velocity(rng.uniform(-0.55, 0.55, 3))
    P = pm + pp
    worst = max(worst, minkowski_dot(P, P))
print(f"largest P·P over 1000 random e⁻e⁺ pairs: {worst:.4f}  (≤ −4m² = {-4*m**2:.4f}; a photon needs 0)")
# two photons in the CM frame, transformed back to the lab
P = m*four_velocity([0.3, -0.2, 0.1]) + m*four_velocity([-0.1, 0.4, 0.0])
vcm = P[1:]/P[0]; Ecm = np.sqrt(-minkowski_dot(P, P))
n = np.array([0, 0.6, 0.8])
k1, k2 = Ecm/2*np.r_[1, n], Ecm/2*np.r_[1, -n]
Linv = np.linalg.inv(boost_general(vcm))
K1, K2 = Linv @ k1, Linv @ k2
print("two photons: k1² =", round(minkowski_dot(K1, K1), 12), " k2² =", round(minkowski_dot(K2, K2), 12),
      " k1 + k2 − P =", np.round(K1 + K2 - P, 12))
```
