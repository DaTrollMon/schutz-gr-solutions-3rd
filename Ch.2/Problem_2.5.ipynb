# Chapter 2, Problem 5 — linear independence

**Problem (paraphrased).** A set of vectors $\{\vec a, \vec b, \vec c, \vec d\}$ is **linearly independent** if no linear combination of them is zero except the trivial one, $0\vec a + 0\vec b + 0\vec c + 0\vec d = 0$.
- **(a)** Show that the basis vectors of Eq. (2.9) are linearly independent.
- **(b)** Is the set $\{\vec a, \vec b, \vec c, 5\vec a + 3\vec b - 2\vec c\}$ linearly independent?

**Eq. (2.9):** $\vec e_0 \xrightarrow{\mathcal O}(1,0,0,0)$, $\vec e_1 \xrightarrow{\mathcal O}(0,1,0,0)$, $\vec e_2 \xrightarrow{\mathcal O}(0,0,1,0)$, $\vec e_3 \xrightarrow{\mathcal O}(0,0,0,1)$.

## (a) $\{\vec e_\alpha\}$ are linearly independent

Suppose some combination vanishes: $c^0\vec e_0 + c^1\vec e_1 + c^2\vec e_2 + c^3\vec e_3 = \vec 0$, i.e. $c^\alpha\vec e_\alpha = \vec 0$.

**Compute its components in $\mathcal O$** using Eq. (2.8) and the components (2.9):
$$
c^0(1,0,0,0) + c^1(0,1,0,0) + c^2(0,0,1,0) + c^3(0,0,0,1) = (c^0, c^1, c^2, c^3).
$$
(In index form: the $\beta$ component is $c^\alpha(\vec e_\alpha)^\beta = c^\alpha\delta_\alpha{}^\beta = c^\beta$, using Eq. 2.10.)

The zero vector has components $(0,0,0,0)$ (in every frame, Problem 8). So
$$
(c^0, c^1, c^2, c^3) = (0,0,0,0)\quad\Longrightarrow\quad c^0 = c^1 = c^2 = c^3 = 0 .
$$
Only the trivial combination vanishes, so the basis vectors are **linearly independent**. $\blacksquare$

(Equivalently: the matrix whose columns are the components of the $\vec e_\alpha$ is the identity, with determinant $1 \neq 0$.)

## (b) $\{\vec a, \vec b, \vec c, \vec d\}$ with $\vec d = 5\vec a + 3\vec b - 2\vec c$

**No — it is linearly dependent**, whatever $\vec a, \vec b, \vec c$ are. Move everything to one side:
$$
5\vec a + 3\vec b - 2\vec c - 1\cdot\vec d = \vec 0 .
$$
This is a linear combination equal to zero with coefficients $(5, 3, -2, -1)$, which are not all zero (the coefficient of $\vec d$ is $-1$ regardless). So the set is not linearly independent.

Geometrically, $\vec d$ lies in the span of $\vec a, \vec b, \vec c$, so these four vectors span at most a 3-dimensional subspace. They can't serve as a basis for 4D spacetime.

```python
import numpy as np
E = np.eye(4)                                   # columns = components of e_0..e_3
print("(a) det[e_0 e_1 e_2 e_3] =", np.linalg.det(E), " rank =", np.linalg.matrix_rank(E))

rng = np.random.default_rng(5)
a, b, c = rng.normal(size=(3, 4))               # three arbitrary vectors
d = 5*a + 3*b - 2*c
M = np.column_stack([a, b, c, d])
print("(b) rank of [a b c d] =", np.linalg.matrix_rank(M), "(< 4 ⇒ dependent)")
print("    5a + 3b − 2c − d =", np.round(5*a + 3*b - 2*c - d, 12))
```
