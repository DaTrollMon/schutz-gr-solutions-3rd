# Chapter 2, Problem 4 — vector arithmetic in components

**Problem.** Given $\vec A \xrightarrow{\mathcal O} (5, -1, 0, 1)$ and $\vec B \xrightarrow{\mathcal O} (-2, 1, 1, -6)$, find the components in $\mathcal O$ of **(a)** $-6\vec A$; **(b)** $3\vec A + \vec B$; **(c)** $-6\vec A + 3\vec B$.

**Rules (Eq. 2.8).** Vectors add component by component, and multiplying by a number multiplies every component:
$$
\vec A + \vec B \xrightarrow{\mathcal O}(A^0 + B^0, \dots, A^3 + B^3),\qquad \mu\vec A \xrightarrow{\mathcal O}(\mu A^0, \dots, \mu A^3).
$$
These rules are consistent between frames because the Lorentz transformation is linear: $\Lambda(\mu\vec A + \vec B) = \mu\Lambda\vec A + \Lambda\vec B$.

## (a) $-6\vec A$
$$
-6\,(5, -1, 0, 1) = (-30,\ 6,\ 0,\ -6) \quad\Rightarrow\quad \boxed{-6\vec A \xrightarrow{\mathcal O} (-30,\ 6,\ 0,\ -6)}
$$

## (b) $3\vec A + \vec B$
$$
3\vec A = (15, -3, 0, 3),\qquad 3\vec A + \vec B = (15 - 2,\ -3 + 1,\ 0 + 1,\ 3 - 6) \quad\Rightarrow\quad \boxed{(13,\ -2,\ 1,\ -3)}
$$

## (c) $-6\vec A + 3\vec B$
$$
-6\vec A = (-30, 6, 0, -6),\quad 3\vec B = (-6, 3, 3, -18),\quad\text{sum} \Rightarrow \boxed{(-36,\ 9,\ 3,\ -24)}
$$

```python
import numpy as np
from schutz_gr import boost_matrix
A = np.array([5, -1, 0, 1]); B = np.array([-2, 1, 1, -6])
for name, V in (("(a) −6A", -6*A), ("(b) 3A + B", 3*A + B), ("(c) −6A + 3B", -6*A + 3*B)):
    print(f"{name:14s} →", V)

# Linearity check: transform first then combine = combine then transform (any boost)
L = boost_matrix(0.6)
print("\nframe-independence of the rules:", np.allclose(L @ (-6*A + 3*B), -6*(L @ A) + 3*(L @ B)))
```
