# Chapter 2, Problem 34 — the dot product is bilinear

**Problem.** Show that, if $\vec A$, $\vec B$, $\vec C$ are any vectors and $\alpha$, $\beta$ any real numbers,
$$
(\alpha\vec A)\cdot\vec B = \alpha(\vec A\cdot\vec B),\qquad
\vec A\cdot(\beta\vec B) = \beta(\vec A\cdot\vec B),
$$
$$
\vec A\cdot(\vec B + \vec C) = \vec A\cdot\vec B + \vec A\cdot\vec C,\qquad
(\vec A + \vec B)\cdot\vec C = \vec A\cdot\vec C + \vec B\cdot\vec C .
$$

**Tools.** Components of sums and multiples, Eq. (2.8): $(\alpha\vec A)^\mu = \alpha A^\mu$, $(\vec B + \vec C)^\mu = B^\mu + C^\mu$. The dot product, Eq. (2.26): $\vec A\cdot\vec B = -A^0B^0 + A^1B^1 + A^2B^2 + A^3B^3 = \eta_{\mu\nu}A^\mu B^\nu$, with $\eta = \mathrm{diag}(-1, 1, 1, 1)$. Since the dot product is frame-independent (§2.5), proving each identity in one frame proves it in all.

## 1. $(\alpha\vec A)\cdot\vec B = \alpha(\vec A\cdot\vec B)$

$$
(\alpha\vec A)\cdot\vec B = -(\alpha A^0)B^0 + (\alpha A^1)B^1 + (\alpha A^2)B^2 + (\alpha A^3)B^3 = \alpha\left(-A^0B^0 + A^1B^1 + A^2B^2 + A^3B^3\right) = \alpha(\vec A\cdot\vec B).
$$
(Factor $\alpha$ out of a sum of real numbers: distributivity of ordinary multiplication.)

## 2. $\vec A\cdot(\beta\vec B) = \beta(\vec A\cdot\vec B)$

Identical, with $\beta$ in the second slot: $-A^0(\beta B^0) + \dots = \beta(-A^0B^0 + \dots)$. (Or: the dot product is symmetric, $\vec A\cdot\vec B = \vec B\cdot\vec A$, since $A^\mu B^\mu = B^\mu A^\mu$ term by term. Then 2 follows from 1.)

## 3. $\vec A\cdot(\vec B + \vec C) = \vec A\cdot\vec B + \vec A\cdot\vec C$

$$
\begin{aligned}
\vec A\cdot(\vec B + \vec C) &= -A^0(B^0 + C^0) + A^1(B^1 + C^1) + A^2(B^2 + C^2) + A^3(B^3 + C^3)\\
&= \left(-A^0B^0 + A^1B^1 + A^2B^2 + A^3B^3\right) + \left(-A^0C^0 + A^1C^1 + A^2C^2 + A^3C^3\right)\\
&= \vec A\cdot\vec B + \vec A\cdot\vec C .
\end{aligned}
$$

## 4. $(\vec A + \vec B)\cdot\vec C = \vec A\cdot\vec C + \vec B\cdot\vec C$

Same steps with the sum in the first slot, or use symmetry plus 3.

**In index notation, all four at once:** $\eta_{\mu\nu}(\alpha A^\mu + \beta B^\mu)C^\nu = \alpha\,\eta_{\mu\nu}A^\mu C^\nu + \beta\,\eta_{\mu\nu}B^\mu C^\nu$, and likewise in the second slot. The dot product is **linear in each argument separately** (bilinear). Chapter 3 builds on exactly this: $\eta$ is a $\binom02$ tensor, the metric. $\blacksquare$

**Why it matters here.** Bilinearity is what §2.5 used to prove frame-independence of $\vec A\cdot\vec B$ from that of magnitudes: $(\vec A + \vec B)^2 = \vec A^{\,2} + 2\vec A\cdot\vec B + \vec B^{\,2}$. It's also used in Problems 18, 24 and 32.

```python
import sympy as sp
from schutz_gr import minkowski_dot
A = sp.Matrix(sp.symbols("A0:4")); B = sp.Matrix(sp.symbols("B0:4")); C = sp.Matrix(sp.symbols("C0:4"))
al, be = sp.symbols("alpha beta", real=True)
checks = {
    "(αA)·B = α(A·B)":     minkowski_dot(al*A, B) - al*minkowski_dot(A, B),
    "A·(βB) = β(A·B)":     minkowski_dot(A, be*B) - be*minkowski_dot(A, B),
    "A·(B+C) = A·B + A·C": minkowski_dot(A, B + C) - minkowski_dot(A, B) - minkowski_dot(A, C),
    "(A+B)·C = A·C + B·C": minkowski_dot(A + B, C) - minkowski_dot(A, C) - minkowski_dot(B, C),
    "A·B = B·A":           minkowski_dot(A, B) - minkowski_dot(B, A),
}
for k, d in checks.items(): print(f"{k:22s}", sp.expand(d) == 0)
```
