# Chapter 2, Problem 17 — every unit future-timelike vector is a four-velocity

**Problem.**
- **(a)** Prove that any timelike vector $\vec U$ with $U^0 > 0$ and $\vec U\cdot\vec U = -1$ is the four-velocity of some world line.
- **(b)** Use this to prove that for any timelike vector $\vec V$ there is a Lorentz frame in which $\vec V$ has zero spatial components.

## (a)

Let $\vec U \xrightarrow{\mathcal O}(U^0, U^1, U^2, U^3)$ with $U^0 > 0$ and $-(U^0)^2 + \sum_i(U^i)^2 = -1$.

**Step 1 — define a candidate velocity.** Since $U^0 > 0$ we can divide: $v^i \equiv U^i/U^0$.

**Step 2 — it's subluminal.** From the normalization, $\sum_i(U^i)^2 = (U^0)^2 - 1 < (U^0)^2$. Dividing by $(U^0)^2$:
$$
|\mathbf v|^2 = \frac{\sum_i(U^i)^2}{(U^0)^2} = 1 - \frac{1}{(U^0)^2} < 1 .
$$
So $\mathbf v$ is the velocity of a possible material particle.

**Step 3 — $U^0$ is the right $\gamma$.** From Step 2, $(U^0)^2 = \dfrac{1}{1 - |\mathbf v|^2}$, and since $U^0 > 0$, $U^0 = (1 - |\mathbf v|^2)^{-1/2} = \gamma(\mathbf v)$. Then $U^i = v^iU^0 = \gamma v^i$.

**Step 4 — conclude.** $\vec U \to \gamma(1, \mathbf v)$, which by Problem 15(b) is exactly the four-velocity of the world line $\mathbf x(t) = \mathbf v t$ (or any line parallel to it). $\blacksquare$

The two conditions are both needed: $U^0 > 0$ makes it *future*-pointing (a past-pointing $-\gamma(1, \mathbf v)$ is no four-velocity), and $\vec U\cdot\vec U = -1$ fixes the length (a timelike vector of the wrong length points the right way but isn't $d\vec x/d\tau$).

## (b)

Let $\vec V$ be timelike: $\vec V\cdot\vec V < 0$. Then $V^0 \neq 0$ (otherwise $\vec V\cdot\vec V = \sum(V^i)^2 \geq 0$).

**Normalize.** Define $\vec U = \pm\vec V/\sqrt{-\vec V\cdot\vec V}$, with the sign chosen so that $U^0 > 0$. Then $\vec U\cdot\vec U = \vec V\cdot\vec V/(-\vec V\cdot\vec V) = -1$ and $U^0 > 0$.

**Apply (a).** $\vec U$ is the four-velocity of a world line with velocity $\mathbf v = \mathbf U/U^0$. Go to the rest frame $\bar{\mathcal O}$ of that world line (a boost by $\mathbf v$, Problem 1.21). There $\vec U = \vec e_{\bar0} \xrightarrow{\bar{\mathcal O}}(1, 0, 0, 0)$.

**Undo the normalization.** $\vec V = \pm\sqrt{-\vec V\cdot\vec V}\,\vec U \xrightarrow{\bar{\mathcal O}}\big(\pm\sqrt{-\vec V\cdot\vec V},\ 0,\ 0,\ 0\big)$: **zero spatial components.** $\blacksquare$

This is the vector version of Problem 1.21(a): a timelike displacement has a frame in which it's purely temporal.

```python
import numpy as np
from schutz_gr import minkowski_dot, boost_general
rng = np.random.default_rng(17)
for _ in range(3):
    s = rng.normal(size=3) * 0.5
    V = np.r_[np.sign(rng.normal()) * (np.linalg.norm(s) + rng.uniform(0.1, 2)), s]    # random timelike vector
    U = np.sign(V[0]) * V / np.sqrt(-minkowski_dot(V, V))
    v = U[1:] / U[0]
    Vbar = boost_general(v) @ V
    print(f"V = {np.round(V, 3)}, V·V = {minkowski_dot(V, V):+.3f}, |v| = {np.linalg.norm(v):.3f} → V in rest frame = {np.round(Vbar, 10)}")
```
