# Chapter 2, Problem 16 — velocity addition from the four-velocity

**Problem.** Derive the Einstein velocity-addition formula by performing a Lorentz transformation with velocity $v$ on the four-velocity of a particle whose speed in the original frame was $W$.

**Idea.** Ordinary velocities transform awkwardly, but the four-velocity is a four-vector: its components transform *linearly* by the Lorentz matrix. Then read off the new 3-velocity as $U^{\bar1}/U^{\bar0}$ (Problem 15c).

## Derivation

**Original frame $\mathcal O$.** The particle moves along $x$ with speed $W$. By Problem 15:
$$
\vec U \xrightarrow{\mathcal O}\big(\gamma_W,\ \gamma_W W,\ 0,\ 0\big),\qquad \gamma_W = (1 - W^2)^{-1/2}.
$$

**Transform to $\bar{\mathcal O}$**, moving at $v$ along $x$ relative to $\mathcal O$ ($\gamma_v = (1-v^2)^{-1/2}$):
$$
U^{\bar0} = \gamma_v\big(U^0 - vU^1\big) = \gamma_v\gamma_W(1 - vW),\qquad
U^{\bar1} = \gamma_v\big(U^1 - vU^0\big) = \gamma_v\gamma_W(W - v).
$$

**Read off the velocity in $\bar{\mathcal O}$:**
$$
\bar W = \frac{U^{\bar1}}{U^{\bar0}} = \frac{\gamma_v\gamma_W(W - v)}{\gamma_v\gamma_W(1 - vW)} \quad\Longrightarrow\quad
\boxed{\bar W = \frac{W - v}{1 - Wv}}
$$

**Matching Schutz's Eq. (1.13).** Eq. (1.13) goes the other way: a particle with speed $W$ in $\bar{\mathcal O}$, which moves at $v$ relative to $\mathcal O$, has speed $W' = (W + v)/(1 + Wv)$ in $\mathcal O$. That's the same computation with the inverse transformation, $\Lambda(-v)$, i.e. $v\to-v$ above:
$$
U^0 = \gamma_v\gamma_W(1 + vW),\quad U^1 = \gamma_v\gamma_W(W + v)\quad\Longrightarrow\quad W' = \frac{W + v}{1 + Wv}.\tag{1.13}
$$

**Bonus: the time component gives the gamma factor of the composed velocity.**
$$
\gamma_{W'} = U^0 = \gamma_v\gamma_W(1 + vW).
$$
(Check: $1 - W'^2 = \frac{(1+vW)^2 - (W+v)^2}{(1+vW)^2} = \frac{(1-v^2)(1-W^2)}{(1+vW)^2}$ ✓.) This is the four-vector method's advantage: one linear transformation gives both the new velocity and its $\gamma$.

```python
import sympy as sp
from schutz_gr import boost_matrix, four_velocity
W, v = sp.symbols("W v", real=True)
U = four_velocity([W, 0, 0])
Ub = sp.simplify(boost_matrix(v) * U)
print("Ū⁰ =", sp.factor(Ub[0]), "   Ū¹ =", sp.factor(Ub[1]))
print("W̄ = Ū¹/Ū⁰ =", sp.simplify(Ub[1] / Ub[0]))
Uo = sp.simplify(boost_matrix(-v) * U)
print("with Λ(−v):  W' =", sp.simplify(Uo[1] / Uo[0]))
Wp = (W + v) / (1 + W * v)
print("γ_{W'} − γ_v γ_W (1 + vW) =", sp.simplify(1/sp.sqrt(1 - Wp**2) - (1 + v*W)/sp.sqrt(1 - v**2)/sp.sqrt(1 - W**2)).subs({W: sp.Rational(1, 3), v: sp.Rational(1, 2)}))
```
