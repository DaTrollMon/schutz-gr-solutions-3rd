# Chapter 2, Problem 19 — uniform acceleration and the 1 g rocket

**Problem (paraphrased).** A body is **uniformly accelerated** if its acceleration four-vector $\vec a$ has constant spatial direction and constant magnitude, $\vec a\cdot\vec a = \alpha^2 > 0$.
- **(a)** Show that $\vec a$ always has the same components in the body's MCRF, and that these are what one would call "acceleration" in Galilean terms (a rocket whose engine always gives the same acceleration).
- **(b)** A body is uniformly accelerated with $\alpha = 10\ \text{m s}^{-2}$, starting from rest. Find its speed after time $t$ (correct units!). How far has it travelled? How long to reach $v = 0.999$?
- **(c)** Find the elapsed proper time as a function of $t$ (integrate $d\tau$ along the world line). How much proper time has elapsed when $v = 0.999$? How much would such a traveller age on a trip from Earth to the centre of our Galaxy, about $2\times10^{20}$ m?

## (a) Constant components in the MCRF

**Orthogonality.** Since $\vec U\cdot\vec U = -1$ always, differentiating gives $\vec U\cdot\vec a = 0$ (Eq. 2.32). In the MCRF, $\vec U = \vec e_{\bar0} \to (1, 0, 0, 0)$, so $\vec U\cdot\vec a = -a^{\bar0} = 0$:
$$
\vec a \xrightarrow{\text{MCRF}}(0,\ a^{\bar1},\ a^{\bar2},\ a^{\bar3}),\qquad \vec a\cdot\vec a = (a^{\bar1})^2 + (a^{\bar2})^2 + (a^{\bar3})^2 = \alpha^2 .
$$
Choosing the MCRF's $x$ axis along the (constant) spatial direction of the acceleration:
$$
\boxed{\vec a \xrightarrow{\text{MCRF}}(0,\ \alpha,\ 0,\ 0)}\quad\text{at every event on the world line.}
$$

**Galilean meaning.** In the MCRF the body is momentarily at rest, $\mathbf v = 0$. Expand to first order in the small velocity it picks up: $\vec U = \gamma(1, \mathbf v) = (1, \mathbf v) + O(v^2)$ and $d\tau = dt\,(1 + O(v^2))$. So
$$
a^{\bar i} = \frac{dU^{\bar i}}{d\tau} = \frac{dv^{\bar i}}{dt}\quad\text{at the moment } \mathbf v = 0 .
$$
That's the ordinary Newtonian acceleration measured by an inertial observer momentarily comoving with the body: what an accelerometer aboard reads. A rocket with a constant thrust per unit mass (constant felt "g") is uniformly accelerated in this sense.

## Solving for the world line (needed for b, c)

Take motion along $x$. Because $\vec U\cdot\vec U = -1$ and $U^0 > 0$, we can always write
$$
U^0 = \cosh\theta(\tau),\qquad U^1 = \sinh\theta(\tau)
$$
($\theta$ is the rapidity, Problem 1.18). Then
$$
\vec a = \frac{d\vec U}{d\tau} = \dot\theta\,(\sinh\theta,\ \cosh\theta),\qquad \vec a\cdot\vec a = \dot\theta^2(-\sinh^2\theta + \cosh^2\theta) = \dot\theta^2 = \alpha^2 .
$$
So $\dot\theta = \alpha$ and, starting from rest ($\theta = 0$ at $\tau = 0$), $\theta = \alpha\tau$: **rapidity grows linearly with proper time.** Integrate $dt/d\tau = \cosh\alpha\tau$ and $dx/d\tau = \sinh\alpha\tau$ from the origin:
$$
t = \frac{\sinh\alpha\tau}{\alpha},\qquad x = \frac{\cosh\alpha\tau - 1}{\alpha},\qquad v = \tanh\alpha\tau .
$$
Eliminating $\tau$: $(x + 1/\alpha)^2 - t^2 = 1/\alpha^2$, a **hyperbola**. (Compare Problem 21.)

## (b) Speed, distance, and time to reach $v = 0.999$ (lab frame)

**Units.** With $c = 1$, an acceleration has units of m$^{-1}$ (Problem 1.2e):
$$
\alpha = \frac{10\ \text{m s}^{-2}}{c^2} = \frac{10}{9\times10^{16}}\ \text{m}^{-1} = 1.11\times10^{-16}\ \text{m}^{-1},
\qquad \frac1\alpha = 9\times10^{15}\ \text{m} \approx 0.95\ \text{light-year}.
$$

**Speed after lab time $t$.** From $\sinh\alpha\tau = \alpha t$: $\cosh\alpha\tau = \sqrt{1 + \alpha^2t^2}$, so
$$
\boxed{v(t) = \tanh\alpha\tau = \frac{\alpha t}{\sqrt{1 + \alpha^2t^2}}}\qquad\Big(\text{SI: } v = \frac{\alpha_{SI}t}{\sqrt{1 + (\alpha_{SI}t/c)^2}}\Big).
$$
Newtonian ($v \approx \alpha t$) for $\alpha t \ll 1$; tends to 1 but never reaches it.

**Distance.**
$$
\boxed{x(t) = \frac{\sqrt{1 + \alpha^2t^2} - 1}{\alpha}}\quad\approx \tfrac12\alpha t^2\ \text{early},\quad \approx t - 1/\alpha\ \text{late}.
$$

**Time to $v = 0.999$.** Invert: $\alpha t = v/\sqrt{1 - v^2} = \gamma v = 0.999\times22.366 = 22.34$.
$$
t = \frac{22.34}{1.11\times10^{-16}\ \text{m}^{-1}} = 2.01\times10^{17}\ \text{m} = 6.7\times10^{8}\ \text{s} \approx \boxed{21.2\ \text{years}}
$$
Distance by then: $x = (\gamma - 1)/\alpha = 21.37\times9\times10^{15} = 1.92\times10^{17}$ m $\approx 20$ light-years.

## (c) Proper time

$$
\tau(t) = \int_0^t \frac{dt'}{\gamma(t')} = \int_0^t\frac{dt'}{\sqrt{1 + \alpha^2t'^2}} = \boxed{\frac{1}{\alpha}\sinh^{-1}(\alpha t)}
$$
(using $1/\gamma = \sqrt{1 - v^2} = 1/\sqrt{1 + \alpha^2t^2}$). Same as $\theta = \alpha\tau$ above. For large $t$, $\tau \approx \ln(2\alpha t)/\alpha$: **proper time grows only logarithmically** with lab time.

**At $v = 0.999$:** $\alpha\tau = \tanh^{-1}(0.999) = \tfrac12\ln(1999) = 3.800$, so
$$
\tau = 3.800\times9\times10^{15}\ \text{m} = 3.42\times10^{16}\ \text{m} = 1.14\times10^8\ \text{s} \approx \boxed{3.6\ \text{years}}\quad(\text{vs. }21.2\text{ yr on Earth}).
$$

**Trip to the Galactic centre, $D = 2\times10^{20}$ m.** From $x = (\cosh\alpha\tau - 1)/\alpha$:
$$
\alpha\tau = \cosh^{-1}(1 + \alpha D) = \cosh^{-1}(1 + 22\,222) \approx \ln(2\times22\,223) = 10.70,
$$
$$
\tau \approx 10.70\times9\times10^{15}\ \text{m} = 9.6\times10^{16}\ \text{m} = 3.2\times10^{8}\ \text{s} \approx \boxed{10\ \text{years}}
$$
(accelerating the whole way; Earth-frame time $\approx D \approx 21\,000$ years). If instead the traveller accelerates halfway and decelerates the second half (to arrive at rest): $\tau = \tfrac2\alpha\cosh^{-1}(1 + \alpha D/2) \approx 19$ years. Either way, a human lifetime, versus $\sim 2\times10^4$ years for people at home.

```python
import numpy as np
c = 3e8
alpha = 10 / c**2                       # m^-1
yr = 3.156e7                            # s
print(f"α = {alpha:.4e} m^-1,  1/α = {1/alpha:.3e} m = {1/alpha/9.46e15:.3f} ly")
v = 0.999; g = 1/np.sqrt(1 - v**2)
t = g*v/alpha; x = (g - 1)/alpha; tau = np.arctanh(v)/alpha
print(f"(b) t(v=0.999) = {t:.3e} m = {t/c:.3e} s = {t/c/yr:.2f} yr;  distance = {x:.3e} m = {x/9.46e15:.1f} ly")
print(f"(c) τ(v=0.999) = {tau:.3e} m = {tau/c:.3e} s = {tau/c/yr:.2f} yr")
D = 2e20
tau_full = np.arccosh(1 + alpha*D)/alpha
tau_half = 2*np.arccosh(1 + alpha*D/2)/alpha
t_full = np.sinh(alpha*tau_full)/alpha
print(f"    Galactic centre, accelerate all the way: τ = {tau_full/c/yr:.2f} yr  (Earth time {t_full/c/yr:,.0f} yr)")
print(f"    accelerate half, decelerate half:        τ = {tau_half/c/yr:.2f} yr")
```

```python
import sympy as sp
a_, tau_, t_ = sp.symbols("alpha tau t", positive=True)
U = sp.Matrix([sp.cosh(a_*tau_), sp.sinh(a_*tau_), 0, 0])
acc = U.diff(tau_)
print("U·U =", sp.simplify(-U[0]**2 + U[1]**2), "   a·a =", sp.simplify(-acc[0]**2 + acc[1]**2), "   U·a =", sp.simplify(-U[0]*acc[0] + U[1]*acc[1]))
print("τ(t) =", sp.integrate(1/sp.sqrt(1 + a_**2*sp.Symbol("s")**2), (sp.Symbol("s"), 0, t_)))
```

```python
import matplotlib.pyplot as plt
from schutz_gr.diagrams import COLORS
%config InlineBackend.figure_format = 'retina'
T = np.linspace(0, 40, 400)          # years of Earth time
at = alpha * T * yr * c
fig, axs = plt.subplots(1, 3, figsize=(13, 3.8)); fig.patch.set_facecolor(COLORS["bg"])
axs[0].plot(T, at/np.sqrt(1 + at**2), color=COLORS["amber"], lw=2, label="relativistic"); mask = at <= 1.6
axs[0].plot(T[mask], at[mask], "--", color=COLORS["grey"], label="Newtonian αt")
axs[0].axhline(1, color=COLORS["gold"], ls=":", lw=1); axs[0].set_ylim(0, 1.6); axs[0].set_ylabel("v", color=COLORS["text"])
axs[1].plot(T, (np.sqrt(1 + at**2) - 1)/alpha/9.46e15, color=COLORS["teal"], lw=2); axs[1].set_ylabel("distance (light-years)", color=COLORS["text"])
axs[2].plot(T, np.arcsinh(at)/alpha/c/yr, color=COLORS["green"], lw=2, label="traveller's proper time τ"); axs[2].plot(T, T, ":", color=COLORS["grey"], label="Earth time t")
axs[2].set_ylabel("years", color=COLORS["text"])
for ax, title in zip(axs, ("speed", "distance travelled", "proper time")):
    ax.set_facecolor(COLORS["panel"]); ax.tick_params(colors=COLORS["grey"]); ax.grid(color=COLORS["grid"], lw=0.5)
    ax.set_xlabel("Earth time t (years)", color=COLORS["text"]); ax.set_title(title, color=COLORS["text"], loc="left", fontsize=10)
    for s in ax.spines.values(): s.set_visible(False)
    if ax.get_legend_handles_labels()[0]:
        leg = ax.legend(fontsize=8); leg.get_frame().set_facecolor(COLORS["bg"])
        for tx in leg.get_texts(): tx.set_color(COLORS["text"])
fig.suptitle("A 1 g (10 m/s²) rocket, starting from rest", color=COLORS["text"])
plt.tight_layout(); plt.show()
```
