# Chapter 2, Problem 33 — inverse Compton scattering off a cosmic-ray proton

**Problem.** Space is filled with cosmic rays (high-energy protons) and the cosmic microwave background (CMB). These can Compton-scatter off one another. Suppose a photon of energy $h\nu = 2\times10^{-4}$ eV scatters off a proton of energy $10^9 m_P = 10^{18}$ eV, both energies measured in the Sun's rest frame. Use Eq. (2.43) in the proton's initial rest frame to calculate the **maximum** final energy the photon can have in the solar rest frame after the scattering. What energy range is this (X-ray, visible, etc.)?

**Plan:** (1) transform the photon into the proton's rest frame (Doppler, Problem 25); (2) Compton-scatter there with Eq. (2.43); (3) transform back. At each step, choose the directions that maximize the final energy.

**Numbers.** $\gamma = E_P/m_P = 10^9$ ($m_P \approx 10^9$ eV $= 1$ GeV, as the problem states. With the precise 0.938 GeV, $\gamma$ for the same $10^{18}$ eV proton is $1.066\times10^9$, and since the answer scales as $\gamma^2$ it comes out about 14% higher, $\approx 9.1\times10^{14}$ eV. Same conclusion.). With $\gamma = 10^9$, $1 - v \approx 1/(2\gamma^2) = 5\times10^{-19}$, so $1 + v \approx 2$.

## Step 1 — the photon in the proton's rest frame

Let the proton move along $+x$ in the solar frame. A photon moving at angle $\theta$ to $+x$ has, in the proton frame, energy (Eq. 2.42)
$$
E' = \gamma(1 - v\cos\theta)\,E .
$$
This is largest for a **head-on** collision, $\theta = \pi$ (photon moving in $-x$): $E' = \gamma(1 + v)E \approx 2\gamma E$.
$$
E' \approx 2\times10^9\times2\times10^{-4}\ \text{eV} = 4\times10^{5}\ \text{eV} = 0.4\ \text{MeV}.
$$
The microwave photon looks like a gamma ray to the proton.

## Step 2 — Compton scattering in the proton's rest frame

The proton is at rest here, so Eq. (2.43) applies. Multiply it by $h$ and write $E = h\nu$ ($c = 1$):
$$
\frac{1}{E'_f} = \frac{1}{E'} + \frac{1 - \cos\theta'}{m_P}.
$$
To end up with the most energy in the **solar** frame, the scattered photon should go back along $+x$, the proton's direction of motion. Then the return boost (step 3) blueshifts it maximally. In the proton frame the incoming photon moved along $-x$, so going along $+x$ is **backscattering**, $\theta' = \pi$:
$$
E'_f = \frac{E'}{1 + 2E'/m_P} = \frac{4\times10^5\ \text{eV}}{1 + 8\times10^{-4}} \approx 3.997\times10^5\ \text{eV}.
$$
The recoil correction is tiny because $E' \ll m_P$ (Thomson regime). Backscattering costs the least energy compared with what the return boost gains.

## Step 3 — back to the solar frame

The proton frame moves at $+v$ relative to the Sun. A photon moving along $+x$ in the proton frame has solar-frame energy (inverse Doppler, $v\to-v$ in Eq. 2.42 with $\theta' = 0$):
$$
E_f = \gamma(1 + v)\,E'_f \approx 2\gamma E'_f = 2\times10^9\times3.997\times10^5\ \text{eV}.
$$
$$
\boxed{E_f^{\max} \approx 8\times10^{14}\ \text{eV} \approx 0.8\ \text{PeV}}
$$

**Compact formula.** Combining the steps: $E_f^{\max} = \dfrac{4\gamma^2E}{1 + 4\gamma E/m_P} \approx 4\gamma^2 E$, the classic inverse-Compton result: each scattering can multiply the photon energy by up to $\sim4\gamma^2$.

## What range is this?

$8\times10^{14}$ eV is far beyond X-rays (keV) and ordinary gamma rays (MeV). It's a **very-high-energy / ultra-high-energy gamma ray** (TeV–PeV range). Wavelength $\lambda = hc/E \approx 1.5\times10^{-21}$ m, smaller than a proton. The photon gains a factor of $4\times10^{18}$ in energy, but the proton barely notices (it loses about $10^{-3}$ of its energy).

```python
import numpy as np
E, EP, mP = 2e-4, 1e18, 1e9            # eV
g = EP/mP
one_plus_v = 1 + np.sqrt(1 - 1/g**2)
Ep = g * one_plus_v * E                 # step 1, head-on
Epf = Ep / (1 + 2*Ep/mP)                # step 2, backscatter
Ef = g * one_plus_v * Epf               # step 3
print(f"γ = {g:.0e},  E' = {Ep:.4e} eV,  E'_f = {Epf:.5e} eV,  E_f = {Ef:.4e} eV")
print(f"compact formula 4γ²E/(1 + 4γE/m): {4*g**2*E/(1 + 4*g*E/mP):.4e} eV")
mP_true = 938.272e6
g2 = EP/mP_true
print(f"with m_P = 938.272 MeV (γ = {g2:.4e}): E_f ≈ {4*g2**2*E/(1 + 4*g2*E/mP_true):.4e} eV")
print(f"wavelength hc/E_f = {1239.84e-9/Ef:.2e} m")
```

```python
# Brute-force check that head-on + backscatter is the maximum: scan initial and scattering angles
th_in = np.linspace(0, np.pi, 721)           # photon direction in solar frame (relative to +x)
th_sc = np.linspace(0, np.pi, 721)           # scattering angle in the proton frame (in the plane)
best = 0.0
np.seterr(invalid='ignore', divide='ignore')
for ti in th_in[::8]:
    # photon in proton frame: energy and direction (aberration)
    Epr = g*(1 - np.sqrt(1 - 1/g**2)*np.cos(ti))*E
    cos_pr = (np.cos(ti) - np.sqrt(1 - 1/g**2)) / (1 - np.sqrt(1 - 1/g**2)*np.cos(ti))
    phi_in = np.arccos(np.clip(cos_pr, -1, 1))
    for sign in (1, -1):
        phi_out = phi_in + sign*th_sc            # outgoing direction in proton frame
        Epf_ = Epr / (1 + Epr/mP*(1 - np.cos(th_sc)))
        Ef_ = g*(1 + np.sqrt(1 - 1/g**2)*np.cos(phi_out))*Epf_
        best = max(best, Ef_.max())
print(f"max over scanned geometries: {best:.4e} eV (analytic {Ef:.4e} eV)")
```
