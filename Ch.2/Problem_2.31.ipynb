# Chapter 2, Problem 31 — radiation pressure on a mirror

**Problem.** A photon of frequency $\nu$ is reflected without change of frequency from a mirror, with angle of incidence $\theta$. Calculate the momentum transferred to the mirror. What momentum would be transferred if the photon were absorbed rather than reflected?

**Setup.** Mirror surface in the $y$–$z$ plane, outward normal along $-x$, so the photon arrives moving in $+x$. The angle of incidence $\theta$ is measured from the normal (the usual optics convention). The photon travels in the $x$–$y$ plane. Work in the mirror's initial rest frame.

"Reflected without change of frequency" means the mirror takes no energy: it's effectively infinitely massive (a finite-mass mirror would recoil, and the reflected photon would be slightly redshifted).

## Four-momenta

Photon energy $E = h\nu$; spatial momentum magnitude $|\mathbf p| = h\nu$ (null, Eq. 2.37).
$$
\vec p_{\text{in}} \to h\nu\,(1,\ \cos\theta,\ \sin\theta,\ 0),\qquad
\vec p_{\text{out}} \to h\nu\,(1,\ -\cos\theta,\ \sin\theta,\ 0).
$$
Reflection reverses the normal component of the momentum and keeps the tangential component (the law of reflection, angle out = angle in). Equal frequencies give equal energies.

## Reflection

Conservation of total four-momentum: $\vec p_{\text{in}} + \vec P_{\text{mirror, before}} = \vec p_{\text{out}} + \vec P_{\text{mirror, after}}$. So the four-momentum transferred to the mirror is
$$
\Delta\vec P_{\text{mirror}} = \vec p_{\text{in}} - \vec p_{\text{out}} \to h\nu\,(0,\ 2\cos\theta,\ 0,\ 0).
$$
$$
\boxed{\Delta\mathbf p = 2h\nu\cos\theta\ \text{ along the normal, into the mirror}}\qquad\big(\text{SI: } 2h\nu\cos\theta/c\big)
$$
No energy transfer ($\Delta P^0 = 0$), consistent with the unchanged frequency. Normal incidence ($\theta = 0$) gives the maximum, $2h\nu$; grazing incidence ($\theta \to \pi/2$) transfers nothing.

## Absorption

Now nothing comes out. The mirror absorbs the photon's entire four-momentum:
$$
\Delta\vec P_{\text{mirror}} = \vec p_{\text{in}} \to h\nu\,(1,\ \cos\theta,\ \sin\theta,\ 0).
$$
$$
\boxed{\Delta\mathbf p = h\nu\ \text{ in the photon's direction:}\ h\nu\cos\theta\ \text{normal},\ h\nu\sin\theta\ \text{tangential}}
$$
The mirror also gains energy $h\nu$, which goes into heat, i.e. an increase of its rest mass (Problem 27). The normal push is **half** that of reflection. That's why a perfectly reflecting solar sail gets twice the normal force of a perfectly absorbing one, though the absorber also gets a sideways push.

```python
import sympy as sp
h, nu, th = sp.symbols("h nu theta", positive=True)
p_in = h*nu*sp.Matrix([1, sp.cos(th), sp.sin(th), 0])
p_out = h*nu*sp.Matrix([1, -sp.cos(th), sp.sin(th), 0])
print("reflected: ΔP_mirror =", list(sp.simplify(p_in - p_out)))
print("absorbed:  ΔP_mirror =", list(p_in), "  |Δp| =", sp.simplify(sp.sqrt(sum(c**2 for c in p_in[1:]))))
```
