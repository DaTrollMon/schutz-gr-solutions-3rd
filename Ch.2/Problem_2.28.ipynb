# Chapter 2, Problem 28 — dot products are frame-independent

**Problem.** Let $\vec A \xrightarrow{\mathcal O}(5, 1, -1, 0)$, $\vec B \xrightarrow{\mathcal O}(-2, 3, 1, 6)$, $\vec C \xrightarrow{\mathcal O}(2, -2, 0, 0)$. Let $\bar{\mathcal O}$ move at $v = 0.6$ in the $+x$ direction relative to $\mathcal O$, axes parallel.
- **(a)** Find the components of $\vec A$, $\vec B$, $\vec C$ in $\bar{\mathcal O}$.
- **(b)** Form $\vec A\cdot\vec B$, $\vec B\cdot\vec C$, $\vec A\cdot\vec C$ and $\vec C\cdot\vec C$ using the components in $\bar{\mathcal O}$. Verify the frame independence of these numbers.
- **(c)** Classify $\vec A$, $\vec B$, $\vec C$ as timelike, spacelike or null.

## (a) Components in $\bar{\mathcal O}$

$\gamma = 1.25$, $\gamma v = 0.75$: $X^{\bar0} = 1.25X^0 - 0.75X^1$, $X^{\bar1} = -0.75X^0 + 1.25X^1$, $y$ and $z$ unchanged.

| | $\mathcal O$ | $\bar{\mathcal O}$ |
|---|---|---|
| $\vec A$ | $(5, 1, -1, 0)$ | $(6.25 - 0.75,\ -3.75 + 1.25,\ -1,\ 0) = \boxed{(5.5,\ -2.5,\ -1,\ 0)}$ |
| $\vec B$ | $(-2, 3, 1, 6)$ | $(-2.5 - 2.25,\ 1.5 + 3.75,\ 1,\ 6) = \boxed{(-4.75,\ 5.25,\ 1,\ 6)}$ |
| $\vec C$ | $(2, -2, 0, 0)$ | $(2.5 + 1.5,\ -1.5 - 2.5,\ 0,\ 0) = \boxed{(4,\ -4,\ 0,\ 0)}$ |

## (b) Dot products, $\vec X\cdot\vec Y = -X^0Y^0 + X^1Y^1 + X^2Y^2 + X^3Y^3$

**In $\bar{\mathcal O}$:**
- $\vec A\cdot\vec B = -(5.5)(-4.75) + (-2.5)(5.25) + (-1)(1) + 0 = 26.125 - 13.125 - 1 = \boxed{12}$
- $\vec B\cdot\vec C = -(-4.75)(4) + (5.25)(-4) + 0 + 0 = 19 - 21 = \boxed{-2}$
- $\vec A\cdot\vec C = -(5.5)(4) + (-2.5)(-4) + 0 + 0 = -22 + 10 = \boxed{-12}$
- $\vec C\cdot\vec C = -16 + 16 = \boxed{0}$

**In $\mathcal O$, for comparison:**
- $\vec A\cdot\vec B = -(5)(-2) + (1)(3) + (-1)(1) + 0 = 10 + 3 - 1 = 12$ ✓
- $\vec B\cdot\vec C = -(-2)(2) + (3)(-2) = 4 - 6 = -2$ ✓
- $\vec A\cdot\vec C = -(5)(2) + (1)(-2) = -12$ ✓
- $\vec C\cdot\vec C = -4 + 4 = 0$ ✓

Every product agrees, even though every individual component changed. The dot product is a Lorentz scalar (§2.5).

## (c) Classification by $\vec X\cdot\vec X$

- $\vec A\cdot\vec A = -25 + 1 + 1 + 0 = -23 < 0$ → **timelike** (also in $\bar{\mathcal O}$: $-30.25 + 6.25 + 1 = -23$).
- $\vec B\cdot\vec B = -4 + 9 + 1 + 36 = 42 > 0$ → **spacelike** ($-22.5625 + 27.5625 + 1 + 36 = 42$).
- $\vec C\cdot\vec C = 0$, $\vec C\neq0$ → **null**. Note $\vec C$ is null but has nonzero components: a null vector is not the zero vector.

```python
import numpy as np
from schutz_gr import boost_matrix, minkowski_dot
vecs = {"A": np.array([5, 1, -1, 0.]), "B": np.array([-2, 3, 1, 6.]), "C": np.array([2, -2, 0, 0.])}
L = boost_matrix(0.6)
bar = {k: L @ X for k, X in vecs.items()}
for k in vecs: print(f"(a) {k}̄ =", bar[k])
for p, q in (("A", "B"), ("B", "C"), ("A", "C"), ("C", "C")):
    print(f"(b) {p}·{q}:  in O {minkowski_dot(vecs[p], vecs[q]):+.4f}   in Ō {minkowski_dot(bar[p], bar[q]):+.4f}")
kind = lambda s: "null" if np.isclose(s, 0) else ("spacelike" if s > 0 else "timelike")
for k, X in vecs.items(): print(f"(c) {k}·{k} = {minkowski_dot(X, X):+.1f} → {kind(minkowski_dot(X, X))}")
```
