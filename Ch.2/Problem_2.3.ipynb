# Chapter 2, Problem 3 — prove Eq. (2.5)

**Problem.** Prove Eq. (2.5):
$$
\Lambda^{\bar\alpha}{}_\beta\,x^\beta = \Lambda^{\bar\alpha}{}_0\,x^0 + \Lambda^{\bar\alpha}{}_i\,x^i . \tag{2.5}
$$

**What it says.** A Greek (spacetime) dummy index runs over $0,1,2,3$; a Latin (spatial) one over $1,2,3$ only. So a Greek sum equals its time term plus a Latin sum. It's also why a Greek dummy index can't be renamed to a Latin one.

## Proof

Fix any value of the free index $\bar\alpha$. By the summation convention the left side means
$$
\Lambda^{\bar\alpha}{}_\beta x^\beta \equiv \sum_{\beta=0}^{3}\Lambda^{\bar\alpha}{}_\beta x^\beta
= \Lambda^{\bar\alpha}{}_0x^0 + \Lambda^{\bar\alpha}{}_1x^1 + \Lambda^{\bar\alpha}{}_2x^2 + \Lambda^{\bar\alpha}{}_3x^3 .
$$
Group the last three terms. By the convention for Latin indices,
$$
\Lambda^{\bar\alpha}{}_ix^i \equiv \sum_{i=1}^{3}\Lambda^{\bar\alpha}{}_ix^i = \Lambda^{\bar\alpha}{}_1x^1 + \Lambda^{\bar\alpha}{}_2x^2 + \Lambda^{\bar\alpha}{}_3x^3 .
$$
Substituting:
$$
\Lambda^{\bar\alpha}{}_\beta x^\beta = \Lambda^{\bar\alpha}{}_0x^0 + \Lambda^{\bar\alpha}{}_ix^i . \qquad\blacksquare
$$
The same steps work for each of the four values of $\bar\alpha$, so all four equations of (2.5) hold. Nothing here uses any property of $\Lambda$ — it's purely about what the index conventions mean.

**Example (boost along $x$).** For $\bar\alpha = \bar0$: $\Lambda^{\bar0}{}_0 = \gamma$, $\Lambda^{\bar0}{}_1 = -v\gamma$, $\Lambda^{\bar0}{}_2 = \Lambda^{\bar0}{}_3 = 0$, so
$$
x^{\bar0} = \gamma\,x^0 + (-v\gamma\,x^1 + 0\cdot x^2 + 0\cdot x^3) = \gamma(t - vx),
$$
which is Eq. (1.12) ✓.

```python
import sympy as sp
L = sp.Matrix(4, 4, lambda a, b: sp.Symbol(f"Lambda^{a}_{b}"))
x = sp.symbols("x0:4")
for a in range(4):
    lhs = sum(L[a, b] * x[b] for b in range(4))          # Λ^ᾱ_β x^β
    rhs = L[a, 0] * x[0] + sum(L[a, i] * x[i] for i in (1, 2, 3))   # Λ^ᾱ_0 x^0 + Λ^ᾱ_i x^i
    print(f"ᾱ = {a}:", sp.expand(lhs - rhs) == 0)
```
