# Chapter 2, Problem 1 — practising the summation convention

**Problem (paraphrased).** Given the numbers
$$
\{A^0 = 5,\ A^1 = 0,\ A^2 = -1,\ A^3 = -6\},\qquad \{B_0 = 0,\ B_1 = -2,\ B_2 = 4,\ B_3 = 0\},
$$
$$
\{C_{00} = 1, C_{01} = 0, C_{02} = 2, C_{03} = 3,\ C_{10} = 5, C_{11} = -2, C_{12} = -2, C_{13} = 0,\ C_{20} = 4, C_{21} = 5, C_{22} = 2, C_{23} = -2,\ C_{30} = -1, C_{31} = -1, C_{32} = -3, C_{33} = 0\},
$$
find: **(a)** $A^\alpha B_\alpha$; **(b)** $A^\alpha C_{\alpha\beta}$ for all $\beta$; **(c)** $A^\gamma C_{\gamma\sigma}$ for all $\sigma$; **(d)** $A^\nu C_{\mu\nu}$ for all $\mu$; **(e)** $A^\alpha B_\beta$ for all $\alpha, \beta$; **(f)** $A^i B_i$; **(g)** $A^j B_k$ for all $j, k$.

> **Index heights.** The text I have of the book loses which indices are raised. I've placed them so every repeated index is one up, one down, which is what the summation convention (§2.1) needs: $A$ up, $B$ and $C$ down. The arithmetic is identical whichever way they're printed; only the notation changes.

**Rules used (§2.1).**
- An index repeated once up and once down is summed over $0,1,2,3$ (a **dummy** index).
- An index appearing once is **free**: the expression stands for one number per value of the free index.
- Latin indices ($i, j, k$) run over $1,2,3$ only.

It helps to arrange $C_{\alpha\beta}$ as a matrix, **first index = row, second = column**:
$$
C = \begin{pmatrix}
1 & 0 & 2 & 3\\
5 & -2 & -2 & 0\\
4 & 5 & 2 & -2\\
-1 & -1 & -3 & 0
\end{pmatrix}.
$$

```python
import numpy as np
A = np.array([5, 0, -1, -6])
B = np.array([0, -2, 4, 0])
C = np.array([[ 1,  0,  2,  3],
              [ 5, -2, -2,  0],
              [ 4,  5,  2, -2],
              [-1, -1, -3,  0]])
```

## (a) $A^\alpha B_\alpha$ — one dummy index, no free index ⇒ a single number

$$
A^\alpha B_\alpha = A^0B_0 + A^1B_1 + A^2B_2 + A^3B_3 = (5)(0) + (0)(-2) + (-1)(4) + (-6)(0) = \boxed{-4}
$$

```python
print("(a)", np.einsum("a,a->", A, B), "  term by term:", [int(A[a]*B[a]) for a in range(4)])
```

## (b) $A^\alpha C_{\alpha\beta}$ for all $\beta$ — dummy $\alpha$, free $\beta$ ⇒ four numbers

The sum runs over the **first** index of $C$ (rows), so for each column $\beta$:
$$
A^\alpha C_{\alpha\beta} = 5\,C_{0\beta} + 0\,C_{1\beta} - 1\,C_{2\beta} - 6\,C_{3\beta}.
$$
- $\beta = 0$: $5(1) + 0(5) - 1(4) - 6(-1) = 5 - 4 + 6 = 7$
- $\beta = 1$: $5(0) + 0(-2) - 1(5) - 6(-1) = -5 + 6 = 1$
- $\beta = 2$: $5(2) + 0(-2) - 1(2) - 6(-3) = 10 - 2 + 18 = 26$
- $\beta = 3$: $5(3) + 0(0) - 1(-2) - 6(0) = 15 + 2 = 17$

$$
\boxed{A^\alpha C_{\alpha\beta} = (7,\ 1,\ 26,\ 17)\ \text{ for } \beta = 0,1,2,3}
$$
In matrix language: the row vector $A^T C$.

```python
print("(b) A^α C_αβ =", np.einsum("a,ab->b", A, C), "  (= A @ C)", A @ C)
```

## (c) $A^\gamma C_{\gamma\sigma}$ for all $\sigma$

This is (b) with the dummy index renamed $\alpha\to\gamma$ and the free index renamed $\beta\to\sigma$. Neither renaming changes anything (§2.1), so
$$
\boxed{A^\gamma C_{\gamma\sigma} = (7,\ 1,\ 26,\ 17)\ \text{ for } \sigma = 0,1,2,3}
$$

```python
print("(c) A^γ C_γσ =", np.einsum("g,gs->s", A, C))
```

## (d) $A^\nu C_{\mu\nu}$ for all $\mu$ — now the sum is over the **second** index of $C$

For each row $\mu$:
$$
A^\nu C_{\mu\nu} = 5\,C_{\mu0} + 0\,C_{\mu1} - 1\,C_{\mu2} - 6\,C_{\mu3}.
$$
- $\mu = 0$: $5(1) + 0 - 1(2) - 6(3) = 5 - 2 - 18 = -15$
- $\mu = 1$: $5(5) + 0 - 1(-2) - 6(0) = 25 + 2 = 27$
- $\mu = 2$: $5(4) + 0 - 1(2) - 6(-2) = 20 - 2 + 12 = 30$
- $\mu = 3$: $5(-1) + 0 - 1(-3) - 6(0) = -5 + 3 = -2$

$$
\boxed{A^\nu C_{\mu\nu} = (-15,\ 27,\ 30,\ -2)\ \text{ for } \mu = 0,1,2,3}
$$
Compare (b): contracting on a *different slot* of $C$ gives a different answer, because $C$ is not symmetric ($C_{\mu\nu}\neq C_{\nu\mu}$). In matrix language, (d) is $CA$ while (b) is $A^TC$.

```python
print("(d) A^ν C_μν =", np.einsum("n,mn->m", A, C), "  (= C @ A)", C @ A)
print("C symmetric?", np.array_equal(C, C.T))
```

## (e) $A^\alpha B_\beta$ for all $\alpha, \beta$ — no repeated index ⇒ no sum, 16 numbers

Each entry is just a product. Row $\alpha$, column $\beta$:
$$
\boxed{A^\alpha B_\beta = \begin{pmatrix}
0 & -10 & 20 & 0\\
0 & 0 & 0 & 0\\
0 & 2 & -4 & 0\\
0 & 12 & -24 & 0
\end{pmatrix}}
$$
(The outer product. Its trace, $\sum_\alpha A^\alpha B_\alpha = 0 + 0 - 4 + 0 = -4$, reproduces (a).)

```python
E = np.einsum("a,b->ab", A, B)
print("(e)\n", E, "\n trace =", np.trace(E))
```

## (f) $A^i B_i$ — Latin dummy index: sum over $1,2,3$ only

$$
A^iB_i = A^1B_1 + A^2B_2 + A^3B_3 = (0)(-2) + (-1)(4) + (-6)(0) = \boxed{-4}
$$
The same number as (a) only because $B_0 = 0$ here, so the $\alpha = 0$ term vanished anyway. In general $A^\alpha B_\alpha = A^0B_0 + A^iB_i$ (Eq. 2.5 / Problem 3).

## (g) $A^j B_k$ for all $j, k$ — two free Latin indices ⇒ $3\times3 = 9$ numbers

The lower-right $3\times3$ block of (e) (rows/columns 1–3):
$$
\boxed{A^jB_k = \begin{pmatrix} 0 & 0 & 0\\ 2 & -4 & 0\\ 12 & -24 & 0 \end{pmatrix}}\qquad (j = \text{row},\ k = \text{column},\ \text{both } 1..3)
$$

```python
print("(f) A^i B_i =", np.einsum("i,i->", A[1:], B[1:]))
print("(g) A^j B_k =\n", np.einsum("j,k->jk", A[1:], B[1:]))
```

## Summary

| | expression | free indices | result |
|---|---|---|---|
| (a) | $A^\alpha B_\alpha$ | none | $-4$ |
| (b) | $A^\alpha C_{\alpha\beta}$ | $\beta$ | $(7, 1, 26, 17)$ |
| (c) | $A^\gamma C_{\gamma\sigma}$ | $\sigma$ | $(7, 1, 26, 17)$ |
| (d) | $A^\nu C_{\mu\nu}$ | $\mu$ | $(-15, 27, 30, -2)$ |
| (e) | $A^\alpha B_\beta$ | $\alpha,\beta$ | 16 numbers (above) |
| (f) | $A^iB_i$ | none | $-4$ |
| (g) | $A^jB_k$ | $j,k$ | 9 numbers (above) |

The number of free indices fixes how many numbers an expression stands for: $4^{(\#\text{Greek free})}\times3^{(\#\text{Latin free})}$.
