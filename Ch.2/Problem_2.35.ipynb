# Chapter 2, Problem 35 — the boosted basis is orthonormal

**Problem.** Show that the vectors $\{\vec e_{\bar\beta}\}$ obtained from $\{\vec e_\alpha\}$ by Eq. (2.15) satisfy $\vec e_{\bar\alpha}\cdot\vec e_{\bar\beta} = \eta_{\bar\alpha\bar\beta}$ for all $\bar\alpha$, $\bar\beta$.

**Eq. (2.15):** $\vec e_{\bar\mu} = \Lambda^\nu{}_{\bar\mu}(-v)\,\vec e_\nu$. **Eq. (2.27):** $\vec e_\mu\cdot\vec e_\nu = \eta_{\mu\nu}$, with $\eta = \mathrm{diag}(-1, 1, 1, 1)$.

## General argument (index form)

Write $L^\nu{}_{\bar\mu} \equiv \Lambda^\nu{}_{\bar\mu}(-v)$. Using bilinearity of the dot product (Problem 34) to pull the numbers out of both slots:
$$
\vec e_{\bar\alpha}\cdot\vec e_{\bar\beta} = \big(L^\mu{}_{\bar\alpha}\vec e_\mu\big)\cdot\big(L^\nu{}_{\bar\beta}\vec e_\nu\big) = L^\mu{}_{\bar\alpha}L^\nu{}_{\bar\beta}\,(\vec e_\mu\cdot\vec e_\nu) = L^\mu{}_{\bar\alpha}\,\eta_{\mu\nu}\,L^\nu{}_{\bar\beta}.
$$
As matrices this is $(L^T\eta L)_{\bar\alpha\bar\beta}$. So the claim is equivalent to
$$
L^T\eta L = \eta ,
$$
the defining property of a Lorentz transformation (it's the statement that $L$ preserves the interval, Problem 1.20). Let's verify it explicitly for the boost.

## Explicit check for a boost along $x$

$L = \Lambda(-v)$ has columns giving $\bar{\mathcal O}$'s basis vectors in $\mathcal O$ (Problem 6):
$$
\vec e_{\bar0} \to (\gamma,\ \gamma v,\ 0,\ 0),\quad \vec e_{\bar1} \to (\gamma v,\ \gamma,\ 0,\ 0),\quad \vec e_{\bar2} \to (0, 0, 1, 0),\quad \vec e_{\bar3} \to (0, 0, 0, 1).
$$
Now all ten independent products (the dot product is symmetric):
- $\vec e_{\bar0}\cdot\vec e_{\bar0} = -\gamma^2 + \gamma^2v^2 = -\gamma^2(1 - v^2) = -1 = \eta_{\bar0\bar0}$ ✓
- $\vec e_{\bar1}\cdot\vec e_{\bar1} = -\gamma^2v^2 + \gamma^2 = 1 = \eta_{\bar1\bar1}$ ✓
- $\vec e_{\bar0}\cdot\vec e_{\bar1} = -\gamma\cdot\gamma v + \gamma v\cdot\gamma = 0 = \eta_{\bar0\bar1}$ ✓
- $\vec e_{\bar2}\cdot\vec e_{\bar2} = \vec e_{\bar3}\cdot\vec e_{\bar3} = 1$ ✓
- $\vec e_{\bar0}\cdot\vec e_{\bar2} = \vec e_{\bar0}\cdot\vec e_{\bar3} = \vec e_{\bar1}\cdot\vec e_{\bar2} = \vec e_{\bar1}\cdot\vec e_{\bar3} = \vec e_{\bar2}\cdot\vec e_{\bar3} = 0$ ✓ (no overlapping nonzero components)

So $\vec e_{\bar\alpha}\cdot\vec e_{\bar\beta} = \eta_{\bar\alpha\bar\beta}$: $\bar{\mathcal O}$'s basis is an orthonormal tetrad too. $\blacksquare$

**Meaning.** Every inertial observer's basis vectors are orthonormal in the Minkowski sense, even though on $\mathcal O$'s diagram $\vec e_{\bar0}$ and $\vec e_{\bar1}$ don't look perpendicular (Fig. 2.4). They make equal angles with the light ray, which is what Minkowski orthogonality looks like. This is what makes all inertial frames equivalent, and it's why $\eta$ has the same components in every inertial frame.

```python
import sympy as sp
from schutz_gr import boost_matrix, boost_general, minkowski_dot
v = sp.symbols("v", real=True)
L = boost_matrix(-v)                          # columns = e_μ̄ in O
eta = sp.diag(-1, 1, 1, 1)
print("Lᵀ η L = η :", sp.simplify(L.T * eta * L) == eta)
G = sp.Matrix(4, 4, lambda a, b: sp.simplify(minkowski_dot(L[:, a], L[:, b])))
print("e_ᾱ·e_β̄ ="); display(G)
# also for a boost in a general direction
vx, vy, vz = sp.Rational(1, 5), sp.Rational(-1, 3), sp.Rational(1, 4)
Lg = boost_general([-vx, -vy, -vz])
print("general-direction boost: Lᵀ η L = η :", sp.simplify(Lg.T * eta * Lg) == eta)
```
