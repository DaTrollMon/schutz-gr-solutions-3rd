# Chapter 2, Problem 30 — a cosmic ray seen from a rocket, two ways

**Problem.** The four-velocity of a rocket ship is $\vec U \xrightarrow{\mathcal O}(2, 1, 1, 1)$. It encounters a high-velocity cosmic ray whose momentum is $\vec P \xrightarrow{\mathcal O}(300, 299, 0, 0)\times10^{-27}$ kg. Compute the energy of the cosmic ray as measured by the rocket's passengers using each method:
- **(a)** Find the Lorentz transformation from $\mathcal O$ to the MCRF of the rocket, and use it to transform the components of $\vec P$.
- **(b)** Use Eq. (2.35).
- **(c)** Which method is quicker? Why?

(I'll write $P$ in units of $10^{-27}$ kg.)

## (a) Via the Lorentz transformation

**Rocket velocity** (Problem 15c): $\mathbf v = \mathbf U/U^0 = (\tfrac12, \tfrac12, \tfrac12)$, $|\mathbf v|^2 = \tfrac34$, $\gamma = 2 = U^0$ ✓.

**The boost to the rocket frame** (velocity not along an axis). The general boost with velocity $\mathbf v$ and axes kept parallel:
$$
\Lambda^{\bar0}{}_0 = \gamma,\qquad \Lambda^{\bar0}{}_i = \Lambda^{\bar i}{}_0 = -\gamma v_i,\qquad \Lambda^{\bar i}{}_j = \delta_{ij} + (\gamma - 1)\frac{v_iv_j}{|\mathbf v|^2}.
$$
(Check: for $\mathbf v = v\hat{\mathbf x}$ it reduces to Eq. 1.12.) With $\gamma = 2$, $v_i = \tfrac12$, $(\gamma - 1)v_iv_j/|\mathbf v|^2 = \tfrac14/\tfrac34 = \tfrac13$:
$$
\Lambda = \begin{pmatrix}2 & -1 & -1 & -1\\ -1 & \tfrac43 & \tfrac13 & \tfrac13\\ -1 & \tfrac13 & \tfrac43 & \tfrac13\\ -1 & \tfrac13 & \tfrac13 & \tfrac43\end{pmatrix}.
$$
(Check: $\Lambda\vec U = (4 - 3,\ -2 + \tfrac43 + \tfrac23,\ \dots) = (1, 0, 0, 0)$ ✓ — the rocket is at rest in this frame.)

**Transform $\vec P$.** Only the energy is asked for:
$$
P^{\bar0} = 2(300) - 1(299) - 1(0) - 1(0) = 600 - 299 = 301 .
$$
$$
\boxed{\bar E = 301\times10^{-27}\ \text{kg}}\quad(= 2.71\times10^{-8}\ \text{J} \approx 1.7\times10^{11}\ \text{eV}).
$$
(The full components are $\vec P \xrightarrow{\text{rocket}} \big(301,\ \tfrac{296}{3},\ -\tfrac{601}{3},\ -\tfrac{601}{3}\big)\times10^{-27}$ kg, computed below; check $-301^2 + (296^2 + 2\cdot601^2)/9 = -599 = -300^2 + 299^2$ ✓.)

## (b) Via Eq. (2.35), $\bar E = -\vec P\cdot\vec U_{\text{obs}}$

One dot product in $\mathcal O$'s components:
$$
\vec P\cdot\vec U = -(300)(2) + (299)(1) + (0)(1) + (0)(1) = -600 + 299 = -301,
$$
$$
\boxed{\bar E = -\vec P\cdot\vec U = 301\times10^{-27}\ \text{kg}}\qquad\checkmark
$$

## (c) Which is quicker, and why

**(b), by far.** It needs four products and a sum. Method (a) needs the full Lorentz matrix for a boost in a general direction (16 entries) before you can apply it.

**Why it works.** $\vec P\cdot\vec U$ is a scalar, the same in every frame. In the rocket's frame $\vec U = \vec e_{\bar0} \to (1, 0, 0, 0)$, so there $\vec P\cdot\vec U = -P^{\bar0} = -\bar E$. Since the number doesn't depend on the frame, compute it in whichever frame you already have components in. Writing physical quantities as frame-invariant expressions like this is almost always the efficient route (as §2.6 advises).

Interesting aside: the rocket sees almost the same energy as $\mathcal O$ (301 vs 300) because it moves partly *away* from the cosmic ray's direction ($+x$). The blueshift from its $\gamma = 2$ is nearly cancelled by the redshift from receding.

```python
import sympy as sp
from schutz_gr import boost_general, minkowski_dot
R = sp.Rational
U = sp.Matrix([2, 1, 1, 1]); P = sp.Matrix([300, 299, 0, 0])
v = [U[i]/U[0] for i in (1, 2, 3)]
L = boost_general(v)
print("(a) Λ ="); display(L)
print("    Λ U =", list(L*U), "   Λ P =", list(L*P), "→ Ē =", (L*P)[0])
print("(b) −P·U =", -minkowski_dot(P, U))
print("    in SI: Ē =", float((L*P)[0])*1e-27*9e16, "J =", float((L*P)[0])*1e-27*9e16/1.602e-19, "eV")
```
