# Chapter 2, Problem 2 — free and dummy indices

**Problem (paraphrased).** Identify the free and dummy indices in the following equations and change them into equivalent expressions with different indices. How many different equations does each represent?

- **(a)** $A^\alpha B_\alpha = 5$
- **(b)** $A^{\bar\mu} = \Lambda^{\bar\mu}{}_\nu\,A^\nu$
- **(c)** $T^{\alpha\mu\lambda}A_\mu C_\lambda{}^\gamma = D^{\gamma\alpha}$
- **(d)** $R_{\mu\nu} - \tfrac12 g_{\mu\nu}R = G_{\mu\nu}$

> **Index heights.** The text I have of the book loses which indices are raised; the placements above follow the convention that a summed index appears once up and once down. Which indices are free or dummy, and how many equations there are, doesn't depend on heights.

**Rules (§2.1).**
1. A **dummy** index appears twice (once up, once down) on the *same side* of a term and is summed. It can be renamed to any unused Greek letter (to a Latin one only if it's meant to run over $1..3$).
2. A **free** index appears once in *every* term of the equation, at the same height. The equation must hold for every value of it, so it stands for several equations. It can be renamed, but **everywhere at once**.
3. An equation with $k$ free Greek indices represents $4^k$ equations (some may be redundant if the objects have symmetries).

## (a) $A^\alpha B_\alpha = 5$

- **Dummy:** $\alpha$ (summed). **Free:** none.
- **Equivalent forms:** $A^\beta B_\beta = 5$, $A^\mu B_\mu = 5$, …
- **Number of equations:** $4^0 = \boxed{1}$. It's the single statement $A^0B_0 + A^1B_1 + A^2B_2 + A^3B_3 = 5$.

## (b) $A^{\bar\mu} = \Lambda^{\bar\mu}{}_\nu A^\nu$

- **Dummy:** $\nu$. **Free:** $\bar\mu$ (barred, i.e. it labels $\bar{\mathcal O}$'s components; it appears once on each side).
- **Equivalent forms:** $A^{\bar\alpha} = \Lambda^{\bar\alpha}{}_\beta A^\beta$ (both renamed), or $A^{\bar\mu} = \Lambda^{\bar\mu}{}_\sigma A^\sigma$ (dummy only). A bar must stay a bar: $A^{\bar\mu} = \Lambda^{\bar\mu}{}_{\bar\nu}A^{\bar\nu}$ is **not** equivalent. Barred and unbarred indices refer to different frames (§2.2).
- **Number of equations:** $\boxed{4}$, one for each $\bar\mu$. This is the Lorentz transformation of the components, Eq. (2.7).

## (c) $T^{\alpha\mu\lambda}A_\mu C_\lambda{}^\gamma = D^{\gamma\alpha}$

- **Dummies:** $\mu$ (on $T$ and $A$) and $\lambda$ (on $T$ and $C$), so this is a double sum, 16 terms on the left.
- **Free:** $\alpha$ and $\gamma$, each appearing once on the left and once on the right.
- **Equivalent forms:** rename dummies, e.g. $T^{\alpha\beta\delta}A_\beta C_\delta{}^\gamma = D^{\gamma\alpha}$; or rename free indices consistently, e.g. $\alpha\to\sigma$, $\gamma\to\tau$: $T^{\sigma\mu\lambda}A_\mu C_\lambda{}^\tau = D^{\tau\sigma}$.
- **Not equivalent:** swapping which free index goes in which slot, e.g. $\dots = D^{\alpha\gamma}$, unless $D$ happens to be symmetric.
- **Number of equations:** $4^2 = \boxed{16}$ (one per pair $(\alpha,\gamma)$).

## (d) $R_{\mu\nu} - \tfrac12 g_{\mu\nu}R = G_{\mu\nu}$

- **Dummy:** none. ($R$ without indices is a scalar; it's a single symbol, not a sum here.)
- **Free:** $\mu$ and $\nu$, in every term.
- **Equivalent form:** $R_{\alpha\beta} - \tfrac12 g_{\alpha\beta}R = G_{\alpha\beta}$.
- **Number of equations:** $4^2 = \boxed{16}$ as written. (Looking ahead: $R_{\mu\nu}$, $g_{\mu\nu}$ and $G_{\mu\nu}$ are all symmetric, so only **10** are independent. These are Einstein's field equations, Chapter 8.)

```python
# A quick numerical demonstration that renaming dummy indices changes nothing, but swapping free-index slots does.
import numpy as np
rng = np.random.default_rng(2)
T = rng.normal(size=(4, 4, 4)); A = rng.normal(size=4); C = rng.normal(size=(4, 4))
D1 = np.einsum("aml,m,lg->ga", T, A, C)     # T^{αμλ} A_μ C_λ^γ  → indexed [γ, α]
D2 = np.einsum("abd,b,dg->ga", T, A, C)     # dummies renamed μ→β, λ→δ
print("renaming dummies changes nothing:", np.allclose(D1, D2))
print("swapping free slots (D^{γα} vs D^{αγ}) matters:", not np.allclose(D1, D1.T))
print("number of equations in (c):", D1.size)
```

## Summary

| | dummy | free | # equations |
|---|---|---|---|
| (a) | $\alpha$ | — | 1 |
| (b) | $\nu$ | $\bar\mu$ | 4 |
| (c) | $\mu, \lambda$ | $\alpha, \gamma$ | 16 |
| (d) | — | $\mu, \nu$ | 16 (10 independent, by symmetry) |
