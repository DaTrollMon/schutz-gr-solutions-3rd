# Chapter 2, Problem 8 — the zero vector, and equality of vectors

**Problem.**
- **(a)** Prove that the zero vector $(0, 0, 0, 0)$ has these same components in all reference frames.
- **(b)** Use (a) to prove that if two vectors have equal components in one frame, they have equal components in all frames.

## (a) The zero vector is zero everywhere

Let $\vec 0 \xrightarrow{\mathcal O}(0,0,0,0)$, i.e. $0^\beta = 0$ for every $\beta$. In any other frame $\bar{\mathcal O}$, Eq. (2.7) gives
$$
0^{\bar\alpha} = \Lambda^{\bar\alpha}{}_\beta\,0^\beta = \sum_{\beta}\Lambda^{\bar\alpha}{}_\beta\cdot 0 = 0\qquad\text{for every }\bar\alpha .
$$
This holds whatever the 16 numbers $\Lambda^{\bar\alpha}{}_\beta$ are. Only **linearity** (and homogeneity) of the transformation is used. So $\vec 0 \xrightarrow{\bar{\mathcal O}}(0,0,0,0)$ in every frame. $\blacksquare$

(Contrast: a *null* vector, $\vec A\cdot\vec A = 0$, is not a zero vector, and its components certainly change between frames — §2.5.)

## (b) Equal in one frame ⇒ equal in all

Suppose $A^\beta = B^\beta$ for all $\beta$ in $\mathcal O$. Consider $\vec D = \vec A - \vec B$. By Eq. (2.8) its components in $\mathcal O$ are $D^\beta = A^\beta - B^\beta = 0$, so $\vec D$ is the zero vector.

By (a), $D^{\bar\alpha} = 0$ in every frame $\bar{\mathcal O}$. And by linearity of Eq. (2.7),
$$
D^{\bar\alpha} = \Lambda^{\bar\alpha}{}_\beta(A^\beta - B^\beta) = \Lambda^{\bar\alpha}{}_\beta A^\beta - \Lambda^{\bar\alpha}{}_\beta B^\beta = A^{\bar\alpha} - B^{\bar\alpha}.
$$
So $A^{\bar\alpha} - B^{\bar\alpha} = 0$, i.e. $A^{\bar\alpha} = B^{\bar\alpha}$ for all $\bar\alpha$, in every frame. $\blacksquare$

**Why it matters.** It makes a vector equation like $\vec A = \vec B$ meaningful without naming a frame: check it in any one frame and it holds in all. That's what lets us prove vector identities in the most convenient frame (often the rest frame / MCRF), a technique used over and over (e.g. Problems 17, 18).

```python
import sympy as sp
from schutz_gr import boost_matrix
v = sp.symbols("v", real=True)
L = boost_matrix(v)
print("(a) Λ(v)·0 =", list(L * sp.zeros(4, 1)))
A = sp.Matrix(sp.symbols("A0:4")); B = A.copy()          # equal components in O
print("(b) Λ(v)A − Λ(v)B =", list(sp.simplify(L * A - L * B)))
```
