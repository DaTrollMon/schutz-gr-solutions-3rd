# Chapter 2, Problem 15 — four-velocity components

**Problem.**
- **(a)** Compute the four-velocity components in $\mathcal O$ of a particle whose speed in $\mathcal O$ is $v$ in the $+x$ direction, by using the Lorentz transformation from the rest frame of the particle.
- **(b)** Generalize to arbitrary velocity $\mathbf v$, $|\mathbf v| < 1$.
- **(c)** Use (b) to express $\mathbf v$ in terms of the components $\{U^\alpha\}$.
- **(d)** Find the three-velocity $\mathbf v$ of a particle whose four-velocity components are $(2, 1, 1, 1)$.

## (a) Motion along $x$

In its rest frame $\bar{\mathcal O}$ the particle's four-velocity is the time basis vector (§2.3): $\vec U = \vec e_{\bar0} \xrightarrow{\bar{\mathcal O}}(1, 0, 0, 0)$.

Transform to $\mathcal O$. $\bar{\mathcal O}$ moves at $+v$ relative to $\mathcal O$, so $\mathcal O$'s components come from the inverse matrix $\Lambda^\alpha{}_{\bar\beta}(-v)$ (Problem 11):
$$
U^\alpha = \Lambda^\alpha{}_{\bar\beta}(-v)\,U^{\bar\beta} = \Lambda^\alpha{}_{\bar0}(-v) \quad(\text{first column}).
$$
$$
\boxed{U^0 = \gamma = \frac{1}{\sqrt{1-v^2}},\quad U^1 = \gamma v = \frac{v}{\sqrt{1-v^2}},\quad U^2 = U^3 = 0}
$$
This is Eq. (2.21).

## (b) Arbitrary $\mathbf v$

**Rotation argument.** Rotate $\mathcal O$'s spatial axes so that $\mathbf v$ lies along the new $x'$ axis. Rotations don't touch time components or the $\gamma$ factor. By (a), in the rotated frame $\vec U \to \gamma(1, |\mathbf v|, 0, 0)$. The spatial part $\gamma|\mathbf v|\hat{\mathbf x}'$ is the 3-vector $\gamma\mathbf v$. Rotating back gives
$$
\boxed{\vec U \xrightarrow{\mathcal O}\big(\gamma,\ \gamma v^x,\ \gamma v^y,\ \gamma v^z\big),\qquad \gamma = \frac{1}{\sqrt{1 - |\mathbf v|^2}}}
$$

**Check from $U = dx/d\tau$ (Eq. 2.31).** Along the world line, $d\vec x \to (dt, \mathbf v\,dt)$ and $d\tau = \sqrt{dt^2 - |d\mathbf x|^2} = dt\sqrt{1 - |\mathbf v|^2} = dt/\gamma$. So $\vec U = d\vec x/d\tau = \gamma(1, \mathbf v)$ ✓. Normalization: $\vec U\cdot\vec U = \gamma^2(-1 + |\mathbf v|^2) = -1$ ✓ (Eq. 2.28).

## (c) $\mathbf v$ from $U^\alpha$

Divide the spatial components by the time component:
$$
\boxed{v^i = \frac{U^i}{U^0}}
$$
(Physically, $U^i/U^0 = (dx^i/d\tau)/(dt/d\tau) = dx^i/dt$.)

## (d) $\vec U \to (2, 1, 1, 1)$

First check it's a valid four-velocity: $\vec U\cdot\vec U = -4 + 1 + 1 + 1 = -1$ ✓, and $U^0 > 0$ ✓ (future-pointing).
$$
\boxed{\mathbf v = \left(\tfrac12,\ \tfrac12,\ \tfrac12\right)},\qquad |\mathbf v| = \tfrac{\sqrt3}{2} \approx 0.866 .
$$
Consistent: $\gamma = 1/\sqrt{1 - 3/4} = 2 = U^0$ ✓.

```python
import sympy as sp
from schutz_gr import boost_matrix, four_velocity, minkowski_dot
v = sp.symbols("v", real=True)
U_a = boost_matrix(-v) * sp.Matrix([1, 0, 0, 0])
print("(a) U =", list(sp.simplify(U_a)))
vx, vy, vz = sp.symbols("v_x v_y v_z", real=True)
U_b = four_velocity([vx, vy, vz])
print("(b) U =", list(U_b), "  U·U =", sp.simplify(minkowski_dot(U_b, U_b)))
U = [2, 1, 1, 1]
print("(d) U·U =", minkowski_dot(U, U), "  v =", [sp.Rational(U[i], U[0]) for i in (1, 2, 3)])
```
