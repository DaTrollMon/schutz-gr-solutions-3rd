# Chapter 2, Problem 12 — transforming a vector twice, and its invariant magnitude

**Problem.** Given $\vec A \xrightarrow{\mathcal O}(0, -2, 3, 5)$, find:
- **(a)** the components of $\vec A$ in $\bar{\mathcal O}$, which moves at speed 0.8 relative to $\mathcal O$ in the $+x$ direction;
- **(b)** the components of $\vec A$ in $\bar{\bar{\mathcal O}}$, which moves at speed 0.6 relative to $\bar{\mathcal O}$ in the $+x$ direction;
- **(c)** the magnitude of $\vec A$ from its components in $\mathcal O$;
- **(d)** the magnitude of $\vec A$ from its components in $\bar{\mathcal O}$.

**Tools.** Boost along $x$ (Eq. 2.7 with the matrix of §2.2): $A^{\bar0} = \gamma(A^0 - vA^1)$, $A^{\bar1} = \gamma(A^1 - vA^0)$, $A^{\bar2} = A^2$, $A^{\bar3} = A^3$. "Magnitude" in Schutz means $\vec A^{\,2} = \vec A\cdot\vec A = -(A^0)^2 + (A^1)^2 + (A^2)^2 + (A^3)^2$ (Eq. 2.24).

## (a) $\mathcal O \to \bar{\mathcal O}$, $v = 0.8$

$\gamma = 1/\sqrt{1 - 0.64} = 1/0.6 = 5/3$.
$$
A^{\bar0} = \tfrac53\big(0 - 0.8\times(-2)\big) = \tfrac53(1.6) = \tfrac83,\qquad
A^{\bar1} = \tfrac53\big(-2 - 0.8\times0\big) = -\tfrac{10}{3},
$$
$$
\boxed{\vec A \xrightarrow{\bar{\mathcal O}}\left(\tfrac83,\ -\tfrac{10}{3},\ 3,\ 5\right) \approx (2.667,\ -3.333,\ 3,\ 5)}
$$

## (b) $\bar{\mathcal O} \to \bar{\bar{\mathcal O}}$, $v = 0.6$ relative to $\bar{\mathcal O}$

Apply the boost to the $\bar{\mathcal O}$ components. $\gamma = 1/\sqrt{1 - 0.36} = 5/4$:
$$
A^{\bar{\bar0}} = \tfrac54\left(\tfrac83 - 0.6\times\left(-\tfrac{10}3\right)\right) = \tfrac54\left(\tfrac83 + 2\right) = \tfrac54\cdot\tfrac{14}{3} = \tfrac{35}{6},
$$
$$
A^{\bar{\bar1}} = \tfrac54\left(-\tfrac{10}3 - 0.6\times\tfrac83\right) = \tfrac54\left(-\tfrac{10}3 - \tfrac85\right) = \tfrac54\cdot\left(-\tfrac{74}{15}\right) = -\tfrac{37}{6},
$$
$$
\boxed{\vec A \xrightarrow{\bar{\bar{\mathcal O}}}\left(\tfrac{35}{6},\ -\tfrac{37}{6},\ 3,\ 5\right) \approx (5.833,\ -6.167,\ 3,\ 5)}
$$

**Cross-check in one step.** $\bar{\bar{\mathcal O}}$ moves relative to $\mathcal O$ at $w = (0.8 + 0.6)/(1 + 0.48) = 1.4/1.48 = 35/37$. Then $\gamma_w = 37/12$ and $\gamma_w w = 35/12$, so $A^{\bar{\bar0}} = \tfrac{37}{12}(0) + \tfrac{35}{12}(2) = \tfrac{35}{6}$ ✓ and $A^{\bar{\bar1}} = \tfrac{37}{12}(-2) - 0 = -\tfrac{37}{6}$ ✓. Two collinear boosts compose into one, with velocities combined by Eq. (1.13).

## (c) Magnitude from $\mathcal O$'s components

$$
\vec A^{\,2} = -(0)^2 + (-2)^2 + 3^2 + 5^2 = 0 + 4 + 9 + 25 = \boxed{38}\qquad(\text{spacelike})
$$

## (d) Magnitude from $\bar{\mathcal O}$'s components

$$
\vec A^{\,2} = -\left(\tfrac83\right)^2 + \left(-\tfrac{10}3\right)^2 + 3^2 + 5^2 = \frac{-64 + 100}{9} + 34 = 4 + 34 = \boxed{38}
$$
The same, as Eq. (2.25) guarantees: the magnitude is a Lorentz scalar. It's also 38 from the $\bar{\bar{\mathcal O}}$ components: $(-1225 + 1369)/36 + 34 = 4 + 34$.

```python
import sympy as sp
from schutz_gr import boost_matrix, minkowski_dot
R = sp.Rational
A = sp.Matrix([0, -2, 3, 5])
Ab = boost_matrix(R(4, 5)) * A
Abb = boost_matrix(R(3, 5)) * Ab
print("(a) Ā  =", list(Ab))
print("(b) Ā̄  =", list(Abb), "  one-step check with w = 35/37:", list(boost_matrix(R(35, 37)) * A))
print("(c),(d) A·A in O, Ō, Ō̄ =", minkowski_dot(A, A), minkowski_dot(Ab, Ab), minkowski_dot(Abb, Abb))
```
