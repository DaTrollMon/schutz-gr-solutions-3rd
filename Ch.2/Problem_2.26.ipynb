# Chapter 2, Problem 26 — the energy cost of the speed of light

**Problem.** Calculate the energy required to accelerate a particle of rest mass $m \neq 0$ from speed $v$ to speed $v + \delta v$ ($\delta v \ll v$), to first order in $\delta v$. Show that it would take an infinite amount of energy to accelerate the particle to the speed of light.

## The energy increment

The particle's energy is $E(v) = m\gamma = m(1 - v^2)^{-1/2}$. The energy needed is the change in $E$ (whatever supplies it), and to first order:
$$
\delta E = \frac{dE}{dv}\,\delta v,\qquad \frac{dE}{dv} = m\cdot\left(-\tfrac12\right)(1 - v^2)^{-3/2}\cdot(-2v) = \frac{mv}{(1 - v^2)^{3/2}} = m\gamma^3v .
$$
$$
\boxed{\delta E = m\gamma^3 v\,\delta v = \frac{m v\,\delta v}{(1 - v^2)^{3/2}}}
$$
**Newtonian limit:** for $v \ll 1$, $\gamma\approx1$ and $\delta E \approx mv\,\delta v = \delta\big(\tfrac12mv^2\big)$ ✓.

**Near light speed**, $\gamma^3$ blows up: each further increment of speed costs more and more. At $v = 0.99$, $\gamma^3 \approx 356$; at $0.999$, $\gamma^3 \approx 11\,000$.

## Infinite energy to reach $v = 1$

Add up the increments from $v_0$ to $V$:
$$
\Delta E = \int_{v_0}^{V}\frac{mv\,dv}{(1 - v^2)^{3/2}} = \left[\frac{m}{\sqrt{1 - v^2}}\right]_{v_0}^{V} = m\gamma(V) - m\gamma(v_0).
$$
As $V \to 1$, $\gamma(V) = (1 - V^2)^{-1/2} \to \infty$, so $\Delta E \to \infty$. $\blacksquare$

No finite energy brings a massive particle to light speed. Equivalently (end of §2.7), $p^1/p^0 = \sqrt{1 - m^2/(p^0)^2} < 1$ for every finite energy $p^0$. Only massless particles move at $v = 1$, and they always do.

```python
import sympy as sp
m, v, V, v0 = sp.symbols("m v V v_0", positive=True)
E = m/sp.sqrt(1 - v**2)
print("dE/dv =", sp.simplify(sp.diff(E, v)))
dE = sp.integrate(sp.diff(E, v), (v, v0, V))
print("ΔE from v0 to V =", sp.simplify(dE))
eps = sp.symbols("epsilon", positive=True)
print("limit V → 1⁻ (V = 1 − ε, ε → 0⁺):", sp.limit(m/sp.sqrt(1 - (1 - eps)**2), eps, 0, "+"))
for vv in (0.5, 0.9, 0.99, 0.999):
    print(f"  v = {vv}: γ³ = {(1 - vv**2)**-1.5:10.1f}")
```
