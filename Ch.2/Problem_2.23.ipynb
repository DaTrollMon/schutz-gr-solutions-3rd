# Chapter 2, Problem 23 — energy to fourth order in velocity

**Problem.** A particle of rest mass $m$ has three-velocity $\mathbf v$. Find its energy correct to terms of order $|\mathbf v|^4$. At what speed $|\mathbf v|$ does the absolute value of the $O(|\mathbf v|^4)$ term equal $\tfrac12$ of the kinetic-energy term $\tfrac12 m|\mathbf v|^2$?

## Expansion

$E = p^0 = m\gamma = m(1 - v^2)^{-1/2}$ (§2.4). Binomial series with $\epsilon = -v^2$, $n = -\tfrac12$ (as in Problem 1.14):
$$
(1 - v^2)^{-1/2} = 1 + \tfrac12v^2 + \frac{(-\frac12)(-\frac32)}{2}v^4 + \frac{(-\frac12)(-\frac32)(-\frac52)}{6}(-v^2)^3 + \cdots = 1 + \tfrac12v^2 + \tfrac38v^4 + \tfrac{5}{16}v^6 + \cdots
$$
$$
\boxed{E = m + \tfrac12 m v^2 + \tfrac38 m v^4 + O(v^6)}
$$
Rest-mass energy, Newtonian kinetic energy, and the first relativistic correction (positive: the true kinetic energy exceeds $\tfrac12mv^2$). In SI: $E = mc^2 + \tfrac12mv^2 + \tfrac38mv^4/c^2 + \dots$

## When is the correction half the Newtonian kinetic energy?

$$
\tfrac38 m v^4 = \tfrac12\cdot\tfrac12 m v^2 \quad\Longrightarrow\quad \tfrac38 v^2 = \tfrac14 \quad\Longrightarrow\quad v^2 = \tfrac23 ,
$$
$$
\boxed{|\mathbf v| = \sqrt{2/3} \approx 0.816}
$$
At that speed the 4th-order truncation is itself a poor approximation: the exact kinetic energy $m(\gamma - 1) = m(\sqrt3 - 1) \approx 0.732m$, versus $\tfrac12mv^2 + \tfrac38mv^4 = \tfrac13m + \tfrac16m = 0.5m$. The neglected terms ($\tfrac5{16}v^6 + \dots$) are no longer small.

```python
import sympy as sp
v, m = sp.symbols("v m", positive=True)
E = m / sp.sqrt(1 - v**2)
print("E =", sp.series(E, v, 0, 8))
sol = sp.solve(sp.Eq(sp.Rational(3, 8)*m*v**4, sp.Rational(1, 2)*sp.Rational(1, 2)*m*v**2), v)
print("v =", sol, "≈", [float(s) for s in sol])
vs = sol[0]
print("at that speed: exact KE/m =", sp.nsimplify(sp.simplify((E - m).subs(v, vs) / m)), "≈", float((E - m).subs(v, vs) / m),
      "  4th-order KE/m =", float((m*vs**2/2 + 3*m*vs**4/8) / m))
```
