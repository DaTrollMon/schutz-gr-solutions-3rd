# Chapter 2, Problem 27 — heat has inertia

**Problem.** Two identical bodies of mass 10 kg are at rest at the same temperature. One of them is heated by the addition of 100 J of heat. Both are then subjected to the same force. Which accelerates faster, and by how much?

## Heating increases the rest mass

The rest mass is the energy in the rest frame (§2.4): $m = p^0$ in the MCRF, i.e. $E_{\text{rest}} = mc^2$. Adding heat $Q$ to a body at rest, without moving it, increases its rest-frame energy by $Q$, so its rest mass by
$$
\Delta m = \frac{Q}{c^2} = \frac{100\ \text{J}}{9\times10^{16}\ \text{m}^2\text{s}^{-2}} = 1.1\times10^{-15}\ \text{kg}.
$$
(In $c = 1$ units this is Problem 1.1(a): 100 J $= 1.1\times10^{-15}$ kg.) The heated body has $m_h = 10 + 1.1\times10^{-15}$ kg; the other has $m_c = 10$ kg.

## Same force ⇒ less acceleration for more mass

Starting from rest, in the MCRF the motion is Newtonian to first order (Problem 19a): $F = m\,a$, with $m$ the **rest mass**. The heated body's extra internal energy behaves exactly like extra mass. So
$$
\frac{a_c}{a_h} = \frac{m_h}{m_c} = 1 + \frac{\Delta m}{m_c} = 1 + \frac{1.1\times10^{-15}}{10} = 1 + 1.1\times10^{-16}.
$$
$$
\boxed{\text{The unheated body accelerates faster, by a fraction } \Delta m/m \approx 1.1\times10^{-16}.}
$$
For example, with $F = 10$ N: $a_c = 1\ \text{m s}^{-2}$ and $a_h$ is smaller by about $1.1\times10^{-16}\ \text{m s}^{-2}$. Far too small to measure directly, but the same principle (energy has inertia) is what makes nuclear binding energies show up as measurable mass defects.

```python
c = 3e8
Q, m = 100.0, 10.0
dm = Q / c**2
print(f"Δm = {dm:.3e} kg,  Δm/m = {dm/m:.3e}")
F = 10.0
ac, ah = F/m, F/(m + dm)
from decimal import Decimal, getcontext
getcontext().prec = 30
acD, ahD = Decimal(F)/Decimal(m), Decimal(F)/(Decimal(m) + Decimal(Q)/Decimal(c)**2)
print(f"with F = 10 N: a_cold = {acD} m/s²,  a_hot = {ahD:.25f} m/s²,  difference = {acD - ahD:.3e} m/s²")
```
