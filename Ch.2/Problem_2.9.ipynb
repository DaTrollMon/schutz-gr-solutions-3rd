# Chapter 2, Problem 9 — the order of a double sum doesn't matter

**Problem.** Prove, by writing out all the terms, that
$$
\sum_{\bar\alpha=0}^{3}\left(\sum_{\beta=0}^{3}\Lambda^{\bar\alpha}{}_\beta A^\beta\vec e_{\bar\alpha}\right)
= \sum_{\beta=0}^{3}\left(\sum_{\bar\alpha=0}^{3}\Lambda^{\bar\alpha}{}_\beta A^\beta\vec e_{\bar\alpha}\right).
$$
Since the order of summation doesn't matter, the Einstein convention can write simply $\Lambda^{\bar\alpha}{}_\beta A^\beta\vec e_{\bar\alpha}$.

## Writing out all 16 terms

Abbreviate $T_{\bar\alpha\beta} \equiv \Lambda^{\bar\alpha}{}_\beta A^\beta\vec e_{\bar\alpha}$ (one term, no sum; each is a number times a vector).

**Left side — inner sum over $\beta$ first, for each $\bar\alpha$ (row by row):**
$$
\begin{aligned}
\bar\alpha = \bar0:&\quad \Lambda^{\bar0}{}_0A^0\vec e_{\bar0} + \Lambda^{\bar0}{}_1A^1\vec e_{\bar0} + \Lambda^{\bar0}{}_2A^2\vec e_{\bar0} + \Lambda^{\bar0}{}_3A^3\vec e_{\bar0}\\
\bar\alpha = \bar1:&\quad \Lambda^{\bar1}{}_0A^0\vec e_{\bar1} + \Lambda^{\bar1}{}_1A^1\vec e_{\bar1} + \Lambda^{\bar1}{}_2A^2\vec e_{\bar1} + \Lambda^{\bar1}{}_3A^3\vec e_{\bar1}\\
\bar\alpha = \bar2:&\quad \Lambda^{\bar2}{}_0A^0\vec e_{\bar2} + \Lambda^{\bar2}{}_1A^1\vec e_{\bar2} + \Lambda^{\bar2}{}_2A^2\vec e_{\bar2} + \Lambda^{\bar2}{}_3A^3\vec e_{\bar2}\\
\bar\alpha = \bar3:&\quad \Lambda^{\bar3}{}_0A^0\vec e_{\bar3} + \Lambda^{\bar3}{}_1A^1\vec e_{\bar3} + \Lambda^{\bar3}{}_2A^2\vec e_{\bar3} + \Lambda^{\bar3}{}_3A^3\vec e_{\bar3}
\end{aligned}
$$

**Right side — inner sum over $\bar\alpha$ first, for each $\beta$ (column by column):**
$$
\begin{aligned}
\beta = 0:&\quad \Lambda^{\bar0}{}_0A^0\vec e_{\bar0} + \Lambda^{\bar1}{}_0A^0\vec e_{\bar1} + \Lambda^{\bar2}{}_0A^0\vec e_{\bar2} + \Lambda^{\bar3}{}_0A^0\vec e_{\bar3}\\
\beta = 1:&\quad \Lambda^{\bar0}{}_1A^1\vec e_{\bar0} + \Lambda^{\bar1}{}_1A^1\vec e_{\bar1} + \Lambda^{\bar2}{}_1A^1\vec e_{\bar2} + \Lambda^{\bar3}{}_1A^1\vec e_{\bar3}\\
\beta = 2:&\quad \Lambda^{\bar0}{}_2A^2\vec e_{\bar0} + \Lambda^{\bar1}{}_2A^2\vec e_{\bar1} + \Lambda^{\bar2}{}_2A^2\vec e_{\bar2} + \Lambda^{\bar3}{}_2A^2\vec e_{\bar3}\\
\beta = 3:&\quad \Lambda^{\bar0}{}_3A^3\vec e_{\bar0} + \Lambda^{\bar1}{}_3A^3\vec e_{\bar1} + \Lambda^{\bar2}{}_3A^3\vec e_{\bar2} + \Lambda^{\bar3}{}_3A^3\vec e_{\bar3}
\end{aligned}
$$

**Compare.** Both sides contain exactly the 16 terms $T_{\bar\alpha\beta}$, one for each pair $(\bar\alpha, \beta)$. The left lists them row by row, the right column by column. Term $T_{\bar1 2}$, for example, is the 3rd term of row $\bar1$ on the left and the 2nd term of row $\beta = 2$ on the right.

Vector addition is commutative and associative (it's component-wise addition of real numbers, Eq. 2.8). So a finite sum doesn't depend on the order or grouping of its terms, and the two sides are equal. $\blacksquare$

Because of this, $\Lambda^{\bar\alpha}{}_\beta A^\beta\vec e_{\bar\alpha}$ is unambiguous: it means "sum over both repeated indices", in any order.

```python
import sympy as sp
L = sp.Matrix(4, 4, lambda a, b: sp.Symbol(f"L^{a}_{b}"))
A = sp.symbols("A0:4")
e = sp.symbols("ebar0:4", commutative=True)          # stand-ins for the basis vectors e_ᾱ
left  = sum(sum(L[a, b] * A[b] * e[a] for b in range(4)) for a in range(4))
right = sum(sum(L[a, b] * A[b] * e[a] for a in range(4)) for b in range(4))
print("number of distinct terms:", len(sp.expand(left).args))
print("left − right =", sp.expand(left - right))
```
