# Chapter 2, Problem 14 — reading a Lorentz matrix

**Problem.** The following matrix gives a Lorentz transformation from $\mathcal O$ to $\bar{\mathcal O}$:
$$
\Lambda = \begin{pmatrix}1.25 & 0 & 0 & 0.75\\ 0 & 1 & 0 & 0\\ 0 & 0 & 1 & 0\\ 0.75 & 0 & 0 & 1.25\end{pmatrix}.
$$
- **(a)** What is the velocity (speed and direction) of $\bar{\mathcal O}$ relative to $\mathcal O$?
- **(b)** What is the inverse matrix?
- **(c)** Find the components in $\mathcal O$ of a vector $\vec A \xrightarrow{\bar{\mathcal O}}(1, 2, 0, 0)$.

## (a) Velocity

Only the $t$ and $z$ rows/columns are mixed, so this is a boost along the $z$ axis. Compare with the standard form for a boost of velocity $v$ along $z$:
$$
\Lambda^{\bar0}{}_0 = \gamma,\quad \Lambda^{\bar0}{}_3 = \Lambda^{\bar3}{}_0 = -\gamma v,\quad \Lambda^{\bar3}{}_3 = \gamma .
$$
So $\gamma = 1.25$ and $-\gamma v = 0.75$, giving $v = -0.75/1.25 = -0.6$. Check: $\gamma = 1/\sqrt{1 - 0.36} = 1.25$ ✓.
$$
\boxed{\bar{\mathcal O}\text{ moves at speed }0.6\text{ in the }-z\text{ direction relative to }\mathcal O.}
$$
(Another check: $\bar{\mathcal O}$'s origin, $A^{\bar i} = 0$, must move with $\Delta z/\Delta t = v$. From the last row, $\bar z = 0.75t + 1.25z = 0 \Rightarrow z = -0.6t$ ✓.)

## (b) Inverse

The inverse transformation is the boost with the opposite velocity (Eq. 2.18), $+0.6$ along $z$, so flip the sign of the off-diagonal entries:
$$
\boxed{\Lambda^{-1} = \begin{pmatrix}1.25 & 0 & 0 & -0.75\\ 0 & 1 & 0 & 0\\ 0 & 0 & 1 & 0\\ -0.75 & 0 & 0 & 1.25\end{pmatrix}}
$$
Check the $t$–$z$ block: $\begin{pmatrix}1.25 & 0.75\\0.75 & 1.25\end{pmatrix}\begin{pmatrix}1.25 & -0.75\\-0.75 & 1.25\end{pmatrix} = \begin{pmatrix}1.5625 - 0.5625 & 0\\ 0 & -0.5625 + 1.5625\end{pmatrix} = I$ ✓.

## (c) $\vec A \xrightarrow{\bar{\mathcal O}}(1, 2, 0, 0)$ in $\mathcal O$

$A^\beta = (\Lambda^{-1})^\beta{}_{\bar\alpha}A^{\bar\alpha}$:
$$
A^0 = 1.25(1) - 0.75(0) = 1.25,\quad A^1 = 2,\quad A^2 = 0,\quad A^3 = -0.75(1) + 1.25(0) = -0.75,
$$
$$
\boxed{\vec A \xrightarrow{\mathcal O}(1.25,\ 2,\ 0,\ -0.75)}
$$
Check the invariant: $-1 + 4 = 3$ in $\bar{\mathcal O}$; $-1.5625 + 4 + 0.5625 = 3$ in $\mathcal O$ ✓.

```python
import numpy as np
from schutz_gr import minkowski_dot, boost_general
L = np.array([[1.25, 0, 0, 0.75], [0, 1, 0, 0], [0, 0, 1, 0], [0.75, 0, 0, 1.25]])
print("(a) matches boost with v = (0, 0, −0.6):", np.allclose(L, boost_general([0, 0, -0.6])))
Linv = np.linalg.inv(L)
print("(b) inverse =\n", Linv)
Ab = np.array([1, 2, 0, 0.0]); A = Linv @ Ab
print("(c) A in O =", A, "  A·A:", minkowski_dot(Ab, Ab), minkowski_dot(A, A))
```
