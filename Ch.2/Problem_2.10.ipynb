# Chapter 2, Problem 10 — deriving the basis transformation, Eq. (2.13)

**Problem.** Prove Eq. (2.13), $\vec e_\alpha = \Lambda^{\bar\beta}{}_\alpha\,\vec e_{\bar\beta}$, from the equation
$$
A^\alpha\left(\Lambda^{\bar\beta}{}_\alpha\vec e_{\bar\beta} - \vec e_\alpha\right) = 0
$$
by making specific choices for the components of the arbitrary vector $\vec A$.

**Where the starting equation comes from (§2.2).** $A^\alpha\vec e_\alpha = A^{\bar\beta}\vec e_{\bar\beta}$ (the same vector expanded in two bases), with $A^{\bar\beta} = \Lambda^{\bar\beta}{}_\alpha A^\alpha$ and dummies relabelled. It holds for **every** choice of the four numbers $A^\alpha$.

## Proof

Define four vectors $\vec V_\alpha \equiv \Lambda^{\bar\beta}{}_\alpha\vec e_{\bar\beta} - \vec e_\alpha$ ($\alpha = 0,1,2,3$; the sum on $\bar\beta$ is implied). The hypothesis is
$$
A^0\vec V_0 + A^1\vec V_1 + A^2\vec V_2 + A^3\vec V_3 = \vec 0\qquad\text{for all }(A^0, A^1, A^2, A^3).
$$

**Choice 1:** $\vec A \xrightarrow{\mathcal O}(1, 0, 0, 0)$. Only the $\alpha = 0$ term survives: $1\cdot\vec V_0 = \vec 0$, so $\vec V_0 = \vec 0$, i.e.
$$
\vec e_0 = \Lambda^{\bar\beta}{}_0\,\vec e_{\bar\beta}.
$$

**Choice 2:** $\vec A \xrightarrow{\mathcal O}(0, 1, 0, 0)$ gives $\vec V_1 = \vec 0$: $\vec e_1 = \Lambda^{\bar\beta}{}_1\vec e_{\bar\beta}$.

**Choice 3:** $(0, 0, 1, 0)$ gives $\vec e_2 = \Lambda^{\bar\beta}{}_2\vec e_{\bar\beta}$.

**Choice 4:** $(0, 0, 0, 1)$ gives $\vec e_3 = \Lambda^{\bar\beta}{}_3\vec e_{\bar\beta}$.

These four statements are exactly Eq. (2.13) for each value of the free index $\alpha$:
$$
\boxed{\vec e_\alpha = \Lambda^{\bar\beta}{}_\alpha\,\vec e_{\bar\beta}}\qquad\blacksquare
$$

**Why the trick works.** These four choices are the components of $\vec e_0, \dots, \vec e_3$ themselves: picking $\vec A = \vec e_\gamma$ (so $A^\alpha = \delta_\gamma{}^\alpha$) "selects" one term. In one line: $0 = \delta_\gamma{}^\alpha\vec V_\alpha = \vec V_\gamma$.

**Why a choice is needed at all.** $A^\alpha\vec V_\alpha = 0$ for *one* particular $\vec A$ would not imply $\vec V_\alpha = 0$; the terms could cancel each other. It's the "for all $\vec A$" that forces each $\vec V_\alpha$ to vanish separately.

**Example (§2.2).** For a boost along $x$: $\vec e_0 = \Lambda^{\bar0}{}_0\vec e_{\bar0} + \Lambda^{\bar1}{}_0\vec e_{\bar1} = \gamma\vec e_{\bar0} - v\gamma\,\vec e_{\bar1}$.

```python
import sympy as sp
from schutz_gr import boost_matrix
v = sp.symbols("v", real=True)
L = boost_matrix(v)                              # Λ^β̄_α : row β̄, column α
# Components in Ō: ē_β̄ → unit vector; e_α → Λ^β̄_α ē_β̄ means "column α of Λ"
for a in range(4):
    print(f"e_{a} = " + " + ".join(f"({sp.simplify(L[b, a])}) ē_{b}" for b in range(4) if L[b, a] != 0))
# check: these e_α, expressed in Ō, have the right components in Ō, namely Λ applied to (e_α in O)
print("consistent with A^β̄ = Λ^β̄_α A^α:", all(L[:, a] == L * sp.Matrix([1 if k == a else 0 for k in range(4)]) for a in range(4)))
```
