# Chapter 2, Problem 22 — four-momentum bookkeeping

**Problem.**
- **(a)** Find the energy, rest mass, and three-velocity $\mathbf v$ of a particle whose four-momentum has components $(4, 1, 1, 0)$ kg.
- **(b)** Two particles with four-momenta $\vec p_1 \xrightarrow{\mathcal O}(3, -1, 0, 0)$ kg and $\vec p_2 \xrightarrow{\mathcal O}(2, 1, 1, 0)$ kg collide; both are destroyed and three new particles produced, two of which have $\vec p_3 \xrightarrow{\mathcal O}(1, 1, 0, 0)$ kg and $\vec p_4 \xrightarrow{\mathcal O}(1, -\tfrac12, 0, 0)$ kg. Find the four-momentum, energy, rest mass and three-velocity of the third particle produced. Find the CM frame's three-velocity.

**Tools.** $\vec p \to (E, \mathbf p)$ (Eq. 2.20); $\vec p\cdot\vec p = -m^2$ (Eq. 2.33), so $m = \sqrt{E^2 - |\mathbf p|^2}$; $\mathbf v = \mathbf p/E$ (since $\vec p = m\vec U = m\gamma(1, \mathbf v)$); and conservation of total four-momentum (Eq. 2.22). Units: $c = 1$, so energy and momentum are in kg.

## (a) $\vec p \to (4, 1, 1, 0)$ kg

- **Energy:** $E = p^0 = \boxed{4\ \text{kg}}$ (i.e. $4c^2 = 3.6\times10^{17}$ J).
- **Rest mass:** $m^2 = E^2 - |\mathbf p|^2 = 16 - 1 - 1 = 14$, so $\boxed{m = \sqrt{14} \approx 3.74\ \text{kg}}$.
- **Velocity:** $\mathbf v = \mathbf p/E = \boxed{\left(\tfrac14,\ \tfrac14,\ 0\right)}$, speed $\sqrt2/4 \approx 0.354$.

Check: $\gamma = E/m = 4/\sqrt{14} = 1.069$ and $1/\sqrt{1 - 1/8} = 1.069$ ✓.

## (b) The collision

**Conservation:** $\vec p_1 + \vec p_2 = \vec p_3 + \vec p_4 + \vec p_5$.

**Total before:** $\vec p_1 + \vec p_2 = (3 + 2,\ -1 + 1,\ 0 + 1,\ 0) = (5, 0, 1, 0)$.

**Known products:** $\vec p_3 + \vec p_4 = (2,\ \tfrac12,\ 0,\ 0)$.

**Third particle:**
$$
\boxed{\vec p_5 = (5, 0, 1, 0) - (2, \tfrac12, 0, 0) = \left(3,\ -\tfrac12,\ 1,\ 0\right)\ \text{kg}}
$$
- Energy $\boxed{E_5 = 3\ \text{kg}}$.
- Rest mass: $m_5^2 = 9 - \tfrac14 - 1 = \tfrac{31}{4}$, $\boxed{m_5 = \tfrac{\sqrt{31}}{2} \approx 2.78\ \text{kg}}$.
- Velocity: $\boxed{\mathbf v_5 = \left(-\tfrac16,\ \tfrac13,\ 0\right)}$, speed $\sqrt5/6 \approx 0.373$.

**Sanity checks on the other particles.** $m_3^2 = 1 - 1 = 0$: particle 3 is massless (moves at speed 1, e.g. a photon). $m_4^2 = 1 - \tfrac14 = \tfrac34$. $m_1^2 = 9 - 1 = 8$, $m_2^2 = 4 - 2 = 2$. Rest mass is **not** conserved: before, $\sqrt8 + \sqrt2 \approx 4.24$ kg; after, $0 + 0.866 + 2.78 \approx 3.65$ kg. Some rest mass became kinetic energy. Only the *total four-momentum* is conserved.

**CM frame velocity.** The CM frame is where the total spatial momentum vanishes (Eq. 2.23). The total four-momentum $\vec P = (5, 0, 1, 0)$ is timelike ($-25 + 1 < 0$), so it is $M\vec U_{CM}$ for the four-velocity of the CM frame (Problem 17). Its 3-velocity is
$$
\boxed{\mathbf v_{CM} = \frac{\mathbf P}{P^0} = \left(0,\ \tfrac15,\ 0\right)}
$$
— $0.2$ in the $+y$ direction. The total rest-frame energy is $M = \sqrt{25 - 1} = \sqrt{24} \approx 4.90$ kg. Boosting $\vec P$ by $\mathbf v_{CM}$ gives $(\sqrt{24}, 0, 0, 0)$ (checked below).

```python
import sympy as sp
from schutz_gr import minkowski_dot, boost_general
R = sp.Rational
def describe(p, name):
    m2 = -minkowski_dot(p, p)
    v = [p[i] / p[0] for i in (1, 2, 3)]
    print(f"{name}: p = {list(p)}, E = {p[0]}, m² = {m2}, m = {sp.sqrt(m2)} ≈ {float(sp.sqrt(m2)):.4f}, v = {v}")
describe(sp.Matrix([4, 1, 1, 0]), "(a)")
p1, p2 = sp.Matrix([3, -1, 0, 0]), sp.Matrix([2, 1, 1, 0])
p3, p4 = sp.Matrix([1, 1, 0, 0]), sp.Matrix([1, R(-1, 2), 0, 0])
p5 = p1 + p2 - p3 - p4
for k, p in enumerate((p1, p2, p3, p4, p5), 1):
    describe(p, f"p{k}")
P = p1 + p2
vcm = [P[i] / P[0] for i in (1, 2, 3)]
print("v_CM =", vcm, "  P in CM frame =", list(sp.simplify(boost_general(vcm) * P)))
```
