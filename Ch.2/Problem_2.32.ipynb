# Chapter 2, Problem 32 — Compton scattering

**Problem.** A particle of charge $e$ and rest mass $m$, initially at rest in the laboratory, scatters a photon of initial frequency $\nu_i$ (Compton scattering). The scattered photon comes off at angle $\theta$ from the incident direction. Use conservation of four-momentum to deduce that the photon's final frequency $\nu_f$ is given by
$$
\frac{1}{\nu_f} = \frac{1}{\nu_i} + h\,\frac{1 - \cos\theta}{m}. \tag{2.43}
$$

## Four-momenta in the lab frame

- Particle before: $\vec P_i \to (m, 0, 0, 0)$.
- Photon before (along $x$): $\vec p_i \to h\nu_i\,(1, 1, 0, 0)$.
- Photon after (angle $\theta$ in the $x$–$y$ plane): $\vec p_f \to h\nu_f\,(1, \cos\theta, \sin\theta, 0)$.
- Particle after: $\vec P_f$, unknown. We don't want it, so we eliminate it.

## Eliminate $\vec P_f$ using its magnitude

**Conservation:** $\vec P_i + \vec p_i = \vec P_f + \vec p_f$, so
$$
\vec P_f = \vec P_i + (\vec p_i - \vec p_f).
$$
**Square both sides** (dot each with itself). The particle is still the same particle, so $\vec P_f\cdot\vec P_f = \vec P_i\cdot\vec P_i = -m^2$ (Eq. 2.33):
$$
-m^2 = \vec P_i\cdot\vec P_i + 2\,\vec P_i\cdot(\vec p_i - \vec p_f) + (\vec p_i - \vec p_f)\cdot(\vec p_i - \vec p_f).
$$
With $\vec P_i\cdot\vec P_i = -m^2$, this leaves
$$
0 = 2\,\vec P_i\cdot(\vec p_i - \vec p_f) + \vec p_i\cdot\vec p_i - 2\,\vec p_i\cdot\vec p_f + \vec p_f\cdot\vec p_f .
$$
**Evaluate each piece:**
- Photons are null: $\vec p_i\cdot\vec p_i = \vec p_f\cdot\vec p_f = 0$.
- $\vec P_i\cdot\vec p_i = -m\,h\nu_i$, and $\vec P_i\cdot\vec p_f = -m\,h\nu_f$ (only time components contribute).
- $\vec p_i\cdot\vec p_f = h^2\nu_i\nu_f(-1 + \cos\theta)$.

Substituting:
$$
0 = -2mh(\nu_i - \nu_f) + 2h^2\nu_i\nu_f(1 - \cos\theta).
$$
**Divide by $2mh\,\nu_i\nu_f$:**
$$
\frac{\nu_i - \nu_f}{\nu_i\nu_f} = \frac{h}{m}(1 - \cos\theta)
\quad\Longrightarrow\quad
\boxed{\frac{1}{\nu_f} = \frac{1}{\nu_i} + \frac{h}{m}(1 - \cos\theta)}\qquad\blacksquare
$$

**In familiar form.** Multiply by $c$ (restoring units, $h/m \to h/(mc^2)$): $\lambda_f - \lambda_i = \frac{h}{mc}(1 - \cos\theta)$. The wavelength shift is independent of the initial wavelength, set by the Compton wavelength $h/mc$ ($2.43\times10^{-12}$ m for an electron).

**Checks.** $\theta = 0$: no scattering, $\nu_f = \nu_i$ ✓. $\theta = \pi$ (backscatter) gives the maximum shift, $1/\nu_f = 1/\nu_i + 2h/m$. Always $\nu_f \leq \nu_i$: a particle initially at rest recoils and takes energy. The charge $e$ never enters: the kinematics is purely conservation of four-momentum.

```python
import sympy as sp
from schutz_gr import minkowski_dot
m, h, ni, nf, th = sp.symbols("m h nu_i nu_f theta", positive=True)
Pi = sp.Matrix([m, 0, 0, 0]); pi_ = h*ni*sp.Matrix([1, 1, 0, 0]); pf = h*nf*sp.Matrix([1, sp.cos(th), sp.sin(th), 0])
Pf = Pi + pi_ - pf
eq = sp.Eq(sp.trigsimp(sp.expand(minkowski_dot(Pf, Pf) + m**2)), 0)   # sin²+cos² = 1 ⇒ ν_f² terms cancel (photons are null): linear in ν_f
print('equation:', sp.factor(eq.lhs), '= 0')
sol = sp.solve(eq, nf)[0]
print("ν_f =", sp.simplify(sol))
print("1/ν_f − 1/ν_i − (h/m)(1 − cos θ) =", sp.simplify(1/sol - 1/ni - h/m*(1 - sp.cos(th))))
lam_c = 6.626e-34 / (9.109e-31 * 3e8)
print(f"electron Compton wavelength h/mc = {lam_c:.3e} m")
```
