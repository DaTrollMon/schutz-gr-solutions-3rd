# Chapter 2, Problem 7 — the basis vectors in index notation

**Problem.**
- **(a)** Verify Eq. (2.10), $(\vec e_\alpha)^\beta = \delta_\alpha{}^\beta$, for all $\alpha, \beta$.
- **(b)** Prove Eq. (2.11), $\vec A = A^\alpha\vec e_\alpha$, from Eq. (2.9).

**Notation.** $(\vec e_\alpha)^\beta$ is the $\beta$-th component (in $\mathcal O$) of the $\alpha$-th basis vector. The Kronecker delta is $\delta_\alpha{}^\beta = 1$ if $\alpha = \beta$, else 0.

## (a) Checking all 16 cases

Eq. (2.9) lists the components of each basis vector. Read them as a table, with row $\alpha$ for the vector and column $\beta$ for the component:

| | $\beta=0$ | $\beta=1$ | $\beta=2$ | $\beta=3$ |
|---|---|---|---|---|
| $\vec e_0$ | 1 | 0 | 0 | 0 |
| $\vec e_1$ | 0 | 1 | 0 | 0 |
| $\vec e_2$ | 0 | 0 | 1 | 0 |
| $\vec e_3$ | 0 | 0 | 0 | 1 |

Each entry is 1 exactly when the row label equals the column label, and 0 otherwise. That is the definition of $\delta_\alpha{}^\beta$. So $(\vec e_\alpha)^\beta = \delta_\alpha{}^\beta$ for all 16 pairs $(\alpha, \beta)$: the table *is* the identity matrix. ✓

## (b) $\vec A = A^\alpha\vec e_\alpha$

Two vectors are equal if their components in $\mathcal O$ are equal (Problem 8 then makes that true in every frame). Compute the $\beta$-th component of the right side, using the component rules (2.8) for sums and multiples:
$$
\big(A^\alpha\vec e_\alpha\big)^\beta = A^\alpha\,(\vec e_\alpha)^\beta \overset{(a)}{=} A^\alpha\,\delta_\alpha{}^\beta = A^\beta .
$$
The last step is the "substitution property" of the delta: in $\sum_\alpha A^\alpha\delta_\alpha{}^\beta$, only the $\alpha = \beta$ term survives, and it has coefficient 1.

So $A^\alpha\vec e_\alpha$ has components $(A^0, A^1, A^2, A^3)$, the same as $\vec A$. Hence $\vec A = A^\alpha\vec e_\alpha$. $\blacksquare$

Written out: $A^0(1,0,0,0) + A^1(0,1,0,0) + A^2(0,0,1,0) + A^3(0,0,0,1) = (A^0, A^1, A^2, A^3)$.

```python
import sympy as sp
e = [sp.Matrix([1 if b == a else 0 for b in range(4)]) for a in range(4)]   # Eq. (2.9)
print("(a) all 16 cases (e_α)^β == δ_α^β:",
      all(e[a][b] == sp.KroneckerDelta(a, b) for a in range(4) for b in range(4)))
A = sp.symbols("A0:4")
rhs = sum((A[a] * e[a] for a in range(4)), sp.zeros(4, 1))
print("(b) A^α e_α =", list(rhs), "== (A^0, A^1, A^2, A^3):", list(rhs) == list(A))
```
