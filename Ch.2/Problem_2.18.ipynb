# Chapter 2, Problem 18 — orthogonality facts

**Problem.**
- **(a)** Show that the sum of any two orthogonal spacelike vectors is spacelike.
- **(b)** Show that a timelike vector and a null vector cannot be orthogonal.

Recall (§2.5): $\vec A$ is spacelike if $\vec A^{\,2} > 0$, timelike if $< 0$, null if $= 0$ (and $\vec A \neq 0$). Orthogonal means $\vec A\cdot\vec B = 0$.

## (a) Orthogonal spacelike + spacelike = spacelike

Let $\vec A^{\,2} > 0$, $\vec B^{\,2} > 0$, $\vec A\cdot\vec B = 0$. The dot product is bilinear and symmetric (Problem 34), so
$$
(\vec A + \vec B)\cdot(\vec A + \vec B) = \vec A^{\,2} + 2\,\vec A\cdot\vec B + \vec B^{\,2} = \vec A^{\,2} + \vec B^{\,2} > 0 .
$$
So $\vec A + \vec B$ is spacelike. $\blacksquare$ (Pythagoras survives for spacelike vectors.)

**Orthogonality is essential.** Without it the sum can be anything. E.g. $\vec A \to (1, 2, 0, 0)$ (spacelike, $A^2 = 3$) and $\vec B \to (1, -2, 0, 0)$ ($B^2 = 3$) have $\vec A\cdot\vec B = -1 - 4 = -5$, and $\vec A + \vec B \to (2, 0, 0, 0)$ is timelike.

## (b) Timelike ⟂ null is impossible

Let $\vec T$ be timelike and $\vec N$ null ($\vec N^{\,2} = 0$, $\vec N \neq 0$). The dot product is frame-independent (§2.5), so compute it in the most convenient frame: the one where $\vec T$ has no spatial components (Problem 17b), $\vec T \to (a, 0, 0, 0)$ with $a \neq 0$.

In that frame write $\vec N \to (n^0, \mathbf n)$. Null means $(n^0)^2 = |\mathbf n|^2$. If $n^0 = 0$ then $\mathbf n = 0$ too and $\vec N = 0$, contradicting $\vec N \neq 0$. So $n^0 \neq 0$. Then
$$
\vec T\cdot\vec N = -a\,n^0 + 0 = -a\,n^0 \neq 0 . \qquad\blacksquare
$$

**Geometrically.** A vector orthogonal to timelike $\vec T$ lies in $\vec T$'s rest-frame "space" ($t = $ const), so it's purely spatial there: spacelike or zero. Null vectors always have a nonzero time component, so they never lie in that space. (Contrast: a null vector *is* orthogonal to itself, and two different null vectors can be orthogonal only if they are parallel.)

```python
import numpy as np
from schutz_gr import minkowski_dot, boost_general
rng = np.random.default_rng(18)
# (a) random orthogonal spacelike pairs: build B ⟂ A by Gram–Schmidt with the Minkowski product
worst = np.inf
for _ in range(2000):
    A = rng.normal(size=4); A[0] *= 0.3
    B = rng.normal(size=4); B[0] *= 0.3
    if minkowski_dot(A, A) <= 0: continue
    B = B - minkowski_dot(A, B) / minkowski_dot(A, A) * A
    if minkowski_dot(B, B) <= 0: continue
    worst = min(worst, minkowski_dot(A + B, A + B))
print("(a) smallest (A+B)² found over random orthogonal spacelike pairs:", round(worst, 6), "(> 0)")
# (b) |T·N| over random timelike T and null N never reaches 0
vals = []
for _ in range(2000):
    n = rng.normal(size=3); N = np.r_[np.linalg.norm(n) * rng.choice([-1, 1]), n]
    s = rng.normal(size=3) * 0.4; T = np.r_[np.linalg.norm(s) + rng.uniform(0.01, 2), s]
    vals.append(abs(minkowski_dot(T, N)))
print("(b) min |T·N| over 2000 random pairs:", round(min(vals), 6), "(never 0)")
```
