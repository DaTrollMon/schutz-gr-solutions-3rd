# Chapter 2, Problem 29 — the product rule for the dot product

**Problem.** Prove, using the component expressions Eqs. (2.24) and (2.26), that
$$
\frac{d}{d\tau}\big(\vec U\cdot\vec U\big) = 2\,\vec U\cdot\frac{d\vec U}{d\tau}.
$$

## Proof

Work in a fixed inertial frame $\mathcal O$. Along a world line, the components $U^\alpha(\tau)$ are four ordinary functions of $\tau$. The basis vectors of an inertial frame don't depend on $\tau$, so the components of $d\vec U/d\tau$ are just $dU^\alpha/d\tau$.

**Left side**, by Eq. (2.24):
$$
\vec U\cdot\vec U = -(U^0)^2 + (U^1)^2 + (U^2)^2 + (U^3)^2 .
$$
Differentiate term by term with the ordinary chain rule, $\frac{d}{d\tau}(f^2) = 2f\frac{df}{d\tau}$:
$$
\frac{d}{d\tau}\big(\vec U\cdot\vec U\big) = -2U^0\frac{dU^0}{d\tau} + 2U^1\frac{dU^1}{d\tau} + 2U^2\frac{dU^2}{d\tau} + 2U^3\frac{dU^3}{d\tau}.
$$

**Right side**, by Eq. (2.26) with $\vec A = \vec U$ and $\vec B = d\vec U/d\tau$:
$$
2\,\vec U\cdot\frac{d\vec U}{d\tau} = 2\left(-U^0\frac{dU^0}{d\tau} + U^1\frac{dU^1}{d\tau} + U^2\frac{dU^2}{d\tau} + U^3\frac{dU^3}{d\tau}\right).
$$
The two expressions are identical. $\blacksquare$

**Compactly:** $\vec U\cdot\vec U = \eta_{\alpha\beta}U^\alpha U^\beta$ with constant $\eta_{\alpha\beta}$, so $\frac{d}{d\tau}(\eta_{\alpha\beta}U^\alpha U^\beta) = \eta_{\alpha\beta}\dot U^\alpha U^\beta + \eta_{\alpha\beta}U^\alpha\dot U^\beta = 2\eta_{\alpha\beta}U^\alpha\dot U^\beta$, using the symmetry of $\eta$.

**Consequence (§2.6).** Since $\vec U\cdot\vec U = -1$ is constant, the left side is 0, so $\vec U\cdot\vec a = 0$: the four-acceleration is always orthogonal to the four-velocity (Eq. 2.32). (In curved spacetime, Chapter 5 onward, the basis vectors change from point to point. Then $d/d\tau$ becomes a covariant derivative and this step needs more care.)

```python
import sympy as sp
from schutz_gr import minkowski_dot
tau = sp.symbols("tau", real=True)
U = sp.Matrix([sp.Function(f"U{a}")(tau) for a in range(4)])
lhs = sp.diff(minkowski_dot(U, U), tau)
rhs = 2 * minkowski_dot(U, U.diff(tau))
print("d/dτ(U·U) − 2 U·dU/dτ =", sp.simplify(lhs - rhs))
```
