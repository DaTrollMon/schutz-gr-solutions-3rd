# Chapter 3, Problem 31 — the projection tensor $P_{\mu\nu} = \eta_{\mu\nu} + U_\mu U_\nu$

**Problem.** Consider a timelike unit four-vector $\vec U$, and the tensor $\mathbf P$ whose components are given by
$$
P_{\mu\nu} = \eta_{\mu\nu} + U_\mu U_\nu .
$$
- **(a)** Show that $\mathbf P$ is a projection operator that projects an arbitrary vector $\vec V$ into one orthogonal to $\vec U$. That is, show that the vector $\vec V_\perp$ whose components are $V^\alpha_\perp = P^\alpha{}_\beta V^\beta = (\eta^\alpha{}_\beta + U^\alpha U_\beta)V^\beta$ is (i) orthogonal to $\vec U$, and (ii) unaffected by $\mathbf P$: $V^\alpha_{\perp\perp} := P^\alpha{}_\beta V^\beta_\perp = V^\alpha_\perp$.
- **(b)** Show that for an arbitrary non-null vector $\vec q$, the tensor that projects orthogonally to it has components $\eta_{\mu\nu} - q_\mu q_\nu/(q^\alpha q_\alpha)$. How does this fail for null vectors? How does this relate to the definition of $\mathbf P$?
- **(c)** Show that $\mathbf P$ defined above is the metric tensor for vectors perpendicular to $\vec U$: $\mathbf P(\vec V_\perp, \vec W_\perp) = \mathbf g(\vec V_\perp, \vec W_\perp) = \vec V_\perp\cdot\vec W_\perp$.

**Given:** $U^\alpha U_\alpha = -1$. Also $\eta^\alpha{}_\beta = \delta^\alpha{}_\beta$ (Problem 24c), so $V^\alpha_\perp = V^\alpha + U^\alpha(U_\beta V^\beta) = V^\alpha + (\vec U\cdot\vec V)\,U^\alpha$.

## (a)(i) $\vec V_\perp$ is orthogonal to $\vec U$

$$
\vec U\cdot\vec V_\perp = U_\alpha V^\alpha_\perp = U_\alpha V^\alpha + (\vec U\cdot\vec V)\,U_\alpha U^\alpha = \vec U\cdot\vec V + (\vec U\cdot\vec V)(-1) = 0. \qquad\blacksquare
$$

## (a)(ii) Projecting twice changes nothing

Apply $\mathbf P$ to $\vec V_\perp$:
$$
V^\alpha_{\perp\perp} = V^\alpha_\perp + (\vec U\cdot\vec V_\perp)\,U^\alpha = V^\alpha_\perp + 0 = V^\alpha_\perp. \qquad\blacksquare
$$
In operator terms, $P^\alpha{}_\mu P^\mu{}_\beta = P^\alpha{}_\beta$ ($\mathbf P^2 = \mathbf P$):
$$
(\delta^\alpha{}_\mu + U^\alpha U_\mu)(\delta^\mu{}_\beta + U^\mu U_\beta) = \delta^\alpha{}_\beta + 2U^\alpha U_\beta + U^\alpha(U_\mu U^\mu)U_\beta = \delta^\alpha{}_\beta + U^\alpha U_\beta .
$$
Also $P^\alpha{}_\beta U^\beta = U^\alpha + U^\alpha(-1) = 0$: $\mathbf P$ annihilates $\vec U$. So $\mathbf P$ kills the component along $\vec U$ and leaves the 3-dimensional orthogonal subspace untouched. In the rest frame of $\vec U$ ($U^\alpha = (1, 0, 0, 0)$, $U_\alpha = (-1, 0, 0, 0)$): $P_{\mu\nu} = \operatorname{diag}(0, 1, 1, 1)$; $\mathbf P$ simply discards the time component.

## (b) Projection orthogonal to a general non-null $\vec q$

Let $Q_{\mu\nu} := \eta_{\mu\nu} - q_\mu q_\nu/(q^\alpha q_\alpha)$, so $Q^\alpha{}_\beta V^\beta = V^\alpha - \dfrac{\vec q\cdot\vec V}{\vec q\cdot\vec q}\,q^\alpha$. Then
$$
\vec q\cdot(\mathbf Q\vec V) = \vec q\cdot\vec V - \frac{\vec q\cdot\vec V}{\vec q\cdot\vec q}\,\vec q\cdot\vec q = 0,
$$
and since $\mathbf Q\vec V \perp \vec q$, a second application subtracts nothing: $\mathbf Q^2 = \mathbf Q$. Also $\mathbf Q\vec q = 0$. So $\mathbf Q$ projects orthogonally to $\vec q$. (This is Gram–Schmidt: subtract the part of $\vec V$ along $\vec q$.)

**Null $\vec q$:** $q^\alpha q_\alpha = 0$, so the formula divides by zero. Not a technicality: a null vector is orthogonal to **itself**, so it lies in its own orthogonal "complement". There's no way to split $\vec V$ uniquely into a part along $\vec q$ plus a part orthogonal to $\vec q$, since $\vec q$ is in both.

**Relation to $\mathbf P$:** for a unit timelike vector $q^\alpha q_\alpha = -1$, and $\eta_{\mu\nu} - q_\mu q_\nu/(-1) = \eta_{\mu\nu} + U_\mu U_\nu$. $\mathbf P$ is the special case $\vec q = \vec U$.

## (c) $\mathbf P$ is the metric on the subspace orthogonal to $\vec U$

For $\vec V_\perp, \vec W_\perp$ orthogonal to $\vec U$:
$$
\mathbf P(\vec V_\perp, \vec W_\perp) = P_{\mu\nu}V^\mu_\perp W^\nu_\perp = \eta_{\mu\nu}V^\mu_\perp W^\nu_\perp + (U_\mu V^\mu_\perp)(U_\nu W^\nu_\perp) = \vec V_\perp\cdot\vec W_\perp + 0\cdot0 = \mathbf g(\vec V_\perp, \vec W_\perp). \qquad\blacksquare
$$
More generally, for arbitrary $\vec V, \vec W$, $\mathbf P(\vec V, \vec W) = \vec V_\perp\cdot\vec W_\perp$: the inner product of their parts orthogonal to $\vec U$. $\mathbf P$ is the **spatial metric** of the observer with four-velocity $\vec U$. This is how "space" for a given observer is extracted from spacetime (e.g. the fluid stress tensor in Chapter 4).

```python
import sympy as sp
from schutz_gr import four_velocity
eta = sp.diag(-1, 1, 1, 1)
U = four_velocity([sp.Rational(1, 3), sp.Rational(-1, 4), sp.Rational(1, 2)])     # a unit timelike vector
Ulo = eta * U
print("U·U =", sp.simplify((U.T * eta * U)[0]))
P_lo = eta + Ulo * Ulo.T                       # P_{μν}
P_mix = eta.inv() * P_lo                       # P^α_β
V = sp.Matrix(sp.symbols("V0:4")); W = sp.Matrix(sp.symbols("W0:4"))
Vp, Wp = P_mix * V, P_mix * W
print("(a)(i)  U·V⊥ =", sp.simplify((Ulo.T * Vp)[0]))
print("(a)(ii) P² = P:", sp.simplify(P_mix * P_mix - P_mix) == sp.zeros(4), "| P U = 0:", sp.simplify(P_mix * U) == sp.zeros(4, 1))
q = sp.Matrix([1, 2, 0, 1]); qlo = eta * q; qq = (q.T * qlo)[0]
Q = eta.inv() * (eta - qlo * qlo.T / qq)
print("(b) q·q =", qq, "| q·(QV) =", sp.simplify((qlo.T * Q * V)[0]), "| Q² = Q:", sp.simplify(Q*Q - Q) == sp.zeros(4))
print("    null q: q·q =", (sp.Matrix([1, 1, 0, 0]).T * eta * sp.Matrix([1, 1, 0, 0]))[0], "→ division by zero")
print("(c) P(V⊥,W⊥) − V⊥·W⊥ =", sp.simplify((Vp.T * P_lo * Wp)[0] - (Vp.T * eta * Wp)[0]))
print("    P(V,W) − V⊥·W⊥ =", sp.simplify((V.T * P_lo * W)[0] - (Vp.T * eta * Wp)[0]))
```
