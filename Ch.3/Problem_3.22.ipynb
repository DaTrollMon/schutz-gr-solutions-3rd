# Chapter 3, Problem 22 — vectors as linear functions of one-forms

**Problem.** Suppose that instead of defining vectors first, we had begun by defining one-forms, aided by pictures like Fig. 3.4. Then we could have introduced vectors as linear real-valued functions of one-forms, and defined vector algebra by the analogs of Eqs. (3.6a) and (3.6b) (i.e. by exchanging arrows for tildes). Prove that, so defined, vectors form a vector space. This is another example of the duality between vectors and one-forms.

**The setup.** Let $\mathcal V^*$ be the vector space of one-forms (taken as given). A "vector" is a map $\vec V: \mathcal V^* \to \mathbb R$ with
$$
\vec V(a\tilde p + b\tilde q) = a\,\vec V(\tilde p) + b\,\vec V(\tilde q)\qquad(\text{linearity}).
$$
Algebra, by the analogs of (3.6a,b):
$$
(\vec V + \vec W)(\tilde p) := \vec V(\tilde p) + \vec W(\tilde p),\qquad (\alpha\vec V)(\tilde p) := \alpha\,\vec V(\tilde p).
$$

## Proof

This is word for word the argument of Problem 2, with the roles exchanged. Two things to check: (i) the operations produce vectors (closure), and (ii) the vector-space axioms hold.

**(i) Closure.** $\vec V + \vec W$ is real-valued; is it linear?
$$
(\vec V + \vec W)(a\tilde p + b\tilde q) = \vec V(a\tilde p + b\tilde q) + \vec W(a\tilde p + b\tilde q) = a\vec V(\tilde p) + b\vec V(\tilde q) + a\vec W(\tilde p) + b\vec W(\tilde q) = a(\vec V + \vec W)(\tilde p) + b(\vec V + \vec W)(\tilde q).
$$
Similarly $(\alpha\vec V)(a\tilde p + b\tilde q) = \alpha[a\vec V(\tilde p) + b\vec V(\tilde q)] = a(\alpha\vec V)(\tilde p) + b(\alpha\vec V)(\tilde q)$. Both are linear functions of one-forms, i.e. vectors.

**(ii) Axioms.** Each one reduces, value by value, to the same property of real numbers. Two maps are equal when they agree on every $\tilde p$.

| axiom | why it holds (evaluate on arbitrary $\tilde p$) |
|---|---|
| $\vec V + \vec W = \vec W + \vec V$ | $\vec V(\tilde p) + \vec W(\tilde p) = \vec W(\tilde p) + \vec V(\tilde p)$ |
| $(\vec U + \vec V) + \vec W = \vec U + (\vec V + \vec W)$ | associativity of real addition |
| zero: $\vec 0(\tilde p) := 0$ | linear, and $(\vec V + \vec 0)(\tilde p) = \vec V(\tilde p)$ |
| negative: $(-\vec V)(\tilde p) := -\vec V(\tilde p)$ | $(\vec V + (-\vec V))(\tilde p) = 0 = \vec 0(\tilde p)$ |
| $\alpha(\beta\vec V) = (\alpha\beta)\vec V$ | $\alpha(\beta\vec V(\tilde p)) = (\alpha\beta)\vec V(\tilde p)$ |
| $1\cdot\vec V = \vec V$ | $1\cdot\vec V(\tilde p) = \vec V(\tilde p)$ |
| $\alpha(\vec V + \vec W) = \alpha\vec V + \alpha\vec W$ | distributivity in $\mathbb R$ |
| $(\alpha + \beta)\vec V = \alpha\vec V + \beta\vec V$ | distributivity in $\mathbb R$ |

So the vectors form a vector space, the **dual** $\mathcal V^{**}$ of the space of one-forms. $\blacksquare$

## The duality, made concrete

If one-forms have a basis $\{\tilde\omega^\alpha\}$, define $\vec e_\beta$ by $\vec e_\beta(\tilde\omega^\alpha) := \delta^\alpha{}_\beta$, the analog of Eq. (3.12). Then $\vec V = V^\beta\vec e_\beta$ with $V^\beta := \vec V(\tilde\omega^\beta)$. Check on $\tilde p = p_\alpha\tilde\omega^\alpha$: $\vec V(\tilde p) = p_\alpha\vec V(\tilde\omega^\alpha) = p_\alpha V^\alpha$, the contraction of Eq. (3.54). Everything in §3.3 runs equally well "backwards".

Notably, no picture of arrows was needed. In finite dimensions the double dual $\mathcal V^{**}$ is naturally the same space as $\mathcal V$ (the map $\vec V \mapsto [\tilde p \mapsto \tilde p(\vec V)]$ is an isomorphism requiring no basis or metric), which is why Schutz can say vectors and one-forms are on equal footing.

```python
import sympy as sp
# model: one-forms = row 4-tuples; a "vector" is any linear map of them, determined by its values V^α on the basis ω^α
p = sp.Matrix(sp.symbols("p0:4")).T; q = sp.Matrix(sp.symbols("q0:4")).T
a, b, al = sp.symbols("a b alpha")
mk = lambda comps: (lambda form: (form * sp.Matrix(comps))[0])
V, W = mk(sp.symbols("V0:4")), mk(sp.symbols("W0:4"))
add = lambda F, G: (lambda form: F(form) + G(form)); smul = lambda c, F: (lambda form: c * F(form))
S = add(V, smul(al, W))
print("V + αW is linear:", sp.expand(S(a*p + b*q) - a*S(p) - b*S(q)) == 0)
```
