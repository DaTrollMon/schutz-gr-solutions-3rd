# Chapter 3, Problem 27 — lowering preserves antisymmetry; equal one-forms give equal vectors

**Problem.**
- **(a)** Suppose $\mathsf A$ is an antisymmetric $\binom20$ tensor. Show that $\{A_{\alpha\beta}\}$, obtained by lowering indices by using the metric tensor, are components of an antisymmetric $\binom02$ tensor.
- **(b)** Suppose $V^\alpha = W^\alpha$. Prove that $V_\alpha = W_\alpha$.

> **Text note.** My copy's (b) reads "Suppose $V_\alpha = W_\alpha$. Prove that $V_\alpha = W_\alpha$", with index heights lost. The two meaningful readings are "$V^\alpha = W^\alpha \Rightarrow V_\alpha = W_\alpha$" and the converse "$V_\alpha = W_\alpha \Rightarrow V^\alpha = W^\alpha$". Both are proved below.

## (a)

**It's a tensor.** $A_{\alpha\beta} := \eta_{\alpha\mu}\eta_{\beta\nu}A^{\mu\nu}$ (Eqs. 3.56–3.58). This is the component form of the $\binom02$ tensor $(\vec U, \vec V) \mapsto \mathsf A(\mathbf g(\vec U,\cdot), \mathbf g(\vec V,\cdot))$, i.e. $\mathsf A$ with the metric-associated one-forms of its arguments fed in. That map is bilinear in $\vec U, \vec V$, so it's a $\binom02$ tensor, with components $A^{\mu\nu}\eta_{\mu\alpha}\eta_{\nu\beta}$.

**It's antisymmetric.**
$$
A_{\beta\alpha} = \eta_{\beta\mu}\eta_{\alpha\nu}A^{\mu\nu}
= \eta_{\alpha\nu}\eta_{\beta\mu}A^{\mu\nu}
= \eta_{\alpha\nu}\eta_{\beta\mu}\big(-A^{\nu\mu}\big)
= -\eta_{\alpha\nu}\eta_{\beta\mu}A^{\nu\mu}
= -A_{\alpha\beta}.
$$
(Second step: reorder the numbers $\eta$. Third: antisymmetry of $A^{\mu\nu}$. Last: the definition of $A_{\alpha\beta}$, with $\nu, \mu$ as the names of the dummies.) $\blacksquare$

The argument used nothing about $\eta$ except that it is the same matrix in both slots, so it holds for any metric. In Minkowski coordinates it's visible directly: $A_{\alpha\beta}$ differs from $A^{\alpha\beta}$ only by the sign of row 0 and column 0, and $A_{0i} = -A^{0i} = A^{i0} = -A_{i0}$.

## (b)

**$V^\alpha = W^\alpha \Rightarrow V_\alpha = W_\alpha$.** Multiply both sides by $\eta_{\beta\alpha}$ and sum over $\alpha$:
$$
V_\beta = \eta_{\beta\alpha}V^\alpha = \eta_{\beta\alpha}W^\alpha = W_\beta . \qquad\blacksquare
$$

**Converse, $V_\alpha = W_\alpha \Rightarrow V^\alpha = W^\alpha$.** Multiply by the inverse metric:
$$
\eta^{\beta\alpha}V_\alpha = \eta^{\beta\alpha}\eta_{\alpha\mu}V^\mu = \delta^\beta{}_\mu V^\mu = V^\beta,
$$
and likewise for $W$. So $V^\beta = \eta^{\beta\alpha}V_\alpha = \eta^{\beta\alpha}W_\alpha = W^\beta$. $\blacksquare$

The point: index raising/lowering is an **invertible** linear map (because $\mathbf g$ is nondegenerate), so a vector and its one-form carry identical information. An equation can be moved freely between index positions as long as the same index is moved on both sides.

```python
import sympy as sp
eta = sp.diag(-1, 1, 1, 1)
Aup = sp.Matrix(4, 4, lambda i, j: 0 if i == j else (sp.Symbol(f"a{i}{j}") if i < j else -sp.Symbol(f"a{j}{i}")))
Alo = eta * Aup * eta.T            # A_{αβ} = η_{αμ} η_{βν} A^{μν}
print("(a) A_{αβ} antisymmetric:", Alo == -Alo.T); display(Alo)
V = sp.Matrix(sp.symbols("V0:4"))
print("(b) lowering is invertible: η^{-1}(η V) = V:", eta.inv() * (eta * V) == V, "  det η =", eta.det())
```
