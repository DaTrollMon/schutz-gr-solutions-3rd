# Chapter 3, Problem 24 — symmetric parts, index gymnastics, and $\eta^\alpha{}_\beta$

**Problem.**
- **(a)** Given the components of a $\binom20$ tensor $M^{\alpha\beta}$ as the matrix
$$
\begin{pmatrix}0&1&0&0\\1&-1&0&2\\2&0&0&1\\1&0&-2&0\end{pmatrix},
$$
find: (i) the components of the symmetric tensor $M^{(\alpha\beta)}$ and the antisymmetric tensor $M^{[\alpha\beta]}$; (ii) the components of $M^\alpha{}_\beta$; (iii) the components of $M_\alpha{}^\beta$; (iv) the components of $M_{\alpha\beta}$.
- **(b)** For the $\binom11$ tensor whose components are $M^\alpha{}_\beta$, does it make sense to speak of its symmetric and antisymmetric parts? If so, define them. If not, say why.
- **(c)** Raise an index of the metric tensor to prove $\eta^\alpha{}_\beta = \delta^\alpha{}_\beta$.

Row index = first index $\alpha$, column index = second index $\beta$.

## (a)(i) Symmetric and antisymmetric parts

$M^{(\alpha\beta)} = \tfrac12(M^{\alpha\beta} + M^{\beta\alpha})$, $M^{[\alpha\beta]} = \tfrac12(M^{\alpha\beta} - M^{\beta\alpha})$. The transpose is
$$
M^T = \begin{pmatrix}0&1&2&1\\1&-1&0&0\\0&0&0&-2\\0&2&1&0\end{pmatrix}.
$$
$$
\boxed{M^{(\alpha\beta)} = \begin{pmatrix}0&1&1&\tfrac12\\1&-1&0&1\\1&0&0&-\tfrac12\\\tfrac12&1&-\tfrac12&0\end{pmatrix}},\qquad
\boxed{M^{[\alpha\beta]} = \begin{pmatrix}0&0&-1&-\tfrac12\\0&0&0&1\\1&0&0&\tfrac32\\\tfrac12&-1&-\tfrac32&0\end{pmatrix}}.
$$
Check a few: $(0,2)$: $\tfrac12(0 + 2) = 1$ and $\tfrac12(0 - 2) = -1$; $(2,3)$: $\tfrac12(1 + (-2)) = -\tfrac12$ and $\tfrac12(1 - (-2)) = \tfrac32$. The parts add to $M$.

## (a)(ii)–(iv) Lowering indices

With $\eta = \operatorname{diag}(-1,1,1,1)$, lowering index $0$ multiplies by $-1$, lowering a spatial index does nothing.

**(ii)** $M^\alpha{}_\beta = M^{\alpha\mu}\eta_{\mu\beta}$: lower the **second** index → flip the sign of **column 0**:
$$
\boxed{M^\alpha{}_\beta = \begin{pmatrix}0&1&0&0\\-1&-1&0&2\\-2&0&0&1\\-1&0&-2&0\end{pmatrix}}
$$
**(iii)** $M_\alpha{}^\beta = \eta_{\alpha\mu}M^{\mu\beta}$: lower the **first** index → flip the sign of **row 0**:
$$
\boxed{M_\alpha{}^\beta = \begin{pmatrix}0&-1&0&0\\1&-1&0&2\\2&0&0&1\\1&0&-2&0\end{pmatrix}}
$$
**(iv)** $M_{\alpha\beta} = \eta_{\alpha\mu}\eta_{\beta\nu}M^{\mu\nu}$: flip both row 0 and column 0 (the $(0,0)$ entry flips twice, so it is unchanged):
$$
\boxed{M_{\alpha\beta} = \begin{pmatrix}0&-1&0&0\\-1&-1&0&2\\-2&0&0&1\\-1&0&-2&0\end{pmatrix}}
$$
In matrix language: (ii) $= M\eta$, (iii) $= \eta M$, (iv) $= \eta M\eta$.

## (b) Symmetry of a $\binom11$ tensor?

**Not as such.** Symmetry means invariance under swapping two arguments, $\mathbf T(X, Y) = \mathbf T(Y, X)$. For a $\binom11$ tensor the two arguments are of **different types** (a one-form and a vector), so they can't be swapped. Correspondingly, "$M^\alpha{}_\beta = M^\beta{}_\alpha$" is not a tensor equation: its free indices are at different heights on the two sides, so it can hold in one frame and fail in another.

Concretely: even a $\binom20$ tensor that *is* symmetric, such as $S^{\alpha\beta} = M^{(\alpha\beta)}$ above, gives a $\binom11$ matrix $S^\alpha{}_\beta = S^{\alpha\mu}\eta_{\mu\beta}$ that is generally **not** a symmetric matrix (column 0 flips sign, row 0 doesn't). So "symmetric matrix $M^\alpha{}_\beta$" isn't the right notion.

**With a metric, a meaning can be given:** call the $\binom11$ tensor symmetric if the $\binom20$ (or $\binom02$) tensor obtained by moving its index is symmetric. Moving the index, swapping, and moving it back gives the definitions
$$
M_{(s)}{}^\alpha{}_\beta := \tfrac12\big(M^\alpha{}_\beta + M_\beta{}^\alpha\big),\qquad M_{(A)}{}^\alpha{}_\beta := \tfrac12\big(M^\alpha{}_\beta - M_\beta{}^\alpha\big).
$$
Each is a legitimate $\binom11$ tensor (free indices $\alpha$ up and $\beta$ down on both sides), and they are just $M^{(\alpha\beta)}$ and $M^{[\alpha\beta]}$ from (i) with the second index lowered. Without a metric there is no such notion.

## (c) $\eta^\alpha{}_\beta = \delta^\alpha{}_\beta$

Raise the first index of $\eta_{\mu\beta}$ with the inverse metric:
$$
\eta^\alpha{}_\beta := \eta^{\alpha\mu}\eta_{\mu\beta} = \delta^\alpha{}_\beta ,
$$
the last step being Eq. (3.49) / Problem 19(a): $\eta^{\alpha\mu}$ is the inverse matrix of $\eta_{\mu\beta}$. $\blacksquare$ So the metric with one index up is the identity $\binom11$ tensor (Eq. 3.59–3.60), which is why raising then lowering an index returns the original components.

```python
import sympy as sp
M = sp.Matrix([[0, 1, 0, 0], [1, -1, 0, 2], [2, 0, 0, 1], [1, 0, -2, 0]])
eta = sp.diag(-1, 1, 1, 1)
S, A = (M + M.T) / 2, (M - M.T) / 2
for name, X in [("M^(αβ)", S), ("M^[αβ]", A), ("M^α_β = Mη", M*eta), ("M_α^β = ηM", eta*M), ("M_αβ = ηMη", eta*M*eta)]:
    print(name); display(X)
print("S + A = M:", S + A == M)
print("(b) S^{αβ} symmetric, but S^α_β = Sη symmetric as a matrix?", S*eta == (S*eta).T)
print("    sym part ½(M^α_β + M_β^α) = (Sη):", (M*eta + (eta*M).T) / 2 == S*eta)
print("(c) η^{αμ}η_{μβ} =", eta.inv()*eta == sp.eye(4))
```
