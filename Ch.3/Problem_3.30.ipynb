# Chapter 3, Problem 30 — calculus with vector fields

**Problem.** In some frame $\mathcal O$, the vector fields $\vec U$ and $\vec D$ have components
$$
\vec U \to (1 + t^2,\ t^2,\ \sqrt2\,t,\ 0),\qquad \vec D \to (x,\ 5tx,\ \sqrt2\,t,\ 0),
$$
and the scalar $\rho$ has the value $\rho = x^2 + t^2 - y^2$.
- **(a)** Find $\vec U\cdot\vec U$, $\vec U\cdot\vec D$, $\vec D\cdot\vec D$. Is $\vec U$ suitable as a four-velocity field? Is $\vec D$?
- **(b)** Find the spatial velocity $\vec v$ of a particle whose four-velocity is $\vec U$, for arbitrary $t$. What happens to it in the limits $t \to 0$, $t \to \infty$?
- **(c)** Find $U_\alpha$ for all $\alpha$.
- **(d)** Find $U^\alpha{}_{,\beta}$ for all $\alpha, \beta$.
- **(e)** Show that $U_\alpha U^\alpha{}_{,\beta} = 0$ for all $\beta$. Show that $U^\alpha U_{\alpha,\beta} = 0$ for all $\beta$.
- **(f)** Find $D^\beta{}_{,\beta}$.
- **(g)** Find $(U^\alpha D^\beta)_{,\beta}$ for all $\alpha$.
- **(h)** Find $U_\alpha(U^\alpha D^\beta)_{,\beta}$ and compare with (f) above. Why are the two answers similar?
- **(i)** Find $\rho_{,\alpha}$ for all $\alpha$. Find $\rho^{,\alpha}$ for all $\alpha$. (Recall that $\rho^{,\alpha} := \eta^{\alpha\beta}\rho_{,\beta}$.) What are the numbers $\{\rho^{,\alpha}\}$ the components of?
- **(j)** Find $\nabla_{\vec U}\rho$, $\nabla_{\vec U}\vec D$, $\nabla_{\vec D}\rho$, $\nabla_{\vec D}\vec U$.

Coordinates $(x^0, x^1, x^2, x^3) = (t, x, y, z)$, metric $\eta = \operatorname{diag}(-1,1,1,1)$. Every answer below is re-derived symbolically in the code cell at the end.

## (a) Inner products

$$
\vec U\cdot\vec U = -(1 + t^2)^2 + t^4 + 2t^2 = -(1 + 2t^2 + t^4) + t^4 + 2t^2 = \boxed{-1}.
$$
$$
\vec U\cdot\vec D = -(1 + t^2)x + t^2(5tx) + (\sqrt2t)(\sqrt2t) = \boxed{-(1 + t^2)x + 5t^3x + 2t^2}.
$$
$$
\vec D\cdot\vec D = -x^2 + 25t^2x^2 + 2t^2 .
$$
- **$\vec U$ is a valid four-velocity field:** $\vec U\cdot\vec U = -1$ everywhere and $U^0 = 1 + t^2 > 0$ (timelike, future-pointing, unit).
- **$\vec D$ is not:** $\vec D\cdot\vec D$ is not $-1$ in general; e.g. at $t = 0$ it's $-x^2$, and at $x = 0$ it's $2t^2 > 0$ (spacelike). Its time component $x$ can also be negative or zero.

## (b) Spatial velocity

$U^\alpha = \gamma(1, \vec v)$, so $v^i = U^i/U^0$:
$$
\vec v = \left(\frac{t^2}{1 + t^2},\ \frac{\sqrt2\,t}{1 + t^2},\ 0\right),\qquad
v^2 = \frac{t^4 + 2t^2}{(1 + t^2)^2} = 1 - \frac{1}{(1 + t^2)^2}.
$$
- $t \to 0$: $\vec v \to 0$; the particle is momentarily at rest ($\gamma = 1 + t^2 \to 1$).
- $t \to \infty$: $v^x \to 1$, $v^y \approx \sqrt2/t \to 0$, so $\vec v \to (1, 0, 0)$: it approaches the speed of light in the $x$ direction (with $\gamma = 1 + t^2 \to \infty$), but $v < 1$ at every finite $t$.

## (c) Lower the index

$U_\alpha = \eta_{\alpha\beta}U^\beta$: flip the time component.
$$
U_\alpha = \big(-(1 + t^2),\ t^2,\ \sqrt2\,t,\ 0\big).
$$

## (d) Derivatives $U^\alpha{}_{,\beta}$

$\vec U$ depends only on $t$, so only $\beta = 0$ is nonzero:
$$
U^\alpha{}_{,0} = (2t,\ 2t,\ \sqrt2,\ 0),\qquad U^\alpha{}_{,1} = U^\alpha{}_{,2} = U^\alpha{}_{,3} = 0 .
$$
As a matrix (row $\alpha$, column $\beta$): first column $(2t, 2t, \sqrt2, 0)^T$, the rest zero.

## (e) $U_\alpha U^\alpha{}_{,\beta} = 0$ and $U^\alpha U_{\alpha,\beta} = 0$

Direct: for $\beta = 0$, $U_\alpha U^\alpha{}_{,0} = -(1 + t^2)(2t) + t^2(2t) + \sqrt2t\cdot\sqrt2 = -2t - 2t^3 + 2t^3 + 2t = 0$; for $\beta \neq 0$ every term has a zero factor.

Why: differentiate $U_\alpha U^\alpha = -1$:
$$
0 = (U_\alpha U^\alpha)_{,\beta} = U_{\alpha,\beta}U^\alpha + U_\alpha U^\alpha{}_{,\beta} = 2\,U_\alpha U^\alpha{}_{,\beta},
$$
using $U_{\alpha,\beta}U^\alpha = \eta_{\alpha\mu}U^\mu{}_{,\beta}U^\alpha = U_\mu U^\mu{}_{,\beta}$ ($\eta$ is constant). So both contractions vanish and are equal. Physically, the four-acceleration is orthogonal to the four-velocity (cf. Problem 2.27).

## (f) Divergence of $\vec D$

$$
D^\beta{}_{,\beta} = \partial_t(x) + \partial_x(5tx) + \partial_y(\sqrt2t) + \partial_z(0) = 0 + 5t + 0 + 0 = \boxed{5t}.
$$

## (g) $(U^\alpha D^\beta)_{,\beta}$

Product rule (Problem 29): $(U^\alpha D^\beta)_{,\beta} = U^\alpha{}_{,\beta}D^\beta + U^\alpha D^\beta{}_{,\beta}$. The first term has only $\beta = 0$: $U^\alpha{}_{,0}D^0 = x\,(2t, 2t, \sqrt2, 0)$. The second is $5t\,U^\alpha$:
$$
(U^\alpha D^\beta)_{,\beta} = \big(2tx + 5t(1 + t^2),\ \ 2tx + 5t^3,\ \ \sqrt2x + 5\sqrt2t^2,\ \ 0\big).
$$

## (h) $U_\alpha(U^\alpha D^\beta)_{,\beta}$

$$
U_\alpha(U^\alpha D^\beta)_{,\beta} = \underbrace{U_\alpha U^\alpha{}_{,\beta}}_{0\text{ by (e)}}D^\beta + \underbrace{U_\alpha U^\alpha}_{-1}D^\beta{}_{,\beta} = \boxed{-5t} = -D^\beta{}_{,\beta}.
$$
(Direct substitution of (c) and (g) confirms it.) The answers agree up to sign because $\vec U$ is a **unit** vector field: projecting onto $\vec U$ strips off $\vec U$ (giving $U\cdot U = -1$), and the derivative of $\vec U$ is orthogonal to $\vec U$, so it drops out.

## (i) Gradient of $\rho$

$$
\rho_{,\alpha} = (\partial_t\rho, \partial_x\rho, \partial_y\rho, \partial_z\rho) = (2t,\ 2x,\ -2y,\ 0),\qquad
\rho^{,\alpha} = \eta^{\alpha\beta}\rho_{,\beta} = (-2t,\ 2x,\ -2y,\ 0).
$$
$\{\rho_{,\alpha}\}$ are the components of the one-form $\tilde d\rho$. $\{\rho^{,\alpha}\}$ are the components of the **vector** associated with $\tilde d\rho$ by the metric, the gradient vector $\vec\nabla\rho = \mathbf g^{-1}(\tilde d\rho)$, normal to the surfaces $\rho = $ const.

## (j) Directional derivatives

$\nabla_{\vec V}f = f_{,\alpha}V^\alpha$; $(\nabla_{\vec V}\vec W)^\beta = W^\beta{}_{,\alpha}V^\alpha$ (Eq. 3.68).

$$
\nabla_{\vec U}\rho = 2t(1 + t^2) + 2x\,t^2 + (-2y)\sqrt2t = \boxed{2t + 2t^3 + 2t^2x - 2\sqrt2\,ty}.
$$
$\nabla_{\vec U}\vec D$: the nonzero derivatives of $\vec D$ are $D^0{}_{,x} = 1$, $D^1{}_{,t} = 5x$, $D^1{}_{,x} = 5t$, $D^2{}_{,t} = \sqrt2$:
$$
\nabla_{\vec U}\vec D \to \big(U^x,\ 5xU^t + 5tU^x,\ \sqrt2\,U^t,\ 0\big) = \boxed{\big(t^2,\ 5x(1 + t^2) + 5t^3,\ \sqrt2(1 + t^2),\ 0\big)}.
$$
$$
\nabla_{\vec D}\rho = 2t\cdot x + 2x\cdot5tx + (-2y)\sqrt2t = \boxed{2tx + 10tx^2 - 2\sqrt2\,ty}.
$$
$$
\nabla_{\vec D}\vec U \to U^\beta{}_{,0}D^0 = \boxed{x\,(2t,\ 2t,\ \sqrt2,\ 0)}.
$$
Note $\nabla_{\vec U}\vec D \neq \nabla_{\vec D}\vec U$; the difference is the commutator (Lie bracket) of the two fields, which appears in Chapter 5.

```python
import sympy as sp
t, x, y, z = sp.symbols("t x y z", real=True); X = [t, x, y, z]
eta = sp.diag(-1, 1, 1, 1)
U = sp.Matrix([1 + t**2, t**2, sp.sqrt(2)*t, 0]); D = sp.Matrix([x, 5*t*x, sp.sqrt(2)*t, 0]); rho = x**2 + t**2 - y**2
dot = lambda A, B: sp.expand((A.T * eta * B)[0])
print("(a) U·U =", sp.simplify(dot(U, U)), "| U·D =", dot(U, D), "| D·D =", dot(D, D))
v = sp.simplify(U[1:, 0] / U[0]); print("(b) v =", list(v), " t→0:", [sp.limit(c, t, 0) for c in v], " t→∞:", [sp.limit(c, t, sp.oo) for c in v])
Ulo = eta * U; print("(c) U_α =", list(Ulo))
dU = sp.Matrix(4, 4, lambda a, b: sp.diff(U[a], X[b])); print("(d) U^α_,β ="); display(dU)
dUlo = sp.Matrix(4, 4, lambda a, b: sp.diff(Ulo[a], X[b]))
print("(e) U_α U^α_,β =", list(sp.simplify(Ulo.T * dU)), "  U^α U_α,β =", list(sp.simplify(U.T * dUlo)))
divD = sum(sp.diff(D[b], X[b]) for b in range(4)); print("(f) D^β_,β =", divD)
g = [sp.expand(sum(sp.diff(U[a]*D[b], X[b]) for b in range(4))) for a in range(4)]; print("(g)", g)
print("(h) U_α(U^α D^β)_,β =", sp.simplify(sum(Ulo[a]*g[a] for a in range(4))))
drho = sp.Matrix([sp.diff(rho, c) for c in X]); print("(i) ρ_,α =", list(drho), " ρ^,α =", list(eta.inv() * drho))
dD = sp.Matrix(4, 4, lambda a, b: sp.diff(D[a], X[b]))
print("(j) ∇_U ρ =", sp.expand((drho.T * U)[0]))
print("    ∇_U D =", [sp.expand(c) for c in dD * U])
print("    ∇_D ρ =", sp.expand((drho.T * D)[0]))
print("    ∇_D U =", [sp.factor(c) for c in dU * D])
```
