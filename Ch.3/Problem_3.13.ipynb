# Chapter 3, Problem 13 — $\tilde df$ is normal to surfaces of constant $f$

**Problem.** Prove, by geometric or algebraic arguments, that $\tilde df$ is normal to surfaces of constant $f$.

**What must be shown.** A one-form is normal to a surface if it gives zero on every vector tangent to that surface. So we must show: for every vector $\vec V$ tangent to a surface $f = $ const, $\tilde df(\vec V) = 0$.

## Algebraic proof

Let $\vec V$ be tangent to the surface $f = f_0$ at a point $\mathcal P$. Then there is a curve $x^\alpha(\lambda)$ lying **in the surface**, passing through $\mathcal P$ at $\lambda = 0$, with tangent $dx^\alpha/d\lambda = V^\alpha$ there. (That's what "tangent to the surface" means.)

Along that curve $f$ is constant: $f\big(x^\alpha(\lambda)\big) = f_0$ for all $\lambda$. Differentiate with the chain rule:
$$
0 = \frac{df}{d\lambda} = \frac{\partial f}{\partial x^\alpha}\,\frac{dx^\alpha}{d\lambda} = f_{,\alpha}\,V^\alpha = \tilde df(\vec V).
$$
The last step is Eq. (3.14)/(3.15): $\tilde df(\vec V)$ is the rate of change of $f$ along $\vec V$. So $\tilde df(\vec V) = 0$ for every tangent $\vec V$: $\tilde df$ is normal. $\blacksquare$

## Geometric proof

Picture $\tilde df$ at $\mathcal P$ as its stack of surfaces (§3.3, Fig. 3.4). Those surfaces are the level surfaces $f = f_0,\ f_0 + 1,\ \dots$ approximated near $\mathcal P$ by their tangent planes. $\tilde df(\vec V)$ is the number of them $\vec V$ pierces.

A vector tangent to $f = f_0$ lies **in** that surface (in the linear approximation), so it runs parallel to all the others in the stack and pierces none. Value: zero. $\blacksquare$

## Remarks

- If $\tilde df \neq 0$ at $\mathcal P$, then by Problem 12(c) every other normal is a multiple of $\tilde df$. And by 12(b), $\tilde df$ is positive on vectors pointing toward **increasing** $f$.
- No metric was used. Normality is a relation between a one-form and a surface. Only to turn $\tilde df$ into a normal *vector* do we need $\vec{\nabla f} = g^{-1}(\tilde df)$ (§3.5). In Euclidean space that is the familiar fact that $\nabla f$ is perpendicular to level surfaces.

**Check** with $f = -t^2 + x^2 + y^2$ (level surfaces are hyperboloids). At a point on $f = 1$, construct two tangent vectors and evaluate $\tilde df$ on them.

```python
import sympy as sp
t, x, y, s, u = sp.symbols("t x y s u", real=True)
X = [t, x, y]
f = -t**2 + x**2 + y**2
df = [sp.diff(f, c) for c in X]
# parametrize the surface f = 1: t = sinh s, x = cosh s cos u, y = cosh s sin u
param = {t: sp.sinh(s), x: sp.cosh(s)*sp.cos(u), y: sp.cosh(s)*sp.sin(u)}
print("f on the surface:", sp.simplify(f.subs(param)))
for name, q in (("∂/∂s", s), ("∂/∂u", u)):
    V = [sp.diff(param[c], q) for c in X]                # tangent vectors to the surface
    val = sp.simplify(sum(d.subs(param) * v for d, v in zip(df, V)))
    print(f"d̃f(tangent {name}) =", val)
```
