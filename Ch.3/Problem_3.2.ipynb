# Chapter 3, Problem 2 — one-forms form a vector space

**Problem.** Prove that the set of all one-forms is a vector space.

**Definitions used (§3.3).** A one-form is a linear real-valued function of vectors. For one-forms $\tilde p, \tilde q$ and a real number $\alpha$, Eqs. (3.6a,b) define
$$
(\tilde p + \tilde q)(\vec A) := \tilde p(\vec A) + \tilde q(\vec A),\qquad (\alpha\tilde p)(\vec A) := \alpha\,\tilde p(\vec A)\qquad\text{for every vector }\vec A .
$$
Two one-forms are equal if they give the same value on every vector.

**What must be shown.** (0) Closure: the sum and the multiple are again one-forms (i.e. linear). Then the eight vector-space axioms: (1) $\tilde p + \tilde q = \tilde q + \tilde p$; (2) $(\tilde p + \tilde q) + \tilde r = \tilde p + (\tilde q + \tilde r)$; (3) there is a zero $\tilde 0$; (4) every $\tilde p$ has a negative; (5) $\alpha(\beta\tilde p) = (\alpha\beta)\tilde p$; (6) $1\tilde p = \tilde p$; (7) $\alpha(\tilde p + \tilde q) = \alpha\tilde p + \alpha\tilde q$; (8) $(\alpha + \beta)\tilde p = \alpha\tilde p + \beta\tilde p$.

The strategy for every axiom: evaluate both sides on an arbitrary vector $\vec A$. The values are real numbers, and real numbers obey the corresponding rule.

## (0) Closure

**Sum.** For vectors $\vec A, \vec B$ and numbers $a, b$:
$$
(\tilde p + \tilde q)(a\vec A + b\vec B) = \tilde p(a\vec A + b\vec B) + \tilde q(a\vec A + b\vec B) = a\tilde p(\vec A) + b\tilde p(\vec B) + a\tilde q(\vec A) + b\tilde q(\vec B) = a(\tilde p + \tilde q)(\vec A) + b(\tilde p + \tilde q)(\vec B).
$$
So $\tilde p + \tilde q$ is linear: a one-form. (The first and last steps use the definition (3.6b), the middle one linearity of $\tilde p$ and $\tilde q$.)

**Multiple.** $(\alpha\tilde p)(a\vec A + b\vec B) = \alpha\big(a\tilde p(\vec A) + b\tilde p(\vec B)\big) = a(\alpha\tilde p)(\vec A) + b(\alpha\tilde p)(\vec B)$: linear ✓.

## The axioms (each evaluated on an arbitrary $\vec A$)

1. **Commutativity:** $(\tilde p + \tilde q)(\vec A) = \tilde p(\vec A) + \tilde q(\vec A) = \tilde q(\vec A) + \tilde p(\vec A) = (\tilde q + \tilde p)(\vec A)$.
2. **Associativity:** $((\tilde p + \tilde q) + \tilde r)(\vec A) = (\tilde p(\vec A) + \tilde q(\vec A)) + \tilde r(\vec A) = \tilde p(\vec A) + (\tilde q(\vec A) + \tilde r(\vec A)) = (\tilde p + (\tilde q + \tilde r))(\vec A)$.
3. **Zero:** define $\tilde 0(\vec A) := 0$ for all $\vec A$ (it's linear). Then $(\tilde p + \tilde 0)(\vec A) = \tilde p(\vec A) + 0 = \tilde p(\vec A)$.
4. **Negatives:** $-\tilde p := (-1)\tilde p$. Then $(\tilde p + (-\tilde p))(\vec A) = \tilde p(\vec A) - \tilde p(\vec A) = 0 = \tilde 0(\vec A)$.
5. **Compatibility:** $(\alpha(\beta\tilde p))(\vec A) = \alpha\,(\beta\,\tilde p(\vec A)) = (\alpha\beta)\,\tilde p(\vec A) = ((\alpha\beta)\tilde p)(\vec A)$.
6. **Identity:** $(1\tilde p)(\vec A) = 1\cdot\tilde p(\vec A) = \tilde p(\vec A)$.
7. **Distributivity over one-form addition:** $(\alpha(\tilde p + \tilde q))(\vec A) = \alpha(\tilde p(\vec A) + \tilde q(\vec A)) = \alpha\tilde p(\vec A) + \alpha\tilde q(\vec A) = (\alpha\tilde p + \alpha\tilde q)(\vec A)$.
8. **Distributivity over number addition:** $((\alpha + \beta)\tilde p)(\vec A) = (\alpha + \beta)\tilde p(\vec A) = \alpha\tilde p(\vec A) + \beta\tilde p(\vec A) = (\alpha\tilde p + \beta\tilde p)(\vec A)$.

Each equality holds for every $\vec A$, so the one-forms on the two sides are equal. All eight axioms hold, so the one-forms form a vector space: the **dual space** of the space of vectors. $\blacksquare$

**In components** ($p_\alpha = \tilde p(\vec e_\alpha)$, Eq. 3.7), the operations are component-wise: $(\tilde p + \tilde q)_\alpha = p_\alpha + q_\alpha$, $(\alpha\tilde p)_\beta = \alpha p_\beta$. So the dual space is a copy of $\mathbb R^4$, of dimension 4. A basis is $\{\tilde\omega^\alpha\}$ (Eq. 3.12).

```python
# Quick numerical spot-check: represent one-forms by component rows, act on vectors by contraction
import numpy as np
rng = np.random.default_rng(2)
p, q, r = rng.normal(size=(3, 4)); A = rng.normal(size=4)
a, b = 1.7, -0.4
act = lambda form, vec: form @ vec
checks = {
    "commutative":  act(p + q, A) - act(q + p, A),
    "associative":  act((p + q) + r, A) - act(p + (q + r), A),
    "a(bp) = (ab)p": act(a*(b*p), A) - act((a*b)*p, A),
    "a(p+q)":       act(a*(p + q), A) - (a*act(p, A) + a*act(q, A)),
    "(a+b)p":       act((a + b)*p, A) - (a*act(p, A) + b*act(p, A)),
}
for k, v in checks.items(): print(f"{k:14s} residual = {v:+.1e}")
```
