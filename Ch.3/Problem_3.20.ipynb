# Chapter 3, Problem 20 — why Euclidean vectors and one-forms look alike

**Problem.** In Euclidean three-space in Cartesian coordinates, we don't normally distinguish between vectors and one-forms, because their components transform identically. Prove this in two steps.
- **(a)** Show that $A^{\bar\alpha} = \Lambda^{\bar\alpha}{}_\beta A^\beta$ and $P_{\bar\beta} = \Lambda^\alpha{}_{\bar\beta}P_\alpha$ are the same transformation if the matrix $\{\Lambda^{\bar\alpha}{}_\beta\}$ equals the transpose of its inverse. Such a matrix is said to be orthogonal.
- **(b)** The metric of such a space has components $\{\delta_{ij},\ i, j = 1, \dots, 3\}$. Prove that a transformation from one Cartesian coordinate system to another must obey
$$
\delta_{\bar i\bar j} = \Lambda^k{}_{\bar i}\,\Lambda^l{}_{\bar j}\,\delta_{kl}
$$
and that this implies $\{\Lambda^k{}_{\bar i}\}$ is an orthogonal matrix. See Exer. 32 for the analog of this in SR.

## (a) When the two laws coincide

Name the matrices: $L$ with entries $L_{\bar\alpha\beta} := \Lambda^{\bar\alpha}{}_\beta$ (row $\bar\alpha$, column $\beta$), and its inverse $M := L^{-1}$ with entries $M_{\alpha\bar\beta} := \Lambda^\alpha{}_{\bar\beta}$ (Eq. 2.18 / Problem 10).

- Vectors: $A^{\bar\alpha} = L_{\bar\alpha\beta}A^\beta$, i.e. $\bar A = L\,A$ (column vectors).
- One-forms: $P_{\bar\beta} = M_{\alpha\bar\beta}P_\alpha = (M^T)_{\bar\beta\alpha}P_\alpha$, i.e. $\bar P = M^T P = (L^{-1})^T P$.

So the one-form law is multiplication by $(L^{-1})^T$, the vector law by $L$. They are the **same** linear map of component lists if and only if
$$
L = (L^{-1})^T \iff L^TL = I \iff L \text{ is orthogonal.} \qquad\blacksquare
$$
When this holds, a list of three numbers transforms the same way whether we call it a vector or a one-form, so no one bothers to distinguish them.

## (b) Cartesian-to-Cartesian transformations are orthogonal

**Metric components transform as a $\binom02$ tensor.** By definition $g_{\bar i\bar j} = \mathbf g(\vec e_{\bar i}, \vec e_{\bar j})$. Insert $\vec e_{\bar i} = \Lambda^k{}_{\bar i}\vec e_k$ and use bilinearity:
$$
g_{\bar i\bar j} = \mathbf g\big(\Lambda^k{}_{\bar i}\vec e_k,\ \Lambda^l{}_{\bar j}\vec e_l\big) = \Lambda^k{}_{\bar i}\Lambda^l{}_{\bar j}\,\mathbf g(\vec e_k, \vec e_l) = \Lambda^k{}_{\bar i}\Lambda^l{}_{\bar j}\,g_{kl}.
$$
**Both systems are Cartesian**, meaning orthonormal: $g_{kl} = \delta_{kl}$ and $g_{\bar i\bar j} = \delta_{\bar i\bar j}$. Substituting,
$$
\delta_{\bar i\bar j} = \Lambda^k{}_{\bar i}\,\Lambda^l{}_{\bar j}\,\delta_{kl}. \qquad\checkmark
$$
**Orthogonality.** The delta collapses the sum over $l$: $\delta_{\bar i\bar j} = \sum_k\Lambda^k{}_{\bar i}\Lambda^k{}_{\bar j} = \sum_k (M^T)_{\bar ik}M_{k\bar j}$, i.e.
$$
M^TM = I .
$$
So $M = \{\Lambda^k{}_{\bar i}\}$ is orthogonal. Then its inverse $L = M^{-1} = M^T$ is orthogonal too ($L^TL = MM^T = I$, since a left inverse of a square matrix is also a right inverse), and by part (a) vectors and one-forms transform identically. $\blacksquare$

**The SR analog (Exer. 32).** In Minkowski space the same argument gives $\eta_{\bar\alpha\bar\beta} = \Lambda^\mu{}_{\bar\alpha}\Lambda^\nu{}_{\bar\beta}\eta_{\mu\nu}$, i.e. $M^T\eta M = \eta$, which is *not* orthogonality. A boost isn't orthogonal, so vector and one-form components transform differently, and we must keep track of index positions.

```python
import sympy as sp
th, ph = sp.symbols("theta phi", real=True)
R = sp.rot_axis3(th) * sp.rot_axis1(ph)           # a general-ish rotation
print("(a) rotation: L == (L^{-1})^T :", sp.simplify(R - R.inv().T) == sp.zeros(3, 3))
M = R.inv()
print("(b) M^T δ M = δ :", sp.simplify(M.T * sp.eye(3) * M) == sp.eye(3))
from schutz_gr import boost_matrix
v = sp.symbols("v", real=True)
L = boost_matrix(v)
print("    a boost is NOT orthogonal: L − (L^{-1})^T =", sp.simplify(L - L.inv().T)[0:2, 0:2])
```
