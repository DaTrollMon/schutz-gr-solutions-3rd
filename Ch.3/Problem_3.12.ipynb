# Chapter 3, Problem 12 — normal one-forms to a plane

**Problem.** Let $S$ be the two-dimensional plane $x = 0$ in three-dimensional Euclidean space. Let $\tilde n \neq 0$ be a normal one-form to $S$.
- **(a)** Show that if $\vec V$ is a vector which is not tangent to $S$, then $\tilde n(\vec V) \neq 0$.
- **(b)** Show that if $\tilde n(\vec V) > 0$, then $\tilde n(\vec W) > 0$ for any $\vec W$ which points toward the same side of $S$ as $\vec V$ does (i.e. any $\vec W$ whose $x$ component has the same sign as $V^x$).
- **(c)** Show that any normal to $S$ is a multiple of $\tilde n$.
- **(d)** Generalize these statements to an arbitrary three-dimensional surface in four-dimensional spacetime.

> **Text note.** The extracted text dropped the "≠" signs ("$\tilde n = 0$", "$\tilde n(\vec V) = 0$"); the statements only make sense with "≠", as restored above.

**Definition used (§3.3).** A one-form is *normal* to a surface if its value is zero on every vector tangent to the surface.

## Setting up: what a normal one-form to $x = 0$ looks like

Use Cartesian coordinates $(x, y, z)$ and the dual basis $\tilde dx, \tilde dy, \tilde dz$. Write $\tilde n = n_x\,\tilde dx + n_y\,\tilde dy + n_z\,\tilde dz$.

The tangent vectors to the plane $x = 0$ are exactly the vectors with no $x$ component: $\vec T = T^y\vec e_y + T^z\vec e_z$. In particular $\vec e_y$ and $\vec e_z$ are tangent, so normality requires
$$
\tilde n(\vec e_y) = n_y = 0,\qquad \tilde n(\vec e_z) = n_z = 0 .
$$
Conversely, if $n_y = n_z = 0$ then $\tilde n(\vec T) = n_yT^y + n_zT^z = 0$ for every tangent $\vec T$. So
$$
\tilde n \text{ normal to } S \iff \tilde n = n_x\,\tilde dx ,
$$
and $\tilde n \neq 0$ means $n_x \neq 0$.

## (a) Non-tangent vectors give a nonzero value

$\tilde n(\vec V) = n_x V^x$. "Not tangent to $S$" means $V^x \neq 0$. Both factors are nonzero, so $\tilde n(\vec V) \neq 0$. $\blacksquare$

(Geometrically: the surfaces of $\tilde n$ are the planes $x = $ const, parallel to $S$. A vector that leaves $S$ must cross some of them.)

## (b) Same side ⟹ same sign

$\tilde n(\vec V) = n_xV^x > 0$ means $n_x$ and $V^x$ have the same sign. If $W^x$ has the same sign as $V^x$, it also has the same sign as $n_x$, so $\tilde n(\vec W) = n_xW^x > 0$. $\blacksquare$

So a normal one-form distinguishes the two sides of $S$. That's what makes "outward normal" meaningful (Problem 21), even with no metric.

## (c) All normals are proportional

Let $\tilde m$ be any normal to $S$. By the argument above, $\tilde m = m_x\,\tilde dx$. Since $n_x \neq 0$,
$$
\tilde m = \frac{m_x}{n_x}\,\big(n_x\,\tilde dx\big) = \frac{m_x}{n_x}\,\tilde n . \qquad\blacksquare
$$
The space of normals to a 2-surface in 3-space is one-dimensional: tangency imposes $2$ conditions on $3$ components.

## (d) Generalization: a 3-surface in 4-d spacetime

Let $\Sigma$ be a smooth three-dimensional hypersurface in spacetime and $\mathcal P$ a point on it. The tangent vectors at $\mathcal P$ form a 3-dimensional subspace $T$ of the 4-dimensional space of vectors. Pick a basis $\vec t_1, \vec t_2, \vec t_3$ of $T$ and a fourth vector $\vec t_0 \notin T$, so that $\{\vec t_0, \vec t_1, \vec t_2, \vec t_3\}$ is a basis. A normal one-form satisfies $\tilde n(\vec t_i) = 0$, $i = 1, 2, 3$, so on the dual basis it has only one component: $\tilde n = n_0\,\tilde\theta^0$.

- **(a′)** If $\tilde n \neq 0$ is normal to $\Sigma$ at $\mathcal P$ and $\vec V$ is not tangent to $\Sigma$, then $\tilde n(\vec V) \neq 0$. (Write $\vec V = V^0\vec t_0 + \vec T$ with $\vec T \in T$; then $\tilde n(\vec V) = n_0V^0$, and $V^0 \neq 0$.)
- **(b′)** If $\tilde n(\vec V) > 0$, then $\tilde n(\vec W) > 0$ for every $\vec W$ pointing to the same side of $\Sigma$ as $\vec V$ (same sign of the $\vec t_0$-component).
- **(c′)** Any two normal one-forms to $\Sigma$ at $\mathcal P$ are proportional. The normal is unique up to a factor, which is why it can be normalized to a *unit* normal ($|\tilde n|^2 = \pm1$) when it is not null.

None of this uses the metric: normality, sidedness, and uniqueness up to scale are all metric-free properties of one-forms. If $\Sigma$ is a level surface $f = $ const, then $\tilde df$ is such a normal (Problem 13).

Note that in spacetime a normal one-form can be timelike, spacelike, or **null** (Problem 21b). A null normal's associated vector is *tangent* to the surface it's normal to, which never happens in Euclidean space.

```python
import sympy as sp
# (c′) in 4-d: the normals to a random 3-plane form a 1-dimensional space
sp.var("a0:4 b0:4")
t1, t2, t3 = sp.Matrix([1, 2, 0, -1]), sp.Matrix([0, 1, 1, 3]), sp.Matrix([2, 0, 5, 1])
Tmat = sp.Matrix.hstack(t1, t2, t3).T      # each row: a tangent vector; normals n satisfy Tmat·n = 0
ns = Tmat.nullspace()
print("dimension of the space of normal one-forms:", len(ns), " basis:", list(ns[0].T))
V = sp.Matrix([1, 0, 0, 0])
print("n(V) for a non-tangent V:", (ns[0].T * V)[0], "≠ 0")
```
