# Chapter 3, Problem 26 — contractions with symmetric and antisymmetric tensors

**Problem.** Suppose $\mathsf A$ is an antisymmetric $\binom20$ tensor, $\mathbf B$ a symmetric $\binom02$ tensor, $\mathbf C$ an arbitrary $\binom02$ tensor, and $\mathsf D$ an arbitrary $\binom20$ tensor. Prove:
- **(a)** $A^{\alpha\beta}B_{\alpha\beta} = 0$;
- **(b)** $A^{\alpha\beta}C_{\alpha\beta} = A^{\alpha\beta}C_{[\alpha\beta]}$;
- **(c)** $B_{\alpha\beta}D^{\alpha\beta} = B_{\alpha\beta}D^{(\alpha\beta)}$.

**Two facts used throughout.**
1. Dummy indices can be renamed: $A^{\alpha\beta}B_{\alpha\beta} = A^{\beta\alpha}B_{\beta\alpha}$ (same sum, letters swapped).
2. $A^{\beta\alpha} = -A^{\alpha\beta}$, $B_{\beta\alpha} = B_{\alpha\beta}$.

## (a) Antisymmetric · symmetric = 0

Let $X := A^{\alpha\beta}B_{\alpha\beta}$. Rename the dummies $\alpha \leftrightarrow \beta$, then use the symmetries:
$$
X = A^{\beta\alpha}B_{\beta\alpha} = (-A^{\alpha\beta})(B_{\alpha\beta}) = -X .
$$
So $2X = 0$, $X = 0$. $\blacksquare$

Term by term: the diagonal terms vanish ($A^{\alpha\alpha} = 0$) and each off-diagonal pair cancels, $A^{01}B_{01} + A^{10}B_{10} = A^{01}B_{01} - A^{01}B_{01} = 0$.

## (b) Only the antisymmetric part of $\mathbf C$ survives

Split $C_{\alpha\beta} = C_{(\alpha\beta)} + C_{[\alpha\beta]}$ (Eq. 3.35). $C_{(\alpha\beta)}$ is symmetric, so by (a) $A^{\alpha\beta}C_{(\alpha\beta)} = 0$, leaving
$$
A^{\alpha\beta}C_{\alpha\beta} = A^{\alpha\beta}C_{(\alpha\beta)} + A^{\alpha\beta}C_{[\alpha\beta]} = A^{\alpha\beta}C_{[\alpha\beta]}. \qquad\blacksquare
$$

## (c) Only the symmetric part of $\mathsf D$ survives

Split $D^{\alpha\beta} = D^{(\alpha\beta)} + D^{[\alpha\beta]}$. The argument of (a) with the index heights reversed gives $B_{\alpha\beta}D^{[\alpha\beta]} = 0$ (rename: $Y = B_{\beta\alpha}D^{[\beta\alpha]} = B_{\alpha\beta}(-D^{[\alpha\beta]}) = -Y$). Hence
$$
B_{\alpha\beta}D^{\alpha\beta} = B_{\alpha\beta}D^{(\alpha\beta)}. \qquad\blacksquare
$$

**Why this matters later.** Results like these are used constantly: e.g. contracting the symmetric metric or stress-energy tensor with something kills its antisymmetric part, and in (b) a $\binom02$ tensor contracted with an antisymmetric one (like the electromagnetic $F^{\alpha\beta}$) only "sees" its antisymmetric part.

```python
import sympy as sp
sym = lambda name, kind: sp.Matrix(4, 4, lambda i, j: sp.Symbol(f"{name}{min(i,j)}{max(i,j)}") * (1 if kind == "s" or i <= j else -1) * (0 if kind == "a" and i == j else 1))
A, B = sym("a", "a"), sym("b", "s")
C = sp.Matrix(4, 4, lambda i, j: sp.Symbol(f"c{i}{j}")); D = sp.Matrix(4, 4, lambda i, j: sp.Symbol(f"d{i}{j}"))
full = lambda X, Y: sp.expand(sum(X[i, j] * Y[i, j] for i in range(4) for j in range(4)))
print("A antisymmetric:", A == -A.T, "| B symmetric:", B == B.T)
print("(a)", full(A, B))
print("(b)", full(A, C) - full(A, (C - C.T) / 2))
print("(c)", full(B, D) - full(B, (D + D.T) / 2))
```
