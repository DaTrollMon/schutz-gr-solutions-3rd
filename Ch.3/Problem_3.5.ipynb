# Chapter 3, Problem 5 — why the contraction $A^\alpha p_\alpha$ is frame-independent

**Problem.** Justify each step leading from Eq. (3.10a) to Eq. (3.10d):
$$
\begin{aligned}
A^{\bar\alpha}p_{\bar\alpha} &= \big(\Lambda^{\bar\alpha}{}_\beta A^\beta\big)\big(\Lambda^\mu{}_{\bar\alpha}p_\mu\big) &(3.10a)\\
&= \Lambda^\mu{}_{\bar\alpha}\Lambda^{\bar\alpha}{}_\beta A^\beta p_\mu &(3.10b)\\
&= \delta^\mu{}_\beta A^\beta p_\mu &(3.10c)\\
&= A^\beta p_\beta . &(3.10d)
\end{aligned}
$$

## Step by step

**Left side.** $A^{\bar\alpha}p_{\bar\alpha}$ is the contraction $\tilde p(\vec A)$ computed with $\bar{\mathcal O}$'s components (Eq. 3.8 in $\bar{\mathcal O}$).

**(3.10a) — substitute the transformation laws.**
- Vector components: $A^{\bar\alpha} = \Lambda^{\bar\alpha}{}_\beta A^\beta$ (Eq. 2.7). The summed index is renamed $\beta$ so it can't clash with others.
- One-form components: $p_{\bar\alpha} = \Lambda^\mu{}_{\bar\alpha}p_\mu$ (Eq. 3.9). These transform with the **inverse** matrix, like the basis vectors $\vec e_{\bar\alpha} = \Lambda^\mu{}_{\bar\alpha}\vec e_\mu$. A fresh dummy $\mu$ is used, distinct from $\beta$.

Three sums are now present: over $\bar\alpha$, $\beta$, $\mu$, each with its own letter, as required (Problem 3.1).

**(3.10a) → (3.10b) — reorder the factors.** Every symbol here, $\Lambda^{\bar\alpha}{}_\beta$, $A^\beta$, $\Lambda^\mu{}_{\bar\alpha}$, $p_\mu$, stands for an ordinary real number. A product of numbers can be reordered, and finite sums can be done in any order (Problem 2.9). So the expression may be written with the two $\Lambda$s adjacent.

**(3.10b) → (3.10c) — the two matrices are inverses.** Do the sum over $\bar\alpha$ first: $\Lambda^\mu{}_{\bar\alpha}\Lambda^{\bar\alpha}{}_\beta$ is the $(\mu, \beta)$ entry of the matrix product $\Lambda(-v)\Lambda(v)$, the identity by Eq. (2.18). So $\Lambda^\mu{}_{\bar\alpha}\Lambda^{\bar\alpha}{}_\beta = \delta^\mu{}_\beta$.

**(3.10c) → (3.10d) — the Kronecker delta substitutes.** In $\sum_\mu\delta^\mu{}_\beta p_\mu$ only the term $\mu = \beta$ survives, with coefficient 1: $\delta^\mu{}_\beta p_\mu = p_\beta$ (Problem 2.11g). So $\delta^\mu{}_\beta A^\beta p_\mu = A^\beta p_\beta$.

**Conclusion.** $A^{\bar\alpha}p_{\bar\alpha} = A^\beta p_\beta$: the contraction gives the same number in every frame. That must be so, since $\tilde p(\vec A)$ is a property of the two geometric objects and makes no reference to a frame. The reason it works is that vector and one-form components transform **oppositely** (by inverse matrices); their product is invariant, just as $A^\alpha\vec e_\alpha$ is (§2.2).

```python
import sympy as sp
from schutz_gr import boost_matrix
v = sp.symbols("v", real=True)
L, Linv = boost_matrix(v), boost_matrix(-v)       # Λ^ᾱ_β(v) and Λ^μ_ᾱ(−v)
A = sp.Matrix(sp.symbols("A0:4")); p = sp.Matrix(sp.symbols("p0:4"))
Abar = L * A                                     # A^ᾱ = Λ^ᾱ_β A^β
pbar = Linv.T * p                                # p_ᾱ = Λ^μ_ᾱ p_μ  (sum over the row index of Λ(−v))
print("(3.10c) Λ(−v)Λ(v) = δ:", sp.simplify(Linv * L) == sp.eye(4))
print("A^ᾱ p_ᾱ − A^β p_β =", sp.simplify((Abar.T * pbar)[0] - (A.T * p)[0]))
```
