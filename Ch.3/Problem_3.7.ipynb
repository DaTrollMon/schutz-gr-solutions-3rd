# Chapter 3, Problem 7 — how the basis one-forms transform

**Problem.** Prove Eq. (3.13):
$$
\tilde\omega^{\bar\alpha} = \Lambda^{\bar\alpha}{}_\beta\,\tilde\omega^\beta .
$$

**Setting.** $\{\tilde\omega^\beta\}$ is the basis dual to $\{\vec e_\alpha\}$ in $\mathcal O$, defined by $\tilde\omega^\beta(\vec e_\alpha) = \delta^\beta{}_\alpha$ (Eq. 3.12); likewise $\tilde\omega^{\bar\alpha}(\vec e_{\bar\mu}) = \delta^{\bar\alpha}{}_{\bar\mu}$ in $\bar{\mathcal O}$. The vector bases are related by $\vec e_{\bar\mu} = \Lambda^\nu{}_{\bar\mu}\vec e_\nu$ (Eq. 2.15).

## Proof (two one-forms are equal if they agree on a basis)

A one-form is linear, so it's determined by its values on any basis (Problem 4b). Evaluate both sides of (3.13) on the basis vectors $\vec e_{\bar\mu}$ of $\bar{\mathcal O}$.

**Left side:** by the duality condition in $\bar{\mathcal O}$,
$$
\tilde\omega^{\bar\alpha}(\vec e_{\bar\mu}) = \delta^{\bar\alpha}{}_{\bar\mu}.
$$

**Right side:** use linearity, then the transformation of $\vec e_{\bar\mu}$, then duality in $\mathcal O$:
$$
\begin{aligned}
\Lambda^{\bar\alpha}{}_\beta\,\tilde\omega^\beta(\vec e_{\bar\mu})
&= \Lambda^{\bar\alpha}{}_\beta\,\tilde\omega^\beta\big(\Lambda^\nu{}_{\bar\mu}\vec e_\nu\big)
= \Lambda^{\bar\alpha}{}_\beta\,\Lambda^\nu{}_{\bar\mu}\,\tilde\omega^\beta(\vec e_\nu)\\
&= \Lambda^{\bar\alpha}{}_\beta\,\Lambda^\nu{}_{\bar\mu}\,\delta^\beta{}_\nu
= \Lambda^{\bar\alpha}{}_\beta\,\Lambda^\beta{}_{\bar\mu}
= \delta^{\bar\alpha}{}_{\bar\mu}.
\end{aligned}
$$
The last step is the inverse property, Problem 2.11(f): $\Lambda(v)\Lambda(-v) = I$.

Both sides give $\delta^{\bar\alpha}{}_{\bar\mu}$ on every basis vector $\vec e_{\bar\mu}$, so they are the same one-form:
$$
\tilde\omega^{\bar\alpha} = \Lambda^{\bar\alpha}{}_\beta\,\tilde\omega^\beta . \qquad\blacksquare
$$

## How to remember it

The basis one-forms transform with $\Lambda^{\bar\alpha}{}_\beta$, the **same** matrix as vector components ($A^{\bar\alpha} = \Lambda^{\bar\alpha}{}_\beta A^\beta$). The basis vectors and one-form components use the inverse. This is the "only equation we can write with the indices in their correct positions": the free index $\bar\alpha$ is up on both sides, and the dummy $\beta$ appears once up, once down. As a consequence $\tilde p = p_\alpha\tilde\omega^\alpha = p_{\bar\alpha}\tilde\omega^{\bar\alpha}$ is frame-independent, exactly as $\vec A = A^\alpha\vec e_\alpha$ is.

**Alternative route.** Since $\tilde\omega^\alpha = \tilde dx^\alpha$ (Eq. 3.20) and $x^{\bar\alpha} = \Lambda^{\bar\alpha}{}_\beta x^\beta$ is linear with constant coefficients, $\tilde dx^{\bar\alpha} = \Lambda^{\bar\alpha}{}_\beta\,\tilde dx^\beta$ directly.

```python
import sympy as sp
from schutz_gr import boost_matrix
v = sp.symbols("v", real=True)
L = boost_matrix(v)          # Λ^ᾱ_β
Linv = boost_matrix(-v)      # Λ^ν_μ̄
# components in O: ω^β → row β of the identity; e_μ̄ → column μ̄ of Λ(−v)
omega = sp.eye(4)
omega_bar = L * omega        # row ᾱ = Λ^ᾱ_β ω^β (components in O)
e_bar = Linv                 # column μ̄ = components in O of e_μ̄
print("ω^ᾱ(e_μ̄) = δ :", sp.simplify(omega_bar * e_bar) == sp.eye(4))
```
