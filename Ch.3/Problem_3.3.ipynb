# Chapter 3, Problem 3 — linearity of a one-form, by hand and with numbers

**Problem.**
- **(a)** Prove, by writing out all the terms, that $\tilde p(A^\alpha\vec e_\alpha) = A^\alpha\,\tilde p(\vec e_\alpha)$.
- **(b)** Let the components of $\tilde p$ be $(-1, 1, 2, 0)$, those of $\vec A$ be $(2, 1, 0, -1)$ and those of $\vec B$ be $(0, 2, 0, 0)$. Find (i) $\tilde p(\vec A)$; (ii) $\tilde p(\vec B)$; (iii) $\tilde p(\vec A - 3\vec B)$; (iv) $\tilde p(\vec A) - 3\tilde p(\vec B)$.

## (a)

Write out the sum inside the argument:
$$
\tilde p(A^\alpha\vec e_\alpha) = \tilde p\big(A^0\vec e_0 + A^1\vec e_1 + A^2\vec e_2 + A^3\vec e_3\big).
$$
A one-form is linear (§3.3): $\tilde p(a\vec U + b\vec V) = a\tilde p(\vec U) + b\tilde p(\vec V)$. Apply it repeatedly, peeling off one term at a time:
$$
\begin{aligned}
\tilde p\big(A^0\vec e_0 + [A^1\vec e_1 + A^2\vec e_2 + A^3\vec e_3]\big)
&= A^0\tilde p(\vec e_0) + \tilde p\big(A^1\vec e_1 + [A^2\vec e_2 + A^3\vec e_3]\big)\\
&= A^0\tilde p(\vec e_0) + A^1\tilde p(\vec e_1) + \tilde p\big(A^2\vec e_2 + A^3\vec e_3\big)\\
&= A^0\tilde p(\vec e_0) + A^1\tilde p(\vec e_1) + A^2\tilde p(\vec e_2) + A^3\tilde p(\vec e_3)\\
&= A^\alpha\,\tilde p(\vec e_\alpha). \qquad\blacksquare
\end{aligned}
$$
With $p_\alpha := \tilde p(\vec e_\alpha)$ (Eq. 3.7) this is Eq. (3.8): $\tilde p(\vec A) = A^\alpha p_\alpha$, the **contraction**: all plus signs, no metric needed.

## (b) The numbers

$\tilde p(\vec V) = p_\alpha V^\alpha = p_0V^0 + p_1V^1 + p_2V^2 + p_3V^3$ with $p_\alpha = (-1, 1, 2, 0)$.

- **(i)** $\tilde p(\vec A) = (-1)(2) + (1)(1) + (2)(0) + (0)(-1) = -2 + 1 = \boxed{-1}$
- **(ii)** $\tilde p(\vec B) = (-1)(0) + (1)(2) + 0 + 0 = \boxed{2}$
- **(iii)** $\vec A - 3\vec B \to (2, 1 - 6, 0, -1) = (2, -5, 0, -1)$, so $\tilde p(\vec A - 3\vec B) = -2 - 5 + 0 + 0 = \boxed{-7}$
- **(iv)** $\tilde p(\vec A) - 3\tilde p(\vec B) = -1 - 3(2) = \boxed{-7}$

(iii) = (iv), as linearity requires. Note there's no minus sign on the time term: $\tilde p(\vec A)$ is a contraction, not a dot product. If $(-1, 1, 2, 0)$ were the components of a *vector* $\vec P$, then $\vec P\cdot\vec A = +2 + 1 = 3$ — a different number.

```python
import numpy as np
p = np.array([-1, 1, 2, 0]); A = np.array([2, 1, 0, -1]); B = np.array([0, 2, 0, 0])
print("(i)  p(A) =", p @ A)
print("(ii) p(B) =", p @ B)
print("(iii) p(A − 3B) =", p @ (A - 3*B))
print("(iv) p(A) − 3p(B) =", p @ A - 3*(p @ B))
```
