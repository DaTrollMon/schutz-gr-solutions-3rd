# Chapter 3, Problem 25 — a full contraction is a scalar

**Problem.** Show that if $\mathsf A$ is a $\binom20$ tensor and $\mathbf B$ a $\binom02$ tensor, then $A^{\alpha\beta}B_{\alpha\beta}$ is frame invariant, i.e. a scalar.

**Transformation laws** (§3.5, the pattern of Eqs. 3.9, 2.7 applied index by index):
$$
A^{\bar\alpha\bar\beta} = \Lambda^{\bar\alpha}{}_\mu\Lambda^{\bar\beta}{}_\nu A^{\mu\nu},\qquad
B_{\bar\alpha\bar\beta} = \Lambda^\rho{}_{\bar\alpha}\Lambda^\sigma{}_{\bar\beta}B_{\rho\sigma}.
$$

## Proof by components

$$
\begin{aligned}
A^{\bar\alpha\bar\beta}B_{\bar\alpha\bar\beta}
&= \Lambda^{\bar\alpha}{}_\mu\Lambda^{\bar\beta}{}_\nu A^{\mu\nu}\;\Lambda^\rho{}_{\bar\alpha}\Lambda^\sigma{}_{\bar\beta}B_{\rho\sigma} &&\text{(substitute; all dummies distinct)}\\
&= \big(\Lambda^\rho{}_{\bar\alpha}\Lambda^{\bar\alpha}{}_\mu\big)\big(\Lambda^\sigma{}_{\bar\beta}\Lambda^{\bar\beta}{}_\nu\big)A^{\mu\nu}B_{\rho\sigma} &&\text{(reorder real-number factors)}\\
&= \delta^\rho{}_\mu\,\delta^\sigma{}_\nu\,A^{\mu\nu}B_{\rho\sigma} &&\text{(inverse property, Problem 10b)}\\
&= A^{\mu\nu}B_{\mu\nu}. &&\text{(deltas substitute)}
\end{aligned}
$$
The value is the same in every frame: a scalar. $\blacksquare$

## Proof without components

Write $\mathbf B = B_{\rho\sigma}\tilde\omega^\rho\otimes\tilde\omega^\sigma$ (Eq. 3.26) and $\mathsf A = A^{\mu\nu}\vec e_\mu\otimes\vec e_\nu$ (Problem 23). The double contraction is the number obtained by feeding the vectors of $\mathsf A$ into the slots of $\mathbf B$:
$$
A^{\mu\nu}\,\mathbf B(\vec e_\mu, \vec e_\nu) = A^{\mu\nu}B_{\mu\nu}.
$$
For a single term $\mathsf A = \vec U\otimes\vec V$ this is $\mathbf B(\vec U, \vec V)$, manifestly frame-independent; general $\mathsf A$ is a sum of such terms and the operation is linear. It's the same mechanism as $p_\alpha A^\alpha = \tilde p(\vec A)$ (Problem 5).

**Contrast:** $A^{\alpha\beta}B^{\alpha\beta}$ (both up) or $\sum_\alpha A^{\alpha\alpha}$ is **not** invariant: the $\Lambda$'s don't pair into inverses.

```python
import sympy as sp
from schutz_gr import boost_general
L = sp.Matrix(boost_general([sp.Rational(3, 10), sp.Rational(-2, 5), sp.Rational(1, 5)]))   # a general boost Λ^ᾱ_μ
Li = L.inv()                                                                                # Λ^ρ_ᾱ
A = sp.Matrix(4, 4, lambda i, j: sp.Symbol(f"A{i}{j}")); B = sp.Matrix(4, 4, lambda i, j: sp.Symbol(f"B{i}{j}"))
Abar = L * A * L.T                      # A^{ᾱβ̄} = Λ^ᾱ_μ Λ^β̄_ν A^{μν}
Bbar = Li.T * B * Li                    # B_{ᾱβ̄} = Λ^ρ_ᾱ Λ^σ_β̄ B_{ρσ}
full = lambda X, Y: sum(X[i, j] * Y[i, j] for i in range(4) for j in range(4))
print("A^{ᾱβ̄}B_{ᾱβ̄} − A^{μν}B_{μν} =", sp.simplify(sp.expand(full(Abar, Bbar) - full(A, B))))
print("non-invariant example, trace A^{αα}: change =", sp.simplify(Abar.trace() - A.trace()) != 0)
```
