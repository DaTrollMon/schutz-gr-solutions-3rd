# Chapter 3, Problem 11 — comma notation

**Problem.** Use the notation $\partial\phi/\partial x^\alpha = \phi_{,\alpha}$ to rewrite Eqs. (3.14), (3.15) and (3.18).

**The notation (Eq. 3.19).** A comma followed by an index means "partial derivative with respect to that coordinate": $\phi_{,\alpha} := \partial\phi/\partial x^\alpha$. The index comes out **down**, because the gradient transforms like a one-form (Eq. 3.17). Applied to a component with its own indices, the derivative index goes after the comma: $T^\alpha{}_{\beta,\gamma} := \partial T^\alpha{}_\beta/\partial x^\gamma$.

## Eq. (3.14): rate of change of $\phi$ along a world line

Original:
$$
\frac{d\phi}{d\tau} = \frac{\partial\phi}{\partial t}U^t + \frac{\partial\phi}{\partial x}U^x + \frac{\partial\phi}{\partial y}U^y + \frac{\partial\phi}{\partial z}U^z .
$$
In comma notation, with the summation convention:
$$
\boxed{\frac{d\phi}{d\tau} = \phi_{,\alpha}\,U^\alpha}
$$
(a contraction of the one-form $\tilde d\phi$ with $\vec U$: $\tilde d\phi(\vec U)$).

## Eq. (3.15): components of the gradient

Original: $\tilde d\phi \to \left(\dfrac{\partial\phi}{\partial t}, \dfrac{\partial\phi}{\partial x}, \dfrac{\partial\phi}{\partial y}, \dfrac{\partial\phi}{\partial z}\right)$.
$$
\boxed{\tilde d\phi \xrightarrow{\mathcal O} (\phi_{,0},\ \phi_{,1},\ \phi_{,2},\ \phi_{,3}) = \{\phi_{,\alpha}\}},\qquad \text{i.e. } (\tilde d\phi)_\alpha = \phi_{,\alpha},\quad \tilde d\phi = \phi_{,\alpha}\,\tilde dx^\alpha .
$$

## Eq. (3.18): derivative of the coordinate transformation

Original: $\partial x^\beta/\partial x^{\bar\alpha} = \Lambda^\beta{}_{\bar\alpha}$.
$$
\boxed{x^\beta{}_{,\bar\alpha} = \Lambda^\beta{}_{\bar\alpha}}
$$
(and similarly $x^{\bar\alpha}{}_{,\beta} = \Lambda^{\bar\alpha}{}_\beta$, $x^\alpha{}_{,\beta} = \delta^\alpha{}_\beta$).

**With these, the transformation of the gradient (3.16–3.17) reads** $\phi_{,\bar\alpha} = x^\beta{}_{,\bar\alpha}\,\phi_{,\beta} = \Lambda^\beta{}_{\bar\alpha}\,\phi_{,\beta}$. The indices line up automatically: the chain rule *is* the one-form transformation law.

```python
import sympy as sp
t, x, y, z, tau = sp.symbols("t x y z tau", real=True)
X = [t, x, y, z]
phi = sp.Function("phi")(*X)
grad = [sp.diff(phi, c) for c in X]                       # φ_,α
paths = [sp.Function(n)(tau) for n in ("t", "x", "y", "z")]
phi_path = phi.subs(dict(zip(X, paths)), simultaneous=True)
lhs = sp.diff(phi_path, tau)
rhs = sum(g.subs(dict(zip(X, paths)), simultaneous=True) * sp.diff(p, tau) for g, p in zip(grad, paths))
print("dφ/dτ − φ_,α U^α =", sp.simplify(lhs - rhs))
```
