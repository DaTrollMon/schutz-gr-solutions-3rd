# Chapter 3, Problem 18 — lowering and raising with the metric

**Problem.**
- **(a)** Find the one-forms mapped by the metric tensor from the vectors
$\vec A \to (1, 0, -1, 0)$, $\vec B \to (0, 1, 1, 0)$, $\vec C \to (-1, 0, -1, 0)$, $\vec D \to (0, 0, 1, 1)$.
- **(b)** Find the vectors mapped by the inverse of the metric tensor from the one-forms
$\tilde p \to (3, 0, -1, -1)$, $\tilde q \to (1, -1, 1, 1)$, $\tilde r \to (0, -5, -1, 0)$, $\tilde s \to (-2, 1, 0, 0)$.

**Rules.** Lowering: $V_\alpha = \eta_{\alpha\beta}V^\beta$ (Eq. 3.39), so $V_0 = -V^0$, $V_i = V^i$ (Eqs. 3.40–3.42). Raising: $p^\alpha = \eta^{\alpha\beta}p_\beta$ (Eq. 3.43) with $\eta^{\alpha\beta} = \operatorname{diag}(-1, 1, 1, 1)$ (Eq. 3.44), so $p^0 = -p_0$, $p^i = p_i$. In Minkowski coordinates both operations just **flip the sign of the time component**.

## (a) Vectors → one-forms, $\tilde V = \mathbf g(\vec V, \cdot)$

| vector | $(V^0, V^1, V^2, V^3)$ | one-form $(V_0, V_1, V_2, V_3) = (-V^0, V^1, V^2, V^3)$ |
|---|---|---|
| $\vec A$ | $(1, 0, -1, 0)$ | $\tilde A \to \boxed{(-1, 0, -1, 0)}$ |
| $\vec B$ | $(0, 1, 1, 0)$ | $\tilde B \to \boxed{(0, 1, 1, 0)}$ |
| $\vec C$ | $(-1, 0, -1, 0)$ | $\tilde C \to \boxed{(1, 0, -1, 0)}$ |
| $\vec D$ | $(0, 0, 1, 1)$ | $\tilde D \to \boxed{(0, 0, 1, 1)}$ |

$\vec B$ and $\vec D$ have no time component, so their one-forms have the same components.

**Meaning check.** $\tilde A(\vec X) = \vec A\cdot\vec X$ for every $\vec X$. E.g. $\tilde A(\vec C) = A_\alpha C^\alpha = (-1)(-1) + 0 + (-1)(-1) + 0 = 2$, and $\vec A\cdot\vec C = -(1)(-1) + (-1)(-1) = 2$ ✓.

## (b) One-forms → vectors, $\vec p = \mathbf g^{-1}(\tilde p, \cdot)$

| one-form | $(p_0, p_1, p_2, p_3)$ | vector $(p^0, p^1, p^2, p^3) = (-p_0, p_1, p_2, p_3)$ |
|---|---|---|
| $\tilde p$ | $(3, 0, -1, -1)$ | $\vec p \to \boxed{(-3, 0, -1, -1)}$ |
| $\tilde q$ | $(1, -1, 1, 1)$ | $\vec q \to \boxed{(-1, -1, 1, 1)}$ |
| $\tilde r$ | $(0, -5, -1, 0)$ | $\vec r \to \boxed{(0, -5, -1, 0)}$ |
| $\tilde s$ | $(-2, 1, 0, 0)$ | $\vec s \to \boxed{(2, 1, 0, 0)}$ |

**Character of each** (norms, which are the same for a one-form and its vector, Eq. 3.51):

$\vec p$: $-9 + 1 + 1 = -7$ timelike (past-pointing, since $p^0 < 0$); $\vec q$: $-1 + 1 + 1 + 1 = 2$ spacelike; $\vec r$: $25 + 1 = 26$ spacelike; $\vec s$: $-4 + 1 = -3$ timelike (future-pointing).

```python
import sympy as sp
eta = sp.diag(-1, 1, 1, 1)
vecs = {"A": [1, 0, -1, 0], "B": [0, 1, 1, 0], "C": [-1, 0, -1, 0], "D": [0, 0, 1, 1]}
forms = {"p": [3, 0, -1, -1], "q": [1, -1, 1, 1], "r": [0, -5, -1, 0], "s": [-2, 1, 0, 0]}
print("(a) lowered:")
for k, V in vecs.items():
    print(f"   {k}_α =", list(eta * sp.Matrix(V)))
print("(b) raised (η^{αβ} = inverse of η_{αβ}):")
etainv = eta.inv()
for k, p in forms.items():
    P = etainv * sp.Matrix(p)
    print(f"   {k}^α =", list(P), "  norm", (sp.Matrix(p).T * etainv * sp.Matrix(p))[0])
```
