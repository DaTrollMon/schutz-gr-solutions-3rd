# Chapter 3, Problem 28 — the gradient of a tensor

**Problem.** Deduce Eq. (3.66) from Eq. (3.65).

**The equations (§3.8).** For a $\binom11$ tensor field $\mathsf T = T^\alpha{}_\beta\,\tilde\omega^\beta\otimes\vec e_\alpha$ (Eq. 3.61), the rate of change along a world line with tangent $\vec U$ is
$$
\frac{d\mathsf T}{d\tau} = \big(T^\alpha{}_{\beta,\gamma}\,\tilde\omega^\beta\otimes\vec e_\alpha\big)\,U^\gamma \qquad(3.65)
$$
and from this we're to deduce that
$$
\nabla\mathsf T := T^\alpha{}_{\beta,\gamma}\,\tilde\omega^\beta\otimes\tilde\omega^\gamma\otimes\vec e_\alpha \qquad(3.66)
$$
is a $\binom12$ tensor.

## The argument

**1. $d\mathsf T/d\tau$ is a $\binom11$ tensor for every $\vec U$.** It is the limit of $[\mathsf T(\tau + \Delta\tau) - \mathsf T(\tau)]/\Delta\tau$, a difference of $\binom11$ tensors divided by a number (Eq. 3.62). Tensors form a vector space (Problem 23), so this is a $\binom11$ tensor.

**2. It depends linearly on $\vec U$.** In (3.65) $U^\gamma$ appears linearly, and the coefficients $T^\alpha{}_{\beta,\gamma}$ don't depend on $\vec U$ (they're derivatives of the field at the point). Rewrite $U^\gamma = \tilde\omega^\gamma(\vec U)$:
$$
\frac{d\mathsf T}{d\tau} = T^\alpha{}_{\beta,\gamma}\,\tilde\omega^\gamma(\vec U)\;\tilde\omega^\beta\otimes\vec e_\alpha .
$$

**3. A linear map from vectors to $\binom11$ tensors is a $\binom12$ tensor.** Define $\nabla\mathsf T$ as the function of one one-form and two vectors
$$
\nabla\mathsf T(\tilde p; \vec A, \vec U) := \frac{d\mathsf T}{d\tau}\Big|_{\vec U}(\tilde p; \vec A),
$$
i.e. "the $\binom11$ tensor $d\mathsf T/d\tau$ for world-line tangent $\vec U$, evaluated on $(\tilde p; \vec A)$". It's linear in $\tilde p$ and $\vec A$ (step 1) and in $\vec U$ (step 2), so it is a $\binom12$ tensor. Its components are found by feeding in basis elements:
$$
(\nabla\mathsf T)^\alpha{}_{\beta\gamma} = \nabla\mathsf T(\tilde\omega^\alpha; \vec e_\beta, \vec e_\gamma) = T^\mu{}_{\nu,\lambda}\,\tilde\omega^\lambda(\vec e_\gamma)\,\tilde\omega^\nu(\vec e_\beta)\,\vec e_\mu(\tilde\omega^\alpha) = T^\alpha{}_{\beta,\gamma}.
$$

**4. Expand on the product basis** (Problem 23b). A $\binom12$ tensor with components $T^\alpha{}_{\beta,\gamma}$ is
$$
\nabla\mathsf T = T^\alpha{}_{\beta,\gamma}\,\tilde\omega^\beta\otimes\tilde\omega^\gamma\otimes\vec e_\alpha ,\qquad(3.66)\ \blacksquare
$$
with the derivative slot $\gamma$ written last among the lower indices. And (3.65) is recovered as $d\mathsf T/d\tau = \nabla_{\vec U}\mathsf T$, i.e. $\nabla\mathsf T$ with $\vec U$ in its $\gamma$ slot (Eqs. 3.67, 3.68).

**Alternative: the transformation law.** Since $x^{\bar\gamma}\mapsto x^\gamma$ is linear, $\partial/\partial x^{\bar\gamma} = \Lambda^\gamma{}_{\bar\gamma}\partial/\partial x^\gamma$ and $\Lambda$ is constant, so
$T^{\bar\alpha}{}_{\bar\beta,\bar\gamma} = \Lambda^{\bar\alpha}{}_\mu\Lambda^\nu{}_{\bar\beta}\Lambda^\lambda{}_{\bar\gamma}T^\mu{}_{\nu,\lambda}$: exactly the law for a $\binom12$ tensor. (This step fails in curvilinear coordinates, where the basis varies, which is the doorway to Chapter 5.)

```python
import sympy as sp
from schutz_gr import boost_matrix
t, x = sp.symbols("t x", real=True); v = sp.Rational(3, 5)
L = boost_matrix(v)[:2, :2]; Li = L.inv()
T = sp.Matrix([[t*x, x**2], [sp.sin(t), t + 3*x]])                  # T^α_β(t, x) in the t–x plane
X = [t, x]
dT = [[[sp.diff(T[a, b], X[c]) for c in range(2)] for b in range(2)] for a in range(2)]   # T^α_{β,γ}
# barred frame: T̄ = Λ T Λ^{-1}, as a function of barred coords
tb, xb = sp.symbols("tbar xbar", real=True)
back = dict(zip(X, list(Li * sp.Matrix([tb, xb]))))
Tbar = (L * T * Li).subs(back, simultaneous=True)
Xb = [tb, xb]
ok = all(sp.simplify(sp.diff(Tbar[a, b], Xb[c]) - sum(L[a, m] * Li[n, b] * Li[l, c] * dT[m][n][l] for m in range(2) for n in range(2) for l in range(2)).subs(back, simultaneous=True)) == 0
         for a in range(2) for b in range(2) for c in range(2))
print("T^ᾱ_{β̄,γ̄} = Λ^ᾱ_μ Λ^ν_β̄ Λ^λ_γ̄ T^μ_{ν,λ}  →  ∇T transforms as a (1,2) tensor:", ok)
```
