# Chapter 3, Problem 23 — $\binom MN$ tensors form a vector space, with a product basis

**Problem.**
- **(a)** Prove that the set of all $\binom MN$ tensors for fixed $M, N$ forms a vector space. (You must define addition of such tensors and their multiplication by numbers.)
- **(b)** Prove that a basis for this space is the set
$$
\{\underbrace{\vec e_\alpha\otimes\vec e_\beta\otimes\cdots\otimes\vec e_\gamma}_{M \text{ vectors}}\otimes\underbrace{\tilde\omega^\mu\otimes\tilde\omega^\nu\otimes\cdots\otimes\tilde\omega^\lambda}_{N \text{ one-forms}}\}.
$$
(You will have to define the outer product of more than two one-forms.)

**Definition (§3.5).** A $\binom MN$ tensor is a function of $M$ one-forms and $N$ vectors into the real numbers, linear in each argument separately.

## (a) Vector-space structure

Write arguments as $(\tilde p, \dots, \tilde r; \vec A, \dots, \vec C)$ ($M$ one-forms then $N$ vectors). Define, pointwise,
$$
(\mathbf S + \mathbf T)(\tilde p, \dots; \vec A, \dots) := \mathbf S(\tilde p, \dots; \vec A, \dots) + \mathbf T(\tilde p, \dots; \vec A, \dots),\qquad
(a\mathbf T)(\tilde p, \dots; \vec A, \dots) := a\,\mathbf T(\tilde p, \dots; \vec A, \dots).
$$

**Closure.** Fix all arguments but one, say the $k$-th. As a function of that slot, $\mathbf S$ and $\mathbf T$ are linear, so their sum and their multiples are too (exactly the computation in Problem 22(i)). Hence $\mathbf S + \mathbf T$ and $a\mathbf T$ are again multilinear, i.e. $\binom MN$ tensors.

**Axioms.** All eight hold because they hold for the real-number **values** at every set of arguments (the table in Problem 22 applies verbatim). The zero tensor gives 0 on everything; $-\mathbf T$ gives the negatives of $\mathbf T$'s values. $\blacksquare$

## (b) The product basis

**Outer product of several factors.** For vectors and one-forms, define
$$
(\vec e_\alpha\otimes\cdots\otimes\vec e_\gamma\otimes\tilde\omega^\mu\otimes\cdots\otimes\tilde\omega^\lambda)(\tilde p, \dots, \tilde r; \vec A, \dots, \vec C) := \tilde p(\vec e_\alpha)\cdots\tilde r(\vec e_\gamma)\;\tilde\omega^\mu(\vec A)\cdots\tilde\omega^\lambda(\vec C),
$$
the product of the values of each factor on its own argument (a vector eats a one-form: $\vec e_\alpha(\tilde p) = \tilde p(\vec e_\alpha) = p_\alpha$). It's a product of factors each linear in one argument, so it's multilinear: a $\binom MN$ tensor. Call these $\mathbf E_{\alpha\cdots\gamma}{}^{\mu\cdots\lambda}$. On basis arguments,
$$
\mathbf E_{\alpha\cdots\gamma}{}^{\mu\cdots\lambda}(\tilde\omega^{\rho}, \dots, \tilde\omega^{\tau}; \vec e_\sigma, \dots, \vec e_\kappa) = \delta^\rho{}_\alpha\cdots\delta^\tau{}_\gamma\;\delta^\mu{}_\sigma\cdots\delta^\lambda{}_\kappa .\qquad(*)
$$

**Spanning.** Given any $\binom MN$ tensor $\mathbf T$, define its components
$$
T^{\alpha\cdots\gamma}{}_{\mu\cdots\lambda} := \mathbf T(\tilde\omega^\alpha, \dots, \tilde\omega^\gamma; \vec e_\mu, \dots, \vec e_\lambda).
$$
Claim: $\mathbf T = T^{\alpha\cdots\gamma}{}_{\mu\cdots\lambda}\,\mathbf E_{\alpha\cdots\gamma}{}^{\mu\cdots\lambda}$. Evaluate both sides on arbitrary arguments, expanded on the bases, $\tilde p = p_\rho\tilde\omega^\rho$, …, $\vec A = A^\sigma\vec e_\sigma$, …. By multilinearity the left side is
$$
\mathbf T(\tilde p, \dots; \vec A, \dots) = p_\rho\cdots A^\sigma\cdots\;T^{\rho\cdots}{}_{\sigma\cdots},
$$
and the right side, using $(*)$, is $T^{\alpha\cdots}{}_{\mu\cdots}\,p_\rho\cdots A^\sigma\cdots\,\delta^\rho{}_\alpha\cdots\delta^\mu{}_\sigma\cdots = p_\rho\cdots A^\sigma\cdots\,T^{\rho\cdots}{}_{\sigma\cdots}$. Equal. So the $\mathbf E$'s span.

**Linear independence.** Suppose $c^{\alpha\cdots\gamma}{}_{\mu\cdots\lambda}\,\mathbf E_{\alpha\cdots\gamma}{}^{\mu\cdots\lambda} = 0$. Evaluate on $(\tilde\omega^\rho, \dots, \tilde\omega^\tau; \vec e_\sigma, \dots, \vec e_\kappa)$. By $(*)$ the deltas pick out one term: $c^{\rho\cdots\tau}{}_{\sigma\cdots\kappa} = 0$. Since this holds for every choice of indices, all coefficients vanish. $\blacksquare$

**Dimension.** Each of the $M + N$ indices ranges over 4 values, so the space has dimension $4^{M+N}$: 4 for vectors and one-forms, 16 for $\binom02$, $\binom11$, $\binom20$ tensors, 64 for $\binom12$, etc.

```python
import itertools, numpy as np
# Concrete check for a (1,2) tensor: the 64 product tensors e_α⊗ω^μ⊗ω^ν (as 4×4×4 arrays) are independent and
# the expansion with components T(ω^α; e_μ, e_ν) reproduces T on random arguments.
rng = np.random.default_rng(0)
E = np.eye(4)
basis = [np.einsum("i,j,k->ijk", E[a], E[m], E[n]) for a, m, n in itertools.product(range(4), repeat=3)]
Mat = np.array([b.ravel() for b in basis])
print("rank of the 64 basis tensors:", np.linalg.matrix_rank(Mat))
T = rng.normal(size=(4, 4, 4))                                   # T^α_{μν}
T_of = lambda p, A, B: np.einsum("ijk,i,j,k->", T, p, A, B)
comps = np.array([[[T_of(E[a], E[m], E[n]) for n in range(4)] for m in range(4)] for a in range(4)])
recon = sum(comps[a, m, n] * basis[16*a + 4*m + n] for a, m, n in itertools.product(range(4), repeat=3))
p, A, B = rng.normal(size=(3, 4))
print("T(p;A,B) − (T^α_{μν} E)(p;A,B) =", T_of(p, A, B) - np.einsum("ijk,i,j,k->", recon, p, A, B))
```
