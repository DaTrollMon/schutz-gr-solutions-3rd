# Chapter 3, Problem 29 — the Leibniz rule for $\nabla$

**Problem.** Prove that tensor differentiation obeys the Leibniz (product) rule:
$$
\nabla(\mathsf A\otimes\mathsf B) = (\nabla\mathsf A)\otimes\mathsf B + \mathsf A\otimes\nabla\mathsf B .
$$

**Reading.** Take $\mathsf A$, $\mathsf B$ to be tensor fields of any types; for definiteness, vectors $\vec A$ and $\vec B$ (the argument is identical for any number of indices). In Minkowski coordinates the basis is constant, so the components of $\nabla\mathsf S$ are just the partial derivatives of the components of $\mathsf S$, with one new lower index (Eq. 3.66).

**Slot bookkeeping.** Schutz's $\nabla$ puts the derivative index in a new lower slot, written last: $(\nabla\vec A)^\alpha{}_\gamma = A^\alpha{}_{,\gamma}$. So the three terms have components with free indices $\alpha, \beta$ (up) and $\gamma$ (down), and the identity is to be read with the indices matched:
$$
(A^\alpha B^\beta)_{,\gamma} = A^\alpha{}_{,\gamma}B^\beta + A^\alpha B^\beta{}_{,\gamma}.
$$
(Strictly, $(\nabla\vec A)\otimes\vec B$ has its slots in the order $(\alpha, \gamma, \beta)$ while $\nabla(\vec A\otimes\vec B)$ has $(\alpha, \beta, \gamma)$; the rule identifies them after that reordering, which is what the index equation expresses.)

## Proof

**Components of each side.** $\vec A\otimes\vec B$ has components $A^\alpha B^\beta$ (definition of outer product evaluated on $\tilde\omega^\alpha, \tilde\omega^\beta$). Hence
$$
\big[\nabla(\vec A\otimes\vec B)\big]^{\alpha\beta}{}_\gamma = \frac{\partial}{\partial x^\gamma}\big(A^\alpha B^\beta\big).
$$
For each fixed $\alpha, \beta, \gamma$, $A^\alpha B^\beta$ is a product of two ordinary functions of position, so the ordinary product rule of calculus applies:
$$
\frac{\partial}{\partial x^\gamma}\big(A^\alpha B^\beta\big) = \frac{\partial A^\alpha}{\partial x^\gamma}B^\beta + A^\alpha\frac{\partial B^\beta}{\partial x^\gamma} = A^\alpha{}_{,\gamma}B^\beta + A^\alpha B^\beta{}_{,\gamma}.
$$
**Identify the terms.** $A^\alpha{}_{,\gamma}B^\beta$ are the components of $(\nabla\vec A)\otimes\vec B$; $A^\alpha B^\beta{}_{,\gamma}$ those of $\vec A\otimes\nabla\vec B$. The two sides have equal components in a basis, hence are the same tensor:
$$
\nabla(\vec A\otimes\vec B) = (\nabla\vec A)\otimes\vec B + \vec A\otimes\nabla\vec B. \qquad\blacksquare
$$

**General types.** For $\mathsf A$ with components $A^{\alpha\cdots}{}_{\mu\cdots}$ and $\mathsf B$ with $B^{\beta\cdots}{}_{\nu\cdots}$, the same calculation gives $(A^{\alpha\cdots}{}_{\mu\cdots}B^{\beta\cdots}{}_{\nu\cdots})_{,\gamma} = A^{\alpha\cdots}{}_{\mu\cdots,\gamma}B^{\beta\cdots}{}_{\nu\cdots} + A^{\alpha\cdots}{}_{\mu\cdots}B^{\beta\cdots}{}_{\nu\cdots,\gamma}$.

**Directional form.** Contracting with $U^\gamma$: $\nabla_{\vec U}(\mathsf A\otimes\mathsf B) = (\nabla_{\vec U}\mathsf A)\otimes\mathsf B + \mathsf A\otimes\nabla_{\vec U}\mathsf B$, the familiar product rule for $d/d\tau$ along a world line. (Problem 30(g) is an instance: $(U^\alpha D^\beta)_{,\beta}$.)

Note the proof used constancy of the basis. In curved spacetime the rule still holds, but proving it needs the covariant derivative of Chapter 5.

```python
import sympy as sp
X = sp.symbols("t x y z", real=True)
A = [sp.Function(f"A{i}")(*X) for i in range(4)]; B = [sp.Function(f"B{i}")(*X) for i in range(4)]
ok = all(sp.simplify(sp.diff(A[a]*B[b], X[g]) - (sp.diff(A[a], X[g])*B[b] + A[a]*sp.diff(B[b], X[g]))) == 0
         for a in range(4) for b in range(4) for g in range(4))
print("(A^α B^β)_,γ = A^α_,γ B^β + A^α B^β_,γ for all 64 components:", ok)
```
