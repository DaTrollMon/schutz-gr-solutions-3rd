## (a) Two different sums

**The double sum** has two independent dummy indices, so it has $4\times4 = 16$ terms:
$$
M_{\alpha\beta}A^\alpha B^\beta = \underbrace{M_{00}A^0B^0 + M_{11}A^1B^1 + M_{22}A^2B^2 + M_{33}A^3B^3}_{\alpha=\beta\ (4\text{ terms})} + \underbrace{M_{01}A^0B^1 + M_{10}A^1B^0 + \cdots}_{\alpha\neq\beta\ (12\text{ terms})}.
$$
**The single sum** forces $\alpha = \beta$ in every term, so it has only the 4 "diagonal" terms:
$$
\sum_\alpha M_{\alpha\alpha}A^\alpha B^\alpha = M_{00}A^0B^0 + M_{11}A^1B^1 + M_{22}A^2B^2 + M_{33}A^3B^3 .
$$
**They differ by the 12 off-diagonal terms**, $\sum_{\alpha\neq\beta}M_{\alpha\beta}A^\alpha B^\beta$, which are not zero in general. To show "not equivalent" one counterexample suffices: take $M_{01} = 1$ and every other $M_{\alpha\beta} = 0$, with $\vec A \to (1, 0, 0, 0)$, $\vec B \to (0, 1, 0, 0)$. Then
$$
M_{\alpha\beta}A^\alpha B^\beta = M_{01}A^0B^1 = 1,\qquad \sum_\alpha M_{\alpha\alpha}A^\alpha B^\alpha = 0 .
$$
**The lesson.** In the summation convention each *independent* sum needs its own dummy letter (§3.1: "the importance of using different indices"). The expression $M_{\alpha\alpha}A^\alpha B^\alpha$, with $\alpha$ appearing four times, is not a legal Einstein-convention expression at all: it has to be written with an explicit $\sum$, and then it means something different. (The two do agree in the special case where $M$ is diagonal.)

## (b) $A^\alpha B^\beta\eta_{\alpha\beta}$

$\eta_{\alpha\beta} = \mathrm{diag}(-1, 1, 1, 1)$ is zero unless $\alpha = \beta$. So of the 16 terms of the double sum only the 4 diagonal ones survive:
$$
A^\alpha B^\beta\eta_{\alpha\beta} = A^0B^0\eta_{00} + A^1B^1\eta_{11} + A^2B^2\eta_{22} + A^3B^3\eta_{33} = -A^0B^0 + A^1B^1 + A^2B^2 + A^3B^3 . \quad\blacksquare
$$
This is the scalar product $\vec A\cdot\vec B$ (Eq. 3.1). It's also why part (a)'s two expressions *happen* to agree when $M = \eta$: the off-diagonal terms are all zero.

```python
import sympy as sp
M = sp.Matrix(4, 4, lambda a, b: sp.Symbol(f"M_{a}{b}"))
A = sp.symbols("A0:4"); B = sp.symbols("B0:4")
double = sp.expand(sum(M[a, b]*A[a]*B[b] for a in range(4) for b in range(4)))
single = sum(M[a, a]*A[a]*B[a] for a in range(4))
print("terms: double sum", len(double.args), "  single sum", len(single.args))
print("difference =", sp.expand(double - single))
eta = sp.diag(-1, 1, 1, 1)
print("(b) A^α B^β η_αβ =", sp.expand(sum(A[a]*B[b]*eta[a, b] for a in range(4) for b in range(4))))
```
