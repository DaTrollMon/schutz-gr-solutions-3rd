# Chapter 3, Problem 10 — $\partial x^\alpha/\partial x^\beta$ and the inverse property

**Problem.**
- **(a)** Given a frame $\mathcal O$ with coordinates $\{x^\alpha\}$, show that $\partial x^\alpha/\partial x^\beta = \delta^\alpha{}_\beta$.
- **(b)** For any two frames, Eq. (3.18) gives $\partial x^\beta/\partial x^{\bar\alpha} = \Lambda^\beta{}_{\bar\alpha}$. Show that (a) and the chain rule imply
$$
\Lambda^\beta{}_{\bar\alpha}\,\Lambda^{\bar\alpha}{}_\mu = \delta^\beta{}_\mu .
$$
This is the inverse property again.

## (a)

The four coordinates $t, x, y, z$ are **independent** variables. A partial derivative with respect to $x^\beta$ holds the other three fixed. So:
- if $\alpha = \beta$: $\partial x^\alpha/\partial x^\alpha = 1$ (e.g. $\partial x/\partial x = 1$);
- if $\alpha \neq \beta$: $x^\alpha$ is one of the variables held fixed, so $\partial x^\alpha/\partial x^\beta = 0$ (e.g. $\partial t/\partial x = 0$).

That's the definition of the Kronecker delta: $\boxed{\partial x^\alpha/\partial x^\beta = \delta^\alpha{}_\beta}$. (In the comma notation of §3.3: $x^\alpha{}_{,\beta} = \delta^\alpha{}_\beta$, which is how Eq. 3.20, $\tilde dx^\alpha = \tilde\omega^\alpha$, was obtained.)

## (b)

Regard $x^\beta$ as a function of the barred coordinates, which are themselves functions of the unbarred ones: $x^\beta = x^\beta\big(x^{\bar\alpha}(x^\mu)\big)$. The multivariable chain rule gives
$$
\frac{\partial x^\beta}{\partial x^\mu} = \sum_{\bar\alpha}\frac{\partial x^\beta}{\partial x^{\bar\alpha}}\,\frac{\partial x^{\bar\alpha}}{\partial x^\mu}.
$$
- **Left side:** $\delta^\beta{}_\mu$, by (a).
- **First factor:** $\partial x^\beta/\partial x^{\bar\alpha} = \Lambda^\beta{}_{\bar\alpha}$, Eq. (3.18) (from $x^\beta = \Lambda^\beta{}_{\bar\alpha}x^{\bar\alpha}$ with constant coefficients).
- **Second factor:** likewise $x^{\bar\alpha} = \Lambda^{\bar\alpha}{}_\mu x^\mu$ gives $\partial x^{\bar\alpha}/\partial x^\mu = \Lambda^{\bar\alpha}{}_\mu$.

Therefore
$$
\boxed{\Lambda^\beta{}_{\bar\alpha}\,\Lambda^{\bar\alpha}{}_\mu = \delta^\beta{}_\mu}\qquad\blacksquare
$$
The matrix of the transformation $\bar{\mathcal O}\to\mathcal O$ is the inverse of that for $\mathcal O\to\bar{\mathcal O}$. This is Eq. (2.18) again, now derived from calculus instead of from basis-vector algebra. The advantage of this derivation: it works for **any** invertible coordinate change, not just Lorentz transformations — which is how it'll be used in curved spacetime (Chapter 5).

```python
import sympy as sp
from schutz_gr import lorentz, inverse_lorentz
t, x, v = sp.symbols("t x v", real=True)
X = [t, x]
print("(a) ∂x^α/∂x^β =", sp.Matrix(2, 2, lambda a, b: sp.diff(X[a], X[b])))
tb, xb = lorentz(t, x, v)                                   # x̄(x)
tbs, xbs = sp.symbols("tbar xbar", real=True)
tt, xx = inverse_lorentz(tbs, xbs, v)                       # x(x̄)
J_bar_of_x = sp.Matrix([[sp.diff(tb, t), sp.diff(tb, x)], [sp.diff(xb, t), sp.diff(xb, x)]])        # Λ^ᾱ_μ
J_x_of_bar = sp.Matrix([[sp.diff(tt, tbs), sp.diff(tt, xbs)], [sp.diff(xx, tbs), sp.diff(xx, xbs)]])  # Λ^β_ᾱ
print("(b) Λ^β_ᾱ Λ^ᾱ_μ =", sp.simplify(J_x_of_bar * J_bar_of_x))
```
