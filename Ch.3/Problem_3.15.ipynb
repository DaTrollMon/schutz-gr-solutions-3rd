# Chapter 3, Problem 15 — from Eq. (3.23) to Eq. (3.24)

**Problem.** Supply the reasoning leading from Eq. (3.23) to Eq. (3.24).

**The passage (§3.4).** Looking for a basis $\{\tilde\omega^{\alpha\beta}\}$ of $\binom02$ tensors, Schutz asks whether
$$
\mathbf f = f_{\alpha\beta}\,\tilde\omega^{\alpha\beta}\qquad(3.23)
$$
can hold, and says this would imply
$$
\tilde\omega^{\alpha\beta}(\vec e_\mu, \vec e_\nu) = \delta^\alpha{}_\mu\,\delta^\beta{}_\nu .\qquad(3.24)
$$

## The reasoning, step by step

**1. Evaluate both sides of (3.23) on a pair of basis vectors.** If (3.23) is an equation between tensors, both sides give the same number on any arguments, in particular on $(\vec e_\mu, \vec e_\nu)$:
$$
\mathbf f(\vec e_\mu, \vec e_\nu) = \big(f_{\alpha\beta}\,\tilde\omega^{\alpha\beta}\big)(\vec e_\mu, \vec e_\nu).
$$

**2. Left side = a component.** By the definition of components, Eq. (3.21): $\mathbf f(\vec e_\mu, \vec e_\nu) = f_{\mu\nu}$.

**3. Right side: use the vector-space operations on tensors.** $f_{\alpha\beta}\tilde\omega^{\alpha\beta}$ is a sum of tensors multiplied by numbers; such a combination is evaluated term by term (that's how addition and scalar multiplication of tensors are defined, the analogs of Eqs. 3.6a,b). So
$$
\big(f_{\alpha\beta}\,\tilde\omega^{\alpha\beta}\big)(\vec e_\mu, \vec e_\nu) = f_{\alpha\beta}\,\tilde\omega^{\alpha\beta}(\vec e_\mu, \vec e_\nu).
$$

**4. Combine:** for all $\mu, \nu$,
$$
f_{\mu\nu} = f_{\alpha\beta}\,\tilde\omega^{\alpha\beta}(\vec e_\mu, \vec e_\nu).\qquad(*)
$$

**5. This must hold for every $\mathbf f$.** We want $\{\tilde\omega^{\alpha\beta}\}$ to work as a basis for *all* $\binom02$ tensors, so $(*)$ must hold for arbitrary numbers $f_{\alpha\beta}$. Write the left side with Kronecker deltas, $f_{\mu\nu} = f_{\alpha\beta}\,\delta^\alpha{}_\mu\delta^\beta{}_\nu$ (the deltas pick out $\alpha = \mu$, $\beta = \nu$). Then
$$
f_{\alpha\beta}\Big[\tilde\omega^{\alpha\beta}(\vec e_\mu, \vec e_\nu) - \delta^\alpha{}_\mu\delta^\beta{}_\nu\Big] = 0\quad\text{for all } f_{\alpha\beta}.
$$
Choose $\mathbf f$ with a single nonzero component, $f_{\alpha\beta} = 1$ for one particular pair $(\alpha, \beta) = (\rho, \sigma)$ and $0$ otherwise. The sum collapses to the single bracket for $(\rho, \sigma)$, which must vanish. Doing this for each of the 16 pairs:
$$
\tilde\omega^{\alpha\beta}(\vec e_\mu, \vec e_\nu) = \delta^\alpha{}_\mu\,\delta^\beta{}_\nu .\qquad(3.24)\ \blacksquare
$$

**6. What it means (leading to 3.25).** $\delta^\alpha{}_\mu = \tilde\omega^\alpha(\vec e_\mu)$ and $\delta^\beta{}_\nu = \tilde\omega^\beta(\vec e_\nu)$ (Eq. 3.12). So $\tilde\omega^{\alpha\beta}$ gives on $(\vec e_\mu, \vec e_\nu)$ the product of the values of $\tilde\omega^\alpha$ on the first and $\tilde\omega^\beta$ on the second. Two bilinear functions that agree on all pairs of basis vectors agree everywhere, so $\tilde\omega^{\alpha\beta} = \tilde\omega^\alpha\otimes\tilde\omega^\beta$, Eq. (3.25).

This is exactly the argument used for one-forms: Eq. (3.11) $\Rightarrow$ Eq. (3.12), doubled.

```python
import sympy as sp
# check: with ω^{αβ} = ω^α ⊗ ω^β, f = f_{αβ} ω^{αβ} reproduces f on arbitrary vectors
F = sp.Matrix(4, 4, lambda a, b: sp.Symbol(f"f{a}{b}"))
A = sp.Matrix(sp.symbols("A0:4")); B = sp.Matrix(sp.symbols("B0:4"))
E = sp.eye(4)
omega = lambda a, b: (lambda U, W: (E[a, :] * U)[0] * (E[b, :] * W)[0])   # ω^α⊗ω^β on (U, W)
print("ω^{αβ}(e_μ,e_ν) = δδ :", all(omega(a, b)(E[:, m], E[:, n]) == (1 if (a, b) == (m, n) else 0)
                                     for a in range(4) for b in range(4) for m in range(4) for n in range(4)))
lhs = (A.T * F * B)[0]                                                    # f(A,B) = f_{αβ} A^α B^β
rhs = sum(F[a, b] * omega(a, b)(A, B) for a in range(4) for b in range(4))
print("f(A,B) − f_{αβ} ω^{αβ}(A,B) =", sp.expand(lhs - rhs))
```
