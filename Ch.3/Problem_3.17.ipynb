# Chapter 3, Problem 17 — when a $\binom02$ tensor is an outer product; $\binom11$ tensors as maps

**Problem.**
- **(a)** Suppose that $\mathbf h$ is a $\binom02$ tensor with the property that, for any two vectors $\vec A$ and $\vec B$ (where $\vec B \neq 0$),
$$
\mathbf h(\ \cdot\ , \vec A) = \alpha\,\mathbf h(\ \cdot\ , \vec B),
$$
where $\alpha$ is a number which may depend on $\vec A$ and $\vec B$. Show that there exist one-forms $\tilde p$ and $\tilde q$ such that $\mathbf h = \tilde p\otimes\tilde q$.
- **(b)** Suppose $\mathsf T$ is a $\binom11$ tensor, $\tilde\omega$ a one-form, $\vec v$ a vector, and $\mathsf T(\tilde\omega; \vec v)$ the value of $\mathsf T$ on $\tilde\omega$ and $\vec v$. Prove that $\mathsf T(\ \cdot\ ; \vec v)$ is a vector and $\mathsf T(\tilde\omega;\ \cdot\ )$ is a one-form, i.e. that a $\binom11$ tensor provides a map of vectors to vectors and one-forms to one-forms.

> **Reading of (a).** The intended hypothesis is that all the one-forms $\mathbf h(\cdot, \vec A)$ are multiples of a single one. To avoid the degenerate case it is cleanest to fix one $\vec B$ with $\mathbf h(\cdot, \vec B) \neq 0$ (if no such $\vec B$ exists, $\mathbf h = 0 = 0\otimes 0$, trivially a product).

## (a) Rank-one tensors are outer products

**Step 1: every slice is a multiple of one fixed one-form.** For fixed $\vec A$, the map $\vec X \mapsto \mathbf h(\vec X, \vec A)$ is linear in $\vec X$, so $\mathbf h(\cdot, \vec A)$ is a one-form. Fix a vector $\vec B$ with $\tilde p := \mathbf h(\cdot, \vec B) \neq 0$. By hypothesis, for every $\vec A$,
$$
\mathbf h(\cdot, \vec A) = \alpha(\vec A)\,\tilde p .
$$
(The dependence on $\vec B$ is now frozen, so write $\alpha(\vec A)$.)

**Step 2: $\alpha$ is a linear function of $\vec A$, i.e. a one-form.** Since $\tilde p \neq 0$ there's a vector $\vec X_0$ with $\tilde p(\vec X_0) \neq 0$. Evaluate the equation at $\vec X_0$:
$$
\alpha(\vec A) = \frac{\mathbf h(\vec X_0, \vec A)}{\tilde p(\vec X_0)} .
$$
The right side is linear in $\vec A$ (because $\mathbf h$ is linear in its second slot) and real-valued. So $\alpha$ is a one-form; call it $\tilde q$: $\tilde q(\vec A) := \alpha(\vec A)$.

**Step 3: assemble.** For all $\vec X, \vec A$:
$$
\mathbf h(\vec X, \vec A) = \alpha(\vec A)\,\tilde p(\vec X) = \tilde p(\vec X)\,\tilde q(\vec A) = (\tilde p\otimes\tilde q)(\vec X, \vec A).
$$
Hence $\mathbf h = \tilde p\otimes\tilde q$. $\blacksquare$

**In components.** The hypothesis says every column of the matrix $h_{\alpha\beta}$ (fix $\beta$, i.e. $\vec A = \vec e_\beta$) is a multiple of one column vector $p_\alpha$: the matrix has **rank 1**. A rank-1 matrix is always an outer product $p_\alpha q_\beta$. A generic $\binom02$ tensor has rank 4 and is *not* a single outer product, only a sum of them (Eq. 3.26).

## (b) A $\binom11$ tensor maps vectors to vectors and one-forms to one-forms

$\mathsf T$ is a function of one one-form and one vector, linear in each.

**$\mathsf T(\ \cdot\ ; \vec v)$ is a vector.** Fix $\vec v$. The map $\tilde\omega \mapsto \mathsf T(\tilde\omega; \vec v)$ takes one-forms to real numbers and is linear by the linearity of $\mathsf T$ in its first slot:
$$
\mathsf T(a\tilde\omega + b\tilde\sigma; \vec v) = a\,\mathsf T(\tilde\omega; \vec v) + b\,\mathsf T(\tilde\sigma; \vec v).
$$
A linear real-valued function of one-forms is a vector (§3.5: vectors are $\binom10$ tensors, $\vec V(\tilde p) := \tilde p(\vec V)$, Eq. 3.54). Its components are $\mathsf T(\tilde\omega^\alpha; \vec v) = T^\alpha{}_\beta v^\beta$. So $\vec v \mapsto \mathsf T(\cdot; \vec v)$ is a map of vectors to vectors, and it's linear: in components, $v^\beta \mapsto T^\alpha{}_\beta v^\beta$, multiplication by the matrix $T^\alpha{}_\beta$. $\blacksquare$

**$\mathsf T(\tilde\omega;\ \cdot\ )$ is a one-form.** Fix $\tilde\omega$. The map $\vec v \mapsto \mathsf T(\tilde\omega; \vec v)$ is real-valued and linear (linearity in the second slot), which is the definition of a one-form. Its components are $\mathsf T(\tilde\omega; \vec e_\beta) = \omega_\alpha T^\alpha{}_\beta$. So $\tilde\omega \mapsto \mathsf T(\tilde\omega; \cdot)$ maps one-forms to one-forms: $\omega_\alpha \mapsto \omega_\alpha T^\alpha{}_\beta$, the same matrix acting from the other side. $\blacksquare$

**Example:** the identity $\binom11$ tensor $\delta^\alpha{}_\beta$ maps every vector to itself and every one-form to itself.

```python
import sympy as sp
# (a) a rank-1 matrix factors as p q^T
p = sp.Matrix([2, -1, 0, 3]); q = sp.Matrix([1, 4, -2, 0])
H = p * q.T
print("rank of p⊗q:", H.rank())
# reconstruct per the proof: p̃ = h(·, B) for some B; q̃(A) = h(X0, A)/p̃(X0)
B = sp.Matrix([1, 0, 0, 0]); ptil = H * B            # column h(·,B) (components h_{α0})
X0 = sp.Matrix([1, 0, 0, 0]); qtil = (X0.T * H).T / (X0.T * ptil)[0]
print("p̃ ⊗ q̃ reconstructs h:", ptil * qtil.T == H)
print("a generic matrix has rank", sp.Matrix(4, 4, lambda i, j: (i + 1)**j + i*j).rank(), "→ not a single outer product")
# (b) T(·; v) components T^α_β v^β, T(ω; ·) components ω_α T^α_β
T = sp.Matrix(4, 4, lambda a, b: sp.Symbol(f"T{a}{b}")); v = sp.Matrix(sp.symbols("v0:4")); w = sp.Matrix(sp.symbols("w0:4"))
print("T(ω;v) = ω·(T v) = (ωT)·v:", sp.expand((w.T * (T * v))[0] - ((w.T * T) * v)[0]) == 0)
```
