# Chapter 1, Problem 15 — the ultra-relativistic limits

**Problem (paraphrased).** Suppose the velocity of $\bar{\mathcal O}$ relative to $\mathcal O$ is nearly that of light, $|v| = 1 - \varepsilon$ with $0 < \varepsilon \ll 1$. Show that the formulae of Problem 14 become

- **(a)** $\Delta t \approx \Delta\bar t\,/\sqrt{2\varepsilon}$,
- **(b)** $\Delta x \approx \Delta\bar x\,\sqrt{2\varepsilon}$,
- **(c)** $w' \approx 1 - \varepsilon\,(1-w)/(1+w)$.

What are the relative errors of these approximations when $\varepsilon = 0.1$ and $w = 0.9$?

> **Note on (b).** The book prints (b) as $\Delta x \approx \Delta\bar x/\sqrt{2\varepsilon}$. The Lorentz contraction is $\Delta x = \Delta\bar x\sqrt{1-v^2}$, which *shrinks* as $v\to1$. So the correct limit is $\Delta x \approx \Delta\bar x\,\sqrt{2\varepsilon}$, multiplied rather than divided. I derive that version below; with division the "approximation" would be off by a factor $1/(2\varepsilon)$.

## The key factor: $1 - v^2$ near $v = 1$

Factor the difference of squares, with $|v| = 1-\varepsilon$:
$$
1 - v^2 = (1 - |v|)(1 + |v|) = \varepsilon\,(2 - \varepsilon) = 2\varepsilon\left(1 - \frac{\varepsilon}{2}\right).
$$
To leading order in $\varepsilon$: $\boxed{1 - v^2 \approx 2\varepsilon}$, with fractional error $-\varepsilon/2$.

## (a) Time dilation

$$
\Delta t = \frac{\Delta\bar t}{\sqrt{1-v^2}} = \frac{\Delta\bar t}{\sqrt{2\varepsilon}}\left(1 - \frac{\varepsilon}{2}\right)^{-1/2} \approx \frac{\Delta\bar t}{\sqrt{2\varepsilon}}\left(1 + \frac{\varepsilon}{4}\right)
\quad\Longrightarrow\quad \boxed{\Delta t \approx \frac{\Delta\bar t}{\sqrt{2\varepsilon}}}
$$
Leading relative error: approximate/exact $- 1 \approx -\varepsilon/4$.

## (b) Lorentz contraction

$$
\Delta x = \Delta\bar x\sqrt{1 - v^2} = \Delta\bar x\sqrt{2\varepsilon}\left(1 - \frac{\varepsilon}{2}\right)^{1/2} \approx \Delta\bar x\sqrt{2\varepsilon}\left(1 - \frac{\varepsilon}{4}\right)
\quad\Longrightarrow\quad \boxed{\Delta x \approx \Delta\bar x\,\sqrt{2\varepsilon}}
$$
Leading relative error $\approx +\varepsilon/4$.

## (c) Velocity addition (with $v = 1 - \varepsilon > 0$)

$$
w' = \frac{w + 1 - \varepsilon}{1 + w(1-\varepsilon)} = \frac{(1 + w) - \varepsilon}{(1 + w) - w\varepsilon}.
$$
Subtract from 1 to see how close $w'$ is to light speed:
$$
1 - w' = \frac{(1+w) - w\varepsilon - (1+w) + \varepsilon}{(1+w) - w\varepsilon} = \frac{\varepsilon(1 - w)}{(1 + w) - w\varepsilon}.
$$
To first order in $\varepsilon$ the $w\varepsilon$ in the denominator is negligible:
$$
\boxed{w' \approx 1 - \varepsilon\,\frac{1 - w}{1 + w}}
$$
Physically: $\bar{\mathcal O}$ is already within $\varepsilon$ of light speed, and adding any $w$ leaves the result within an even smaller distance $\varepsilon(1-w)/(1+w)$ of light speed. Adding velocities can never cross $c$. (This holds for $-1 < w < 1$; for $w\to -1$ the formula gives $1-w'\to\infty\cdot\varepsilon$ and breaks down, because $w'$ is then not close to 1.)

```python
import sympy as sp
e, w = sp.symbols("epsilon w", positive=True)
v = 1 - e
print("1 − v² =", sp.factor(1 - v**2))
print("γ expansion:   ", sp.series(1/sp.sqrt(1 - v**2), e, 0, 2))
print("√(1−v²) expansion:", sp.series(sp.sqrt(1 - v**2), e, 0, 2))
wp = (w + v)/(1 + w*v)
print("1 − w' exact  =", sp.simplify(1 - wp))
print("1 − w' series =", sp.series(sp.simplify(1 - wp), e, 0, 3))
```

## Relative errors at $\varepsilon = 0.1$, $w = 0.9$

Here $v = 0.9$ and $1 - v^2 = 0.19$ (versus $2\varepsilon = 0.2$).

**(a)** exact $\gamma = 1/\sqrt{0.19} = 2.2942$; approx $1/\sqrt{0.2} = 2.2361$. Relative error $= 2.2361/2.2942 - 1 = -2.5\%$ (cf. $-\varepsilon/4 = -2.5\%$).

**(b)** exact $\sqrt{0.19} = 0.43589$; approx $\sqrt{0.2} = 0.44721$. Relative error $= +2.6\%$.

**(c)** exact $w' = (0.9 + 0.9)/(1 + 0.81) = 1.8/1.81 = 0.994475$; approx $1 - 0.1\times0.1/1.9 = 0.994737$. Relative error in $w'$: $+2.6\times10^{-4}$. A fairer measure is the relative error in the small quantity $1 - w'$: exact $0.005525$, approx $0.005263$, i.e. $-4.7\%$.

Even with $\varepsilon = 0.1$, which is hardly "$\ll 1$", the approximations are good to a few percent. The errors fall in proportion to $\varepsilon$.

```python
import numpy as np
eps, W = 0.1, 0.9
V = 1 - eps
rows = [
    ("(a) Δt/Δt̄", 1/np.sqrt(1 - V**2), 1/np.sqrt(2*eps)),
    ("(b) Δx/Δx̄", np.sqrt(1 - V**2), np.sqrt(2*eps)),
    ("(c) w'", (W + V)/(1 + W*V), 1 - eps*(1 - W)/(1 + W)),
    ("(c) 1 − w'", 1 - (W + V)/(1 + W*V), eps*(1 - W)/(1 + W)),
]
for name, ex, ap in rows:
    print(f"{name:12s} exact = {ex:.6f}  approx = {ap:.6f}  relative error = {(ap - ex)/ex:+.2%}")
print("\n(book's printed (b), Δx̄/√(2ε):", f"{1/np.sqrt(2*eps):.4f} vs exact {np.sqrt(1 - V**2):.4f} — off by a factor 1/(2ε))")
```
