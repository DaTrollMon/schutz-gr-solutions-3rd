# Chapter 1, Problem 13 — the half-life of a moving pion

**Problem (paraphrased).** The half-life of the pi meson (pion) is $2.5\times10^{-8}$ s when the pion is at rest relative to the observer measuring its decay time. Show, by the principle of relativity, that pions moving at $v = 0.999$ must have a half-life of $5.6\times10^{-7}$ s as measured by an observer at rest.

## Step 1 — the principle of relativity fixes the *proper* half-life

The pion's decay is internal physics. Its rest frame is an inertial frame like any other, so by the principle of relativity the decay law in that frame is the same as for a pion at rest in the lab. In particular: half of a sample of pions decays in a **proper time**
$$
\tau_{1/2} = 2.5\times10^{-8}\ \text{s},
$$
measured by a clock moving with the pions — whatever their velocity relative to anyone else. A pion moving at $v = 0.999$ relative to the lab has a rest frame $\bar{\mathcal O}$, in which it is at rest, so in $\bar{\mathcal O}$ its half-life is $2.5\times10^{-8}$ s.

## Step 2 — relate proper time to lab time (time dilation)

Consider two events on the pion's world line: creation, and the moment by which half the sample has decayed. In the pion frame they're at the same place, separated by $\Delta\bar t = \tau_{1/2}$. In the lab frame $\mathcal O$ they're separated by $\Delta t$ and $\Delta x = v\,\Delta t$. Invariance of the interval:
$$
-(\Delta\bar t)^2 = -(\Delta t)^2 + (\Delta x)^2 = -(\Delta t)^2(1 - v^2)
\quad\Longrightarrow\quad
\Delta t = \frac{\Delta\bar t}{\sqrt{1 - v^2}} = \gamma\,\tau_{1/2}. \tag{Eq. 1.8}
$$

## Step 3 — numbers

$$
1 - v^2 = 1 - 0.999^2 = 1 - 0.998001 = 0.001999,
\qquad \sqrt{0.001999} = 0.044710,
\qquad \gamma = \frac{1}{0.044710} = 22.366 .
$$
$$
\Delta t_{1/2} = 22.366\times2.5\times10^{-8}\ \text{s} = \boxed{5.6\times10^{-7}\ \text{s}}
$$

A handy shortcut for $v$ close to 1: $1 - v^2 = (1-v)(1+v) \approx 2(1 - v) = 0.002$, giving $\gamma \approx 1/\sqrt{0.002} = 22.36$ (Problem 15 develops this).

```python
import numpy as np
tau = 2.5e-8          # s, proper half-life
v = 0.999
g = 1 / np.sqrt(1 - v**2)
print(f"1 − v² = {1 - v**2:.6f},  γ = {g:.4f}")
print(f"lab half-life = γ τ = {g * tau:.3e} s")
print(f"approximation γ ≈ 1/sqrt(2(1−v)) = {1/np.sqrt(2*(1 - v)):.4f}")
```

## What it means physically

**Decay length.** In one lab half-life a pion at $v = 0.999$ travels
$$
v\,\Delta t = 0.999 \times 5.6\times10^{-7}\ \text{s} \times 3\times10^{8}\ \text{m s}^{-1} \approx 168\ \text{m}.
$$
Without time dilation it would be $0.999 \times 2.5\times10^{-8} \times 3\times10^{8} \approx 7.5$ m. This is directly observable: pion beams at accelerators survive over distances ~22 times longer than a naive calculation predicts. In $c=1$ units: $\tau_{1/2} = 2.5\times10^{-8}\times3\times10^{8} = 7.5$ m of time, lab half-life $\approx 168$ m of time.

**Consistency check from the pion's frame.** The pion "sees" the lab moving at 0.999. A lab distance of 168 m is Lorentz-contracted to $168/\gamma \approx 7.5$ m, which passes by in the proper half-life of $7.5$ m of time. The two descriptions agree.

```python
c = 3e8
print(f"distance in one lab half-life: {v * g * tau * c:.0f} m   (naive, no dilation: {v * tau * c:.2f} m)")
print(f"that distance contracted in the pion frame: {v * g * tau * c / g:.2f} m")
```
