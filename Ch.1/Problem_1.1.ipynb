## The method

Setting $c = 1$ is a statement that time and distance are the *same kind* of quantity, related by a fixed conversion factor:

$$
1 = c = 2.998\times10^{8}\ \text{m s}^{-1}
\quad\Longrightarrow\quad
1\ \text{s} = 2.998\times10^{8}\ \text{m} \approx 3\times10^{8}\ \text{m}.
$$

So every second appearing in an SI unit can be replaced by $c$ metres. If a quantity has SI units $\text{kg}^{a}\,\text{m}^{b}\,\text{s}^{n}$, then

$$
X\ \text{kg}^{a}\,\text{m}^{b}\,\text{s}^{n}
= X\ \text{kg}^{a}\,\text{m}^{b}\,\big(c\ \text{m}\big)^{n}
= \big(X\,c^{\,n}\big)\ \text{kg}^{a}\,\text{m}^{\,b+n}.
$$

Recipe for every part:

1. Reduce the SI unit to base units $\text{kg}^a\,\text{m}^b\,\text{s}^n$ (e.g. $\text{J} = \text{kg m}^2\text{s}^{-2}$).
2. Multiply the number by $c^{\,n}$ (with $c = 3\times10^8$, the value Schutz uses; the exact $c$ is shown alongside).
3. Add $n$ to the power of metres; the seconds are gone.

A useful special case: since $1\ \text{s}^{-1} = (3\times10^{8})^{-1}\ \text{m}^{-1}$, **dividing by a second ⇒ dividing by $3\times10^{8}$ and by a metre**.

```python
from schutz_gr import Quantity, to_natural, C_SCHUTZ, C_EXACT

c = C_SCHUTZ      # Schutz's rounded value, 3e8 m/s
print(f"c (Schutz) = {C_SCHUTZ:.3e} m/s,   c (exact) = {C_EXACT:.9e} m/s")
print(f"c^2 = {c**2:.2e},   c^3 = {c**3:.2e}")
```

## (a) $10\ \text{J}$ — the worked example

**Step 1 — base units.** $1\ \text{J} = 1\ \text{N m} = 1\ \text{kg m s}^{-2}\cdot\text{m} = 1\ \text{kg m}^2\text{s}^{-2}$, so $a=1,\ b=2,\ n=-2$.

**Step 2 — replace seconds.** $\text{s}^{-2} = (3\times10^{8}\ \text{m})^{-2} = \dfrac{1}{9\times10^{16}}\ \text{m}^{-2}$.

**Step 3 — combine.**
$$
10\ \text{J} = 10\ \text{kg m}^2 \times \frac{1}{9\times10^{16}\ \text{m}^2}
= \frac{10}{9\times10^{16}}\ \text{kg}
= \boxed{1.1\times10^{-16}\ \text{kg}}
$$

**Interpretation.** With $c=1$, $E = mc^2$ becomes $E = m$: energy *is* mass, measured in kilograms. $10$ J is the rest-mass energy of about $10^{-16}$ kg of matter — this is how much mass a battery loses when it delivers 10 J.

## (b) Power, $100\ \text{W}$

**Step 1.** $1\ \text{W} = 1\ \text{J s}^{-1} = 1\ \text{kg m}^2\text{s}^{-3}$, so $n = -3$.

**Step 2.** $\text{s}^{-3} = (3\times10^{8}\ \text{m})^{-3} = \dfrac{1}{2.7\times10^{25}}\ \text{m}^{-3}$.

**Step 3.**
$$
100\ \text{W} = 100\ \text{kg m}^2\times\frac{1}{2.7\times10^{25}\ \text{m}^3}
= \frac{100}{2.7\times10^{25}}\ \text{kg m}^{-1}
= \boxed{3.7\times10^{-24}\ \text{kg m}^{-1}}
$$

**Check the dimension independently.** Power = energy / time. Energy is in kg (part a), time is in m, so power is in $\text{kg m}^{-1}$ ✓. The bulb converts $3.7\times10^{-24}$ kg of mass-energy into light and heat for every metre of (light-travel) time — i.e. every $3.3$ ns.

## (c) Planck's reduced constant, $\hbar = 1.05\times10^{-34}\ \text{J s}$

**Step 1.** $\text{J s} = \text{kg m}^2\text{s}^{-2}\cdot\text{s} = \text{kg m}^2\text{s}^{-1}$, so $n = -1$. (These are the units of angular momentum.)

**Step 2.** $\text{s}^{-1} = \dfrac{1}{3\times10^{8}}\ \text{m}^{-1}$.

**Step 3.**
$$
\hbar = 1.05\times10^{-34}\ \text{kg m}^2\times\frac{1}{3\times10^{8}\ \text{m}}
= \boxed{3.5\times10^{-43}\ \text{kg m}}
$$

**Check.** Energy × time → kg × m ✓; equivalently momentum × length, where momentum is in kg (see part e) and length in m ✓.

## (d) Velocity of a car, $30\ \text{m s}^{-1}$

**Step 1.** $\text{m s}^{-1}$: $a=0,\ b=1,\ n=-1$.

**Step 2–3.**
$$
30\ \text{m s}^{-1} = 30\ \text{m}\times\frac{1}{3\times10^{8}\ \text{m}} = \boxed{10^{-7}}\quad(\text{dimensionless})
$$

**Interpretation.** With $c=1$ a velocity is a pure number: the fraction of the speed of light. Light is $v=1$; a car is $v = 10^{-7}$. This is why the Newtonian approximation ($v\ll1$) works so well in daily life: relativistic corrections go like $v^2 \sim 10^{-14}$.

## (e) Momentum of a car, $3\times10^{4}\ \text{kg m s}^{-1}$

**Step 1.** $\text{kg m s}^{-1}$: $a=1,\ b=1,\ n=-1$.

**Step 2–3.**
$$
3\times10^{4}\ \text{kg m s}^{-1} = 3\times10^{4}\ \text{kg m}\times\frac{1}{3\times10^{8}\ \text{m}}
= \boxed{10^{-4}\ \text{kg}}
$$

**Check.** $p = mv$ with $v$ dimensionless (part d) ⇒ momentum has the units of mass ✓. Consistency: a car of mass $10^{3}$ kg at $v = 10^{-7}$ has $p = 10^{3}\times10^{-7} = 10^{-4}$ kg, matching exactly ($3\times10^{4}/30 = 10^{3}$ kg).

So energy and momentum share the unit kg — a first hint that they are components of one object, the four-momentum (Chapter 2).

## (f) Pressure of one atmosphere, $10^{5}\ \text{N m}^{-2}$

**Step 1.** $\text{N m}^{-2} = \text{kg m s}^{-2}\cdot\text{m}^{-2} = \text{kg m}^{-1}\text{s}^{-2}$, so $a=1,\ b=-1,\ n=-2$.

**Step 2.** $\text{s}^{-2} = \dfrac{1}{9\times10^{16}}\ \text{m}^{-2}$.

**Step 3.**
$$
10^{5}\ \text{kg m}^{-1}\text{s}^{-2} = \frac{10^{5}}{9\times10^{16}}\ \text{kg m}^{-1}\,\text{m}^{-2}
= \boxed{1.1\times10^{-12}\ \text{kg m}^{-3}}
$$

**Interpretation.** Pressure has the *same units as density* when $c = 1$. That's no accident: pressure is force per area = energy per volume, and energy is mass. Compare with (g): atmospheric pressure is $10^{-15}$ of the mass-density of water — which is why pressure is negligible as a *source of gravity* in everyday situations, though not inside neutron stars (Chapter 10).

## (g) Density of water, $10^{3}\ \text{kg m}^{-3}$

**Step 1.** $\text{kg m}^{-3}$ contains no seconds: $n = 0$.

**Step 2–3.** Nothing to convert:
$$
\boxed{10^{3}\ \text{kg m}^{-3}}
$$

Quantities built only from kg and m are unchanged by setting $c=1$.

## (h) Luminosity flux, $10^{6}\ \text{J s}^{-1}\,\text{cm}^{-2}$

**Step 0 — get to SI first.** $1\ \text{cm}^{-2} = (10^{-2}\ \text{m})^{-2} = 10^{4}\ \text{m}^{-2}$, so
$$
10^{6}\ \text{J s}^{-1}\text{cm}^{-2} = 10^{6}\times10^{4}\ \text{J s}^{-1}\text{m}^{-2} = 10^{10}\ \text{W m}^{-2}.
$$

**Step 1.** $\text{W m}^{-2} = \text{kg m}^{2}\text{s}^{-3}\cdot\text{m}^{-2} = \text{kg s}^{-3}$, so $a=1,\ b=0,\ n=-3$.

**Step 2–3.**
$$
10^{10}\ \text{kg s}^{-3} = \frac{10^{10}}{2.7\times10^{25}}\ \text{kg m}^{-3}
= \boxed{3.7\times10^{-16}\ \text{kg m}^{-3}}
$$

**Interpretation.** Energy flux is energy per area per time; with $c = 1$ that is (kg)/(m²·m) = kg m⁻³, the same as an energy density. Physically: radiation moving at speed 1 with energy density $u$ carries flux $u\cdot 1$, so flux and energy density are literally the same number for a beam of light.

## Check every part in code

`to_natural` implements $X\,\text{kg}^a\text{m}^b\text{s}^n \to X c^n\,\text{kg}^a\text{m}^{b+n}$. The table shows the result with Schutz's $c = 3\times10^8$ and with the exact $c$.

```python
parts = {
    "(a) energy 10 J":                 Quantity(10,      kg=1, m=2,  s=-2),
    "(b) power 100 W":                 Quantity(100,     kg=1, m=2,  s=-3),
    "(c) hbar 1.05e-34 J s":           Quantity(1.05e-34, kg=1, m=2, s=-1),
    "(d) velocity 30 m/s":             Quantity(30,              m=1,  s=-1),
    "(e) momentum 3e4 kg m/s":         Quantity(3e4,     kg=1, m=1,  s=-1),
    "(f) pressure 1e5 N/m^2":          Quantity(1e5,     kg=1, m=-1, s=-2),
    "(g) density 1e3 kg/m^3":          Quantity(1e3,     kg=1, m=-3),
    "(h) flux 1e6 J/s/cm^2 = 1e10 W/m^2": Quantity(1e6 * 1e4, kg=1, s=-3),
}

print(f"{'part':38s} {'SI unit':16s} {'c = 3e8':>12s}  {'exact c':>12s}  natural unit")
for name, q in parts.items():
    n1, n2 = to_natural(q, c=C_SCHUTZ), to_natural(q, c=C_EXACT)
    print(f"{name:38s} {q.unit:16s} {n1.value:12.3e}  {n2.value:12.3e}  {n1.unit}")
```

**Patterns worth remembering** (they recur all through the book):

- energy, momentum, mass → kg
- velocity → dimensionless
- pressure, energy density, energy flux, mass density → $\text{kg m}^{-3}$
- power → $\text{kg m}^{-1}$
