# Chapter 1, Problem 9 — why the clock's world line in Fig. 1.7 is drawn as it is

**Problem (paraphrased).** Explain why the line PQ in Fig. 1.7 is drawn in the manner described in the text.

**The setup (§1.6, Figs. 1.6–1.7).** A rod at rest in $\mathcal O$ lies along the $y$ axis. Its ends at $t = 0$ are events $\mathcal A$ and $\mathcal B$. Fig. 1.7 adds the world line of one clock of $\bar{\mathcal O}$ (which moves at $v$ along $x$). The text says the line is drawn

> perpendicular to the $y$ axis and parallel to the $t$–$x$ plane, i.e. parallel to the $\bar t$ axis,

and that the clock is "the one clock in $\bar{\mathcal O}$ which passes through the $y$ axis at $t=0$" and can send out light rays at event $\mathcal P$ that reach $\mathcal A$ and $\mathcal B$. The reflections come back to the clock at a single event (called $\mathcal L$ in the text).

> **Labelling note.** The exercise says "PQ", but the 3rd-edition text calls the two events on the line $\mathcal P$ (emission) and $\mathcal L$ (return). "PQ" seems to be a leftover from an earlier edition's labels. Either way it means this clock's world line through the emission and return events.

The world line has three features to justify: (1) its direction, parallel to the $\bar t$ axis; (2) constant $y$ (and $z$), i.e. perpendicular to the $y$ axis; (3) its particular position, through the midpoint of the rod at $t=0$.

## (1) Parallel to the $\bar t$ axis

It is the world line of a clock **of $\bar{\mathcal O}$'s frame**, so the clock is at rest relative to $\bar{\mathcal O}$. Every such clock moves at velocity $v$ in $x$ relative to $\mathcal O$, so its world line has $dx/dt = v$ — the same direction as $\bar{\mathcal O}$'s own time axis, $x = vt$ (§1.5). Clocks of $\bar{\mathcal O}$ at other spatial positions just have world lines parallel to the $\bar t$ axis, displaced.

## (2) Constant $y$ and $z$ — perpendicular to the $y$ axis

$\bar{\mathcal O}$'s motion is purely along $x$, so the clock's $y$ and $z$ never change. Its world line therefore lies in a plane $y = \text{const}$, which is perpendicular to the $y$ axis and parallel to the $t$–$x$ plane. That's what the text means by "perpendicular to the $y$ axis and parallel to the $t$–$x$ plane".

## (3) Why *this* clock — through the midpoint of the rod at $t = 0$

The purpose of the construction is to let $\bar{\mathcal O}$ test whether $\mathcal A$ and $\mathcal B$ are simultaneous using his own radar method (§1.5): send light from one of his clocks, let it reflect at the two events, and see whether the echoes come back together. The echoes return at the **same event** only if the clock is **equidistant** from the two reflection points in $\bar{\mathcal O}$'s frame. Then the out-and-back light paths have equal lengths, and since light has speed 1 for $\bar{\mathcal O}$, equal travel times.

- In $\mathcal O$'s frame, the set of points equidistant from the two ends of the rod is the perpendicular bisector plane $y = y_{\text{mid}}$. A clock that stays in that plane for all time needs constant $y = y_{\text{mid}}$, which is property (2) again — so (2) is necessary, not decorative.
- The clock must actually be *at* the rod when the reflections happen, at $t = 0$, so that a single emission $\mathcal P$ can reach both ends at $t=0$ and both echoes can meet again. Hence "the one clock which passes through the $y$ axis at $t = 0$", at the rod's midpoint. (A clock elsewhere in the plane $y = y_{\text{mid}}$ would also be equidistant, but its rays would not hit $\mathcal A$ and $\mathcal B$ at the given events. The text picks the one whose geometry is cleanest.)
- By the mirror symmetry $y - y_{\text{mid}} \to -(y - y_{\text{mid}})$, which leaves the clock's world line fixed and swaps $\mathcal A \leftrightarrow \mathcal B$, the two light paths $\mathcal P\to\mathcal A\to\mathcal L$ and $\mathcal P\to\mathcal B\to\mathcal L$ are mirror images. So they must return at the same event $\mathcal L$.

**Conclusion drawn in the text.** $\bar{\mathcal O}$ emits at $\mathcal P$ and receives both echoes at $\mathcal L$, so by his own definition of simultaneity $\mathcal A$ and $\mathcal B$ are simultaneous (both at the $\bar t$ midway between $\mathcal P$ and $\mathcal L$). Then the interval $\Delta\bar s^2_{\mathcal{AB}}$ is the square of the rod's length *as measured by* $\bar{\mathcal O}$, which is what the first half of the proof of $\phi(v) = \phi(|v|)$ needs.

## Explicit coordinates (a check)

Put the rod from $\mathcal A = (t,x,y) = (0,0,0)$ to $\mathcal B = (0,0,2\ell)$; midpoint $y = \ell$. The clock's world line is $(t,\ x = vt,\ y = \ell)$.

**Emission $\mathcal P$** at $t_P < 0$, position $(vt_P, \ell)$. Light reaches $\mathcal A$ at $t=0$ if the elapsed time equals the distance:
$$
-t_P = \sqrt{(v t_P)^2 + \ell^2} \;\Rightarrow\; t_P^2(1 - v^2) = \ell^2 \;\Rightarrow\; t_P = -\gamma\ell .
$$
The distance to $\mathcal B$ is identical ($|y_B - \ell| = |y_A - \ell| = \ell$), so the same emission reaches both ends at $t = 0$ ✓.

**Return $\mathcal L$** at $t_L > 0$: $t_L = \sqrt{(vt_L)^2 + \ell^2} \Rightarrow t_L = +\gamma\ell$, the same for both echoes ✓.

**In $\bar{\mathcal O}$'s frame** the clock is at rest; its own time reads $\bar t = t/\gamma$ along it, so $\bar t_P = -\ell$ and $\bar t_L = +\ell$. The midpoint is $\bar t = 0$. And indeed the Lorentz transformation gives $\bar t_{\mathcal A} = \gamma(0 - v\cdot0) = 0 = \bar t_{\mathcal B}$ ✓. The radar distance from the clock to each end is $(\bar t_L - \bar t_P)/2 = \ell$ in $\bar{\mathcal O}$'s frame. So $\bar{\mathcal O}$ measures the rod as $2\ell$ — unchanged, as the text concludes for lengths perpendicular to the motion.

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from schutz_gr.diagrams import COLORS
%config InlineBackend.figure_format = 'retina'

v_, l_ = sp.symbols("v ell", positive=True)
tP = sp.symbols("t_P", negative=True); tL = sp.symbols("t_L", positive=True)
print("t_P =", sp.solve(sp.Eq(-tP, sp.sqrt((v_*tP)**2 + l_**2)), tP))
print("t_L =", sp.solve(sp.Eq(tL, sp.sqrt((v_*tL)**2 + l_**2)), tL))
```

```python
v, ell = 0.5, 1.0
g = 1 / np.sqrt(1 - v**2)
A, B = np.array([0, 0, 0.0]), np.array([0, 0, 2*ell])          # (t, x, y)
P = np.array([-g*ell, -g*ell*v, ell]); L = np.array([g*ell, g*ell*v, ell])

fig = plt.figure(figsize=(8, 7)); fig.patch.set_facecolor(COLORS["bg"])
ax = fig.add_subplot(projection="3d"); ax.set_facecolor(COLORS["bg"])
def seg(p, q, **kw): ax.plot([p[1], q[1]], [p[2], q[2]], [p[0], q[0]], **kw)   # plot as (x, y, t)

T = np.linspace(-1.8, 1.8, 2)
for yy, lab in ((0, "world lines of rod ends"), (2*ell, None)):
    ax.plot([0, 0], [yy, yy], T, color=COLORS["grey"], ls=":", lw=1.5, label=lab)
ax.plot(v*T, [ell, ell], T, color=COLORS["amber"], lw=2.4, label="clock of Ō: x = vt, y = ℓ")
for Q, c in ((A, "teal"), (B, "teal")):
    seg(P, Q, color=COLORS["gold"], lw=1.5); seg(Q, L, color=COLORS["gold"], lw=1.5)
ax.plot([], [], color=COLORS["gold"], label="light rays P→A,B→L")
seg(A, B, color=COLORS["teal"], lw=3, label="rod at t = 0 (A to B)")
for name, p in dict(A=A, B=B, P=P, L=L).items():
    ax.scatter(p[1], p[2], p[0], color=COLORS["text"], s=30)
    ax.text(p[1] + 0.08, p[2], p[0] + 0.08, name, color=COLORS["text"])
# the perpendicular-bisector plane y = ℓ
xx, tt = np.meshgrid(np.linspace(-1.2, 1.2, 2), np.linspace(-1.8, 1.8, 2))
ax.plot_surface(xx, np.full_like(xx, ell), tt, color=COLORS["violet"], alpha=0.12)
ax.set_xlabel("x", color=COLORS["teal"]); ax.set_ylabel("y", color=COLORS["teal"]); ax.set_zlabel("t", color=COLORS["teal"])
for axis in (ax.xaxis, ax.yaxis, ax.zaxis):
    axis.set_pane_color((0, 0, 0, 0)); axis.label.set_color(COLORS["teal"])
    axis._axinfo["grid"].update(color=COLORS["grid"], linewidth=0.6)
ax.tick_params(colors=COLORS["grey"])
ax.view_init(elev=14, azim=-58)
ax.set_title("Fig. 1.7 reconstructed (v = 0.5): the clock lies in the bisector plane y = ℓ",
             color=COLORS["text"], fontsize=11)
leg = ax.legend(loc="upper left", fontsize=8); leg.get_frame().set_facecolor(COLORS["bg"])
for t_ in leg.get_texts(): t_.set_color(COLORS["text"])
plt.tight_layout(); plt.show()
print("light-travel check |P→A| time vs distance:", -P[0], np.hypot(P[1], P[2]))
```

## Summary

The line through $\mathcal P$ and $\mathcal L$ is the world line of a clock of $\bar{\mathcal O}$. It is drawn
1. **parallel to the $\bar t$ axis**, because every clock of $\bar{\mathcal O}$ moves with velocity $v$ along $x$;
2. **at constant $y$ (perpendicular to the $y$ axis)**, because $\bar{\mathcal O}$ doesn't move in $y$, and because only a clock staying in the bisector plane $y = y_{\text{mid}}$ is always equidistant from the rod's two ends;
3. **through the rod's midpoint at $t = 0$**, so that one flash $\mathcal P$ reaches both ends at $\mathcal A$ and $\mathcal B$ and both echoes return together at $\mathcal L$.

Together these make the radar test symmetric, which proves $\mathcal A$ and $\mathcal B$ are simultaneous for $\bar{\mathcal O}$ as well as for $\mathcal O$. That's the key step in showing $\phi(v) = \phi(|v|)$.
