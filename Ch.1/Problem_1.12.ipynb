## The events of Fig. 1.14 (from §1.8)

$\bar{\mathcal O}$ moves at $v$ along $+x$; origins coincide.

| event | meaning | $(t, x)$ in $\mathcal O$ |
|---|---|---|
| A | origin: $\bar{\mathcal O}$'s clock passes $\mathcal O$'s clock at $x=0$; both read 0 | $(0, 0)$ |
| B | $\bar{\mathcal O}$'s clock reads $\bar t = 1$ (on the $\bar t$ axis, on the hyperbola $t^2 - x^2 = 1$) | $(\gamma,\ \gamma v)$ |
| D | where that hyperbola crosses $\mathcal O$'s $t$ axis | $(1,\ 0)$ |
| C | $\mathcal O$'s clock at $x = 0$ at the moment $\mathcal O$ regards as simultaneous with B | $(\gamma,\ 0)$ |
| E | $\mathcal O$'s clock at $x=0$ at the moment **$\bar{\mathcal O}$** regards as simultaneous with B | $(\sqrt{1-v^2},\ 0)$ |
| F | $\mathcal O$'s second clock (the one that meets B) at $t = 0$ | $(0,\ \gamma v)$ |

with $\gamma = 1/\sqrt{1-v^2}$. The caption of Fig. 1.14 says: "The proper length of AB is the time ticked by a clock at rest in $\bar{\mathcal O}$, while that of AC is the time it takes to do so as measured by $\mathcal O$".

## Locating E: the tangent at B is $\bar{\mathcal O}$'s line of simultaneity

The hyperbola through D and B is $t^2 - x^2 = 1$ (interval $-1$ from A). Differentiate implicitly: $2t\,dt - 2x\,dx = 0 \Rightarrow dt/dx = x/t$. At B $= (\gamma, \gamma v)$:
$$
\left.\frac{dt}{dx}\right|_B = \frac{\gamma v}{\gamma} = v ,
$$
the slope of $\bar{\mathcal O}$'s $\bar x$ axis, as stated in §1.7. Every line of constant $\bar t$ has this slope, so the tangent at B is the line $\bar t = 1$:
$$
t - \gamma = v\,(x - \gamma v)\quad\Longleftrightarrow\quad t = v x + \gamma(1 - v^2) = v x + \sqrt{1 - v^2}.
$$
(Check with the Lorentz transformation: $\bar t = \gamma(t - vx) = 1 \Leftrightarrow t = vx + 1/\gamma$ ✓.)

It meets $\mathcal O$'s time axis $x = 0$ at
$$
t_E = \sqrt{1 - v^2}\quad\Longrightarrow\quad E = (\sqrt{1-v^2},\ 0).
$$

## (a) AE is shorter than $\bar{\mathcal O}$'s clock time AB

The time $\bar{\mathcal O}$'s clock records from A to B is the proper time along AB:
$$
\tau_{AB} = \sqrt{-(\Delta s^2)_{AB}} = 1
$$
(B was chosen on the hyperbola $\Delta s^2 = -1$).

**Geometric argument.** D is on the same hyperbola, so $\tau_{AD} = 1$ too: AD has the same proper time as AB. The upper branch $t = \sqrt{1+x^2}$ is **convex** ($d^2t/dx^2 = (1+x^2)^{-3/2} > 0$), so it lies entirely above each of its tangent lines, touching only at the point of tangency. The tangent at B meets the $t$ axis at E. Since D is the hyperbola's point on that axis, E lies strictly below D for any $v \neq 0$:
$$
t_E < t_D = 1 \quad\Longrightarrow\quad AE < AD = \tau_{AB}.
$$
Both A and E are on the world line of $\mathcal O$'s clock at $x=0$, so AE is the time *that* clock ticks between A and E. And E is the event $\bar{\mathcal O}$ calls simultaneous with B. So during the time $\bar{\mathcal O}$'s own clock ticks 1, $\bar{\mathcal O}$ sees $\mathcal O$'s clock tick less: $AE = \sqrt{1-v^2} < 1$.

**Algebraic version.** $t_D - t_E = 1 - \sqrt{1-v^2} > 0$ for $0 < |v| < 1$.

## (b) The intervals

Using $\Delta s^2 = -(\Delta t)^2 + (\Delta x)^2$:
$$
(\Delta s^2)_{AB} = -\gamma^2 + \gamma^2 v^2 = -\gamma^2(1-v^2) = -1 ,
$$
$$
(\Delta s^2)_{AE} = -(\sqrt{1-v^2})^2 + 0 = -(1 - v^2) ,
$$
$$
(\Delta s^2)_{AC} = -\gamma^2 + 0 = -\frac{1}{1-v^2} .
$$
Therefore
$$
\boxed{(\Delta s^2)_{AE} = (1 - v^2)\,(\Delta s^2)_{AB}}\qquad\text{and}\qquad (\Delta s^2)_{AB} = (1-v^2)\,(\Delta s^2)_{AC}.
$$

The two relations are mirror images. The second is $\mathcal O$'s time dilation: $\mathcal O$ says $\bar{\mathcal O}$'s clock ticked $\tau_{AB} = \sqrt{1-v^2}\,\tau_{AC}$ while his own ticked $\tau_{AC}$. The first is $\bar{\mathcal O}$'s: $\bar{\mathcal O}$ says $\mathcal O$'s clock ticked $\tau_{AE} = \sqrt{1-v^2}\,\tau_{AB}$ while his own ticked $\tau_{AB}$.

These hold for any point B on the $\bar t$ axis, not just $\bar t = 1$, by linearity (all coordinates scale together). The code below checks it with B at a general $\bar t = \tau$.

```python
import sympy as sp
import numpy as np
from schutz_gr import inverse_lorentz, interval_tx
v, tau = sp.symbols("v tau", positive=True)
g = 1 / sp.sqrt(1 - v**2)

tB, xB = inverse_lorentz(tau, 0, v)                      # B: t̄ = τ, x̄ = 0
tangent = sp.solve(sp.Eq(g * (sp.Symbol("t") - v * sp.Symbol("x")), tau), sp.Symbol("t"))[0]
tE = sp.simplify(tangent.subs(sp.Symbol("x"), 0))
tC = tB
print("B =", (sp.simplify(tB), sp.simplify(xB)), "  E = (", tE, ", 0)   C = (", sp.simplify(tC), ", 0)")

sAB = sp.simplify(interval_tx(tB, xB)); sAE = sp.simplify(interval_tx(tE, 0)); sAC = sp.simplify(interval_tx(tC, 0))
print("Δs²_AB =", sAB, "  Δs²_AE =", sAE, "  Δs²_AC =", sAC)
print("Δs²_AE / Δs²_AB =", sp.simplify(sAE / sAB), "   Δs²_AB / Δs²_AC =", sp.simplify(sAB / sAC))
print("slope of hyperbola t²−x²=τ² at B:", sp.simplify((xB / tB)))
```

```python
from schutz_gr.diagrams import SpacetimeDiagram, COLORS
%config InlineBackend.figure_format = 'retina'
vf = 0.6; gf = 1/np.sqrt(1 - vf**2)
d = SpacetimeDiagram(xlim=(-0.5, 2.0), tlim=(-0.3, 1.9), title=f"Fig. 1.14 reconstructed (v = {vf})", size=7)
d.boosted_axes(vf, color="amber", label="Ō axes")
T = np.linspace(-0.3, 1.9, 400)
X = np.linspace(-0.5, 2.0, 400)
d.line(X, np.sqrt(1 + X**2), color="violet", lw=1.6, label="hyperbola DB: t² − x² = 1")
d.line(X, vf*X + np.sqrt(1 - vf**2), color="rose", ls="--", lw=1.4, label="tangent at B = Ō-simultaneity t̄ = 1")
d.line(X, gf + 0*X, color="teal", ls="--", lw=1.2, label="O-simultaneity t = γ")
d.line(gf*vf + 0*T, T, color="grey", ls=":", lw=1.4, label="O's 2nd clock (through F and B)")
for name, (t_, x_), off in (("A", (0, 0), (-12, -14)), ("B", (gf, gf*vf), (8, -4)), ("C", (gf, 0), (-16, 4)),
                             ("D", (1, 0), (-16, 2)), ("E", (np.sqrt(1 - vf**2), 0), (-16, -6)), ("F", (0, gf*vf), (6, -14))):
    d.event(t_, x_, name=name, offset=off)
d.show(loc="lower right")
print(f"t_E = {np.sqrt(1-vf**2):.3f} < t_D = 1 < t_C = γ = {gf:.3f}")
```

## (c) $\bar{\mathcal O}$ sees $\mathcal O$'s clocks run slow — at the "right" rate

$\bar{\mathcal O}$ compares **one** of $\mathcal O$'s clocks (the one at $x=0$, through A and E) with **two** of his own: the one through A, and the one at B (which is the same clock, B being on the $\bar t$ axis). His clocks are synchronized in his frame, so his comparison is legitimate. Between A and the event he regards as simultaneous with B:

- his own clock ticks $\tau_{AB} = \sqrt{-(\Delta s^2)_{AB}}$;
- $\mathcal O$'s clock ticks $\tau_{AE} = \sqrt{-(\Delta s^2)_{AE}}$.

By (b),
$$
\tau_{AE} = \sqrt{(1 - v^2)\big(-(\Delta s^2)_{AB}\big)} = \sqrt{1 - v^2}\;\tau_{AB}.
$$

So $\bar{\mathcal O}$ concludes $\mathcal O$'s clocks run slow by the factor $\sqrt{1-v^2}$. That's **exactly** the factor $\mathcal O$ found for $\bar{\mathcal O}$'s clocks, Eq. (1.8): $\tau_{AB} = \sqrt{1-v^2}\,\tau_{AC}$. This is the "right" rate the principle of relativity demands: the situation is symmetric, so each must see the other's clocks slowed by the same factor (it can depend only on $|v|$).

**Why there's no contradiction.** The two measurements are different experiments. $\mathcal O$ compares $\bar{\mathcal O}$'s clock at B with his own clock at C (simultaneous with B *for $\mathcal O$*). $\bar{\mathcal O}$ compares $\mathcal O$'s clock at E (simultaneous with B *for $\bar{\mathcal O}$*) with his own at B. Each uses his own definition of "now", and C ≠ E.
