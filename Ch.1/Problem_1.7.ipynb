# Chapter 1, Problem 7 — the numbers $M_{\alpha\beta}$ for a general linear transformation

**Problem (paraphrased).** In the discussion leading to Eq. (1.2), suppose $\bar{\mathcal O}$'s coordinates are the linear combinations
$$
\bar t = \alpha t + \beta x,\qquad \bar x = \mu t + \nu x,\qquad \bar y = a y,\qquad \bar z = b z,
$$
where $\alpha, \beta, \mu, \nu, a, b$ may depend on the relative velocity $v$ but not on the coordinates. Find the numbers $\{M_{\alpha\beta}\}$ of Eq. (1.2) in terms of $\alpha,\beta,\mu,\nu,a,b$.

*(Notation clash: here $\alpha$ and $\beta$ are coefficients, not indices. Below I write matrix entries as $M_{00}, M_{01},\dots$ with explicit numbers to avoid confusion. Also the book prints "$v$" in $\bar x = \mu t + vx$; from the list of coefficients it is $\nu$, not the velocity.)*

## Step 1 — coordinate *differences* transform the same way

The transformation is linear and homogeneous (origins coincide), so differences obey the same equations:
$$
\Delta\bar t = \alpha\,\Delta t + \beta\,\Delta x,\quad \Delta\bar x = \mu\,\Delta t + \nu\,\Delta x,\quad \Delta\bar y = a\,\Delta y,\quad \Delta\bar z = b\,\Delta z .
$$
(Subtract the equations for two events: $\bar t_2 - \bar t_1 = \alpha(t_2-t_1) + \beta(x_2-x_1)$, etc.)

## Step 2 — substitute into $\Delta\bar s^2$

$$
\Delta\bar s^2 = -(\Delta\bar t)^2 + (\Delta\bar x)^2 + (\Delta\bar y)^2 + (\Delta\bar z)^2
= -(\alpha\Delta t + \beta\Delta x)^2 + (\mu\Delta t + \nu\Delta x)^2 + a^2\Delta y^2 + b^2\Delta z^2 .
$$

## Step 3 — expand

$$
\begin{aligned}
-(\alpha\Delta t + \beta\Delta x)^2 &= -\alpha^2\Delta t^2 - 2\alpha\beta\,\Delta t\,\Delta x - \beta^2\Delta x^2\\
(\mu\Delta t + \nu\Delta x)^2 &= \mu^2\Delta t^2 + 2\mu\nu\,\Delta t\,\Delta x + \nu^2\Delta x^2
\end{aligned}
$$
so
$$
\Delta\bar s^2 = (\mu^2 - \alpha^2)\,\Delta t^2 + 2(\mu\nu - \alpha\beta)\,\Delta t\,\Delta x + (\nu^2 - \beta^2)\,\Delta x^2 + a^2\,\Delta y^2 + b^2\,\Delta z^2 .
$$

## Step 4 — read off $M$ (taking it symmetric, Problem 6)

Comparing with $\sum M_{\alpha\beta}\Delta x^\alpha\Delta x^\beta$, where the $\Delta t\,\Delta x$ coefficient is $M_{01}+M_{10} = 2M_{01}$:

$$
\boxed{
\begin{aligned}
M_{00} &= \mu^2 - \alpha^2, & M_{01} = M_{10} &= \mu\nu - \alpha\beta, & M_{11} &= \nu^2 - \beta^2,\\
M_{22} &= a^2, & M_{33} &= b^2, & \text{all others} &= 0 .
\end{aligned}}
$$

As a matrix:
$$
M = \begin{pmatrix}
\mu^2-\alpha^2 & \mu\nu-\alpha\beta & 0 & 0\\
\mu\nu-\alpha\beta & \nu^2-\beta^2 & 0 & 0\\
0 & 0 & a^2 & 0\\
0 & 0 & 0 & b^2
\end{pmatrix}.
$$

Compactly: if $\Delta\bar x = \Lambda\,\Delta x$ with $\Lambda$ the transformation matrix, then $\Delta\bar s^2 = \Delta x^T\,(\Lambda^T\eta\Lambda)\,\Delta x$, so $M = \Lambda^T\eta\Lambda$. That's automatically symmetric.

```python
import sympy as sp
al, be, mu, nu, a, b = sp.symbols("alpha beta mu nu a b", real=True)
dt, dx, dy, dz = sp.symbols(r"\Delta{t} \Delta{x} \Delta{y} \Delta{z}", real=True)

Lam = sp.Matrix([[al, be, 0, 0], [mu, nu, 0, 0], [0, 0, a, 0], [0, 0, 0, b]])
eta = sp.diag(-1, 1, 1, 1)
M = sp.simplify(Lam.T * eta * Lam)
display(M)

D = sp.Matrix([dt, dx, dy, dz])
direct = -(al*dt + be*dx)**2 + (mu*dt + nu*dx)**2 + a**2*dy**2 + b**2*dz**2
print("x^T M x equals the direct substitution:", sp.expand((D.T * M * D)[0] - direct) == 0)
```

## Check: the Lorentz transformation

For the boost of Eq. (1.12): $\alpha = \nu = \gamma$, $\beta = \mu = -\gamma v$, $a = b = 1$. Then
$$
M_{00} = \gamma^2v^2 - \gamma^2 = -\gamma^2(1-v^2) = -1,\quad
M_{01} = (-\gamma v)\gamma - \gamma(-\gamma v) = 0,\quad
M_{11} = \gamma^2 - \gamma^2v^2 = 1,
$$
so $M = \mathrm{diag}(-1,1,1,1) = \eta$. That's the invariance of the interval, $\Delta\bar s^2 = \Delta s^2$ ✓. It also matches the general result of §1.6: $M_{0i} = 0$, $M_{ij} = -M_{00}\,\delta_{ij}$, with $\phi(v) = -M_{00} = 1$.

```python
v = sp.symbols("v", real=True)
g = 1 / sp.sqrt(1 - v**2)
M_boost = sp.simplify(M.subs({al: g, nu: g, be: -g*v, mu: -g*v, a: 1, b: 1}))
display(M_boost)
```
