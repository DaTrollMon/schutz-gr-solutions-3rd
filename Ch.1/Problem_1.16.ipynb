# Chapter 1, Problem 16 — time dilation and Lorentz contraction from the Lorentz transformation

**Problem (paraphrased).** Use the Lorentz transformation, Eq. (1.12), to derive (a) the time dilation and (b) the Lorentz contraction formulae. Do this by identifying the pairs of events whose separations (in time or space) are to be compared, then use the Lorentz transformation to do the algebra that the invariant hyperbolae did in the text.

**Eq. (1.12)** ($\bar{\mathcal O}$ moving at $v$ along $+x$), and its inverse (swap bars, $v\to-v$):
$$
\bar t = \gamma(t - vx),\quad \bar x = \gamma(x - vt);\qquad
t = \gamma(\bar t + v\bar x),\quad x = \gamma(\bar x + v\bar t);\qquad \gamma = \frac{1}{\sqrt{1-v^2}} .
$$
Because the transformation is linear, the same equations hold for **differences** $\Delta t, \Delta x, \dots$ between two events. That's all we use.

The whole trick is choosing the right two events. A "duration" or "length" is only defined once you say which two events bound it and in which frame they're compared.

## (a) Time dilation

**What is being measured.** A clock at rest in $\bar{\mathcal O}$ (say at $\bar x = 0$) ticks. Take two ticks as events $\mathcal E_1$, $\mathcal E_2$:

- **In $\bar{\mathcal O}$** (the clock's rest frame) they're at the same place: $\Delta\bar x = 0$, $\Delta\bar t = \tau$, the proper time between ticks.
- **In $\mathcal O$** they're at different places (the clock moves). $\mathcal O$ measures the time between them, $\Delta t$, using two of his own synchronized clocks.

**Algebra.** Use the inverse transformation, since we know the barred differences:
$$
\Delta t = \gamma(\Delta\bar t + v\,\Delta\bar x) = \gamma(\tau + v\cdot0) = \gamma\,\tau .
$$
$$
\boxed{\Delta t = \frac{\Delta\bar t}{\sqrt{1 - v^2}}}\qquad\text{(Eq. 1.8)}
$$
$\mathcal O$ finds the moving clock's ticks spread out by $\gamma > 1$: the clock runs slow.

**Check.** $\Delta x = \gamma(\Delta\bar x + v\Delta\bar t) = \gamma v\tau = v\,\Delta t$ ✓ (the clock moves at $v$). And $-\Delta t^2 + \Delta x^2 = -\gamma^2\tau^2(1 - v^2) = -\tau^2 = -\Delta\bar t^2$ ✓ — the hyperbola argument of the text, now pure algebra.

**Pitfall.** Using the *forward* transformation $\Delta\bar t = \gamma(\Delta t - v\Delta x)$ with $\Delta x = 0$ gives $\Delta\bar t = \gamma\Delta t$ — the *opposite*. That formula is right for a different pair of events: ticks of a clock at rest in $\mathcal O$ ($\Delta x = 0$), seen by $\bar{\mathcal O}$. It's the reciprocal effect of Problem 12, not a contradiction. The physics is in which quantity vanishes.

## (b) Lorentz contraction

**What is being measured.** A rod at rest in $\bar{\mathcal O}$ along the $x$ axis, with proper length $\ell = \Delta\bar x$ (its ends are at fixed $\bar x$, so $\bar{\mathcal O}$ can measure them at leisure).

$\mathcal O$ measures its length by marking the positions of both ends **at the same time in $\mathcal O$**. So the relevant pair of events is: front end at $t$, rear end at the *same* $t$:
$$
\Delta t = 0 ,\qquad \Delta x = L = \text{length in }\mathcal O .
$$
These two events are not simultaneous in $\bar{\mathcal O}$, but that doesn't matter there: the ends don't move in $\bar{\mathcal O}$, so their $\bar x$ separation is $\ell$ whatever the $\bar t$'s are.

**Algebra.** Use the forward transformation for $\Delta\bar x$ (known: $\ell$) in terms of $\Delta t$ (known: 0) and $\Delta x$ (unknown):
$$
\ell = \Delta\bar x = \gamma(\Delta x - v\,\Delta t) = \gamma\,L
\quad\Longrightarrow\quad
\boxed{L = \ell\,\sqrt{1 - v^2}}\qquad\text{(Eq. 1.11)}
$$
The moving rod is shorter in $\mathcal O$ by $\sqrt{1-v^2}$.

**The time difference $\bar{\mathcal O}$ assigns to $\mathcal O$'s two "simultaneous" marks:** $\Delta\bar t = \gamma(0 - vL) = -\gamma v L = -v\ell$. Nonzero, so $\bar{\mathcal O}$ says $\mathcal O$ marked the ends at different times, which is why $\bar{\mathcal O}$ doesn't accept $\mathcal O$'s measurement as a length measurement in his frame. Compare the text's event C (on the $\bar x$ axis, $\bar t = 0$) and B (on the $x$ axis, $t = 0$), Fig. 1.13: the rear end is at the origin for both, the front end gives B for $\mathcal O$'s measurement and C for $\bar{\mathcal O}$'s.

```python
import sympy as sp
from schutz_gr import lorentz, inverse_lorentz, interval_tx
v = sp.symbols("v", positive=True)
tau, ell, L = sp.symbols("tau ell L", positive=True)
g = 1/sp.sqrt(1 - v**2)

# (a) two ticks of a clock at rest in Ō: Δx̄ = 0, Δt̄ = τ
dt, dx = inverse_lorentz(tau, 0, v)
print("(a) Δt =", sp.simplify(dt), "  Δx =", sp.simplify(dx), "  Δx/Δt =", sp.simplify(dx/dt))
print("    interval check:", sp.simplify(interval_tx(dt, dx)), "= −τ²")

# (b) ends of a rod at rest in Ō, marked simultaneously in O: Δt = 0, Δx = L, and Δx̄ = ℓ
dtb, dxb = lorentz(0, L, v)
sol = sp.solve(sp.Eq(dxb, ell), L)[0]
print("(b) Δx̄ =", sp.simplify(dxb), " = ℓ   ⇒   L =", sp.simplify(sol))
print("    Δt̄ of O's 'simultaneous' marks =", sp.simplify(dtb.subs(L, sol)))
```

```python
import numpy as np
from schutz_gr.diagrams import SpacetimeDiagram, COLORS
%config InlineBackend.figure_format = 'retina'
vf, lf = 0.6, 1.0
gf = 1/np.sqrt(1 - vf**2)
d = SpacetimeDiagram(xlim=(-0.4, 2.2), tlim=(-0.6, 1.8), title=f"(b) Rod at rest in Ō (v = {vf}, ℓ = 1): the two pairs of events", size=7)
d.boosted_axes(vf, color="amber", label="Ō axes")
T = np.linspace(-0.6, 1.8, 2)
d.line(vf*T, T, color="green", lw=2.2, label="rear end (x̄ = 0)")
d.line(vf*T + lf/gf, T, color="green", lw=2.2, ls="--", label="front end (x̄ = ℓ)")
d.event(0, 0, name="rear @ origin", offset=(-90, -14))
d.event(0, lf/gf, name="B: front at t = 0", offset=(6, -14))
tC, xC = gf*vf*lf, gf*lf
d.event(tC, xC, name="C: front at t̄ = 0", offset=(6, 4))
d.show(loc="upper left")
print(f"O's length (t = 0): {lf/gf:.3f} = ℓ√(1−v²);   Ō's length (t̄ = 0): ℓ = {lf}")
```
