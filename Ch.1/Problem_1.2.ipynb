# Chapter 1, Problem 2 — natural units ($c = 1$) → SI

**Problem (paraphrased).** Convert the following from $c = 1$ units back to SI units:

| | quantity | $c = 1$ value |
|---|---|---|
| (a) | velocity | $10^{-2}$ |
| (b) | pressure | $10^{19}\ \text{kg m}^{-3}$ |
| (c) | time | $10^{18}\ \text{m}$ |
| (d) | energy density | $1\ \text{kg m}^{-3}$ |
| (e) | acceleration | $10\ \text{m}^{-1}$ |

## The method — why the *kind* of quantity matters

Going from SI to $c=1$ is mechanical: replace every $\text{s}$ by $c\ \text{m}$. Going **back** is not: once the seconds are gone, the unit alone doesn't say how many there were. For example $\text{kg m}^{-3}$ could be a density (no seconds), a pressure ($\text{kg m}^{-1}\text{s}^{-2}$), or an energy flux ($\text{kg s}^{-3}$).

So for each part we use the **physical kind** of quantity to look up its SI unit, $\text{kg}^a\,\text{m}^{b_{SI}}\,\text{s}^{n}$. From Problem 1, the conversion SI → natural is

$$
X_{SI}\ \text{kg}^a\,\text{m}^{b_{SI}}\,\text{s}^{n} = \big(X_{SI}\,c^{\,n}\big)\ \text{kg}^a\,\text{m}^{\,b_{SI}+n}.
$$

Inverting:

$$
\boxed{X_{SI} = X_{\text{nat}}\,c^{-n}}
\qquad\text{and the natural unit must equal } \text{kg}^a\,\text{m}^{\,b_{SI}+n}\ \text{(a consistency check).}
$$

In words: **multiply by $c$ once for every $\text{s}^{-1}$ in the SI unit, divide by $c$ once for every $\text{s}$.**

```python
from schutz_gr import Quantity, to_si, C_SCHUTZ, C_EXACT
c = C_SCHUTZ
```

## (a) Velocity $10^{-2}$

**SI unit:** $\text{m s}^{-1}$, so $n=-1$, $b_{SI}=1$. **Check:** natural unit should be $\text{m}^{1-1} = $ dimensionless ✓.

**Convert:**
$$
v = 10^{-2}\times c = 10^{-2}\times3\times10^{8}\ \text{m s}^{-1} = \boxed{3\times10^{6}\ \text{m s}^{-1}}
$$

1% of light speed; about the speed of electrons in a hydrogen atom ($\alpha c \approx c/137$).

## (b) Pressure $10^{19}\ \text{kg m}^{-3}$

**SI unit:** $\text{Pa} = \text{N m}^{-2} = \text{kg m}^{-1}\text{s}^{-2}$, so $n=-2$, $b_{SI}=-1$. **Check:** $\text{kg m}^{-1-2} = \text{kg m}^{-3}$ ✓.

**Convert** (two factors of $\text{s}^{-1}$ ⇒ multiply by $c^2$):
$$
p = 10^{19}\times c^{2} = 10^{19}\times9\times10^{16}\ \text{kg m}^{-1}\text{s}^{-2} = \boxed{9\times10^{35}\ \text{N m}^{-2}}
$$

That's ~$10^{31}$ atmospheres — the regime of pressure deep inside a neutron star, where pressure (as an energy density, $10^{19}\ \text{kg m}^{-3}$) rivals the nuclear mass density ($\sim10^{17}\text{–}10^{18}\ \text{kg m}^{-3}$) and therefore matters as a source of gravity.

## (c) Time $10^{18}\ \text{m}$

**SI unit:** $\text{s}$, so $n=+1$, $b_{SI}=0$. **Check:** $\text{m}^{0+1} = \text{m}$ ✓.

**Convert** (one factor of $\text{s}$ ⇒ divide by $c$):
$$
t = \frac{10^{18}}{c} = \frac{10^{18}}{3\times10^{8}}\ \text{s} = \boxed{3.3\times10^{9}\ \text{s}}\ \approx 106\ \text{years}.
$$

A time of $10^{18}$ m is the time light takes to travel $10^{18}$ m ≈ 106 light-years.

## (d) Energy density $1\ \text{kg m}^{-3}$

**SI unit:** $\text{J m}^{-3} = \text{kg m}^{2}\text{s}^{-2}\cdot\text{m}^{-3} = \text{kg m}^{-1}\text{s}^{-2}$, so $n = -2$. **Check:** $\text{kg m}^{-3}$ ✓ (same SI unit as pressure, as found in Problem 1).

**Convert:**
$$
u = 1\times c^{2} = \boxed{9\times10^{16}\ \text{J m}^{-3}}
$$

This is just $E = mc^2$ per unit volume: the rest-mass energy of $1$ kg of matter in each cubic metre.

## (e) Acceleration $10\ \text{m}^{-1}$

**SI unit:** $\text{m s}^{-2}$, so $n=-2$, $b_{SI}=1$. **Check:** $\text{m}^{1-2}=\text{m}^{-1}$ ✓. (Acceleration = velocity/time = dimensionless/m.)

**Convert:**
$$
a = 10\times c^{2} = 10\times9\times10^{16}\ \text{m s}^{-2} = \boxed{9\times10^{17}\ \text{m s}^{-2}}
$$

About $10^{17}\,g$. **Meaning of $a = 10\ \text{m}^{-1}$:** an acceleration $a$ (in m⁻¹) changes your velocity by a fraction $a\,\Delta t$ of light speed in a time $\Delta t$ (in m). So $1/a = 0.1$ m is the time (≈ 0.33 ns) in which this acceleration would, Newtonianly, bring you from rest to light speed. Earth's $g = 9.8\ \text{m s}^{-2} \approx 1.1\times10^{-16}\ \text{m}^{-1}$, i.e. $1/g \approx$ one light-year — a classic result: a 1 g rocket gets relativistic in about a year.

## Check every part in code

```python
# natural-unit value as Quantity(kg=a, m=b_nat), plus the SI time exponent n from the physical kind
parts = {
    "(a) velocity 1e-2":           (Quantity(1e-2),            -1),
    "(b) pressure 1e19 kg/m^3":     (Quantity(1e19, kg=1, m=-3), -2),
    "(c) time 1e18 m":              (Quantity(1e18, m=1),         1),
    "(d) energy density 1 kg/m^3":  (Quantity(1.0,  kg=1, m=-3), -2),
    "(e) acceleration 10 /m":       (Quantity(10.0, m=-1),       -2),
}
print(f"{'part':30s} {'c = 3e8':>12s} {'exact c':>12s}  SI unit")
for name, (q, n) in parts.items():
    s1, s2 = to_si(q, n, c=C_SCHUTZ), to_si(q, n, c=C_EXACT)
    print(f"{name:30s} {s1.value:12.3e} {s2.value:12.3e}  {s1.unit}")

year = 365.25 * 24 * 3600
print(f"\n(c) in years: {to_si(Quantity(1e18, m=1), 1).value / year:.1f} yr")
print(f"(e) in units of g: {to_si(Quantity(10.0, m=-1), -2).value / 9.81:.2e} g")
```

## Summary

| part | $c=1$ | SI |
|---|---|---|
| (a) velocity | $10^{-2}$ | $3\times10^{6}\ \text{m s}^{-1}$ |
| (b) pressure | $10^{19}\ \text{kg m}^{-3}$ | $9\times10^{35}\ \text{N m}^{-2}$ |
| (c) time | $10^{18}\ \text{m}$ | $3.3\times10^{9}\ \text{s}$ |
| (d) energy density | $1\ \text{kg m}^{-3}$ | $9\times10^{16}\ \text{J m}^{-3}$ |
| (e) acceleration | $10\ \text{m}^{-1}$ | $9\times10^{17}\ \text{m s}^{-2}$ |

The one idea to keep: converting *back* to SI needs the physical meaning of the quantity, because $c = 1$ erases the bookkeeping of seconds.
