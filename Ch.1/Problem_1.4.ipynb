## Notation reminders

- **Superscripts are labels, not powers.** $x^2$ means "coordinate number 2", i.e. $y$; $\Delta x^\alpha$ is the $\alpha$-th coordinate *difference* between two events.
- $\alpha$ runs over $0,1,2,3$, with $0$ the time coordinate:
$$
\Delta x^0 = \Delta t,\quad \Delta x^1 = \Delta x,\quad \Delta x^2 = \Delta y,\quad \Delta x^3 = \Delta z.
$$
- Here the $V_\alpha$ are *just numbers* — no transformation law is attached to them yet. Each term is an ordinary product of numbers. (Lower-index position anticipates Chapter 3, where $V_\alpha$ will become the components of a one-form, but nothing in this problem depends on that.)

## (a) $\sum_\alpha V_\alpha\,\Delta x^\alpha$

**Step 1 — one term per value of the summed index.** The sum instructs: set $\alpha = 0$, then $1$, $2$, $3$, form $V_\alpha\,\Delta x^\alpha$ each time, and add.

$$
\begin{aligned}
\alpha = 0:&\quad V_0\,\Delta x^0\\
\alpha = 1:&\quad V_1\,\Delta x^1\\
\alpha = 2:&\quad V_2\,\Delta x^2\\
\alpha = 3:&\quad V_3\,\Delta x^3
\end{aligned}
$$

**Step 2 — add them.**
$$
\sum_{\alpha=0}^{3} V_\alpha\,\Delta x^\alpha = V_0\,\Delta x^0 + V_1\,\Delta x^1 + V_2\,\Delta x^2 + V_3\,\Delta x^3 .
$$

**Step 3 — substitute coordinate names.**
$$
\boxed{\sum_{\alpha=0}^{3} V_\alpha\,\Delta x^\alpha = V_0\,\Delta t + V_1\,\Delta x + V_2\,\Delta y + V_3\,\Delta z}
$$

Four terms, each linear in one coordinate difference: the result is a general **linear** function of $(\Delta t, \Delta x, \Delta y, \Delta z)$.

```python
import sympy as sp
import numpy as np

dt, dx, dy, dz = sp.symbols(r"\Delta{t} \Delta{x} \Delta{y} \Delta{z}")
DX = [dt, dx, dy, dz]                         # Δx^α, α = 0..3
V = sp.symbols("V_0:4")                       # V_α

S1 = sum(V[a] * DX[a] for a in range(4))
display(sp.Eq(sp.Symbol(r"\sum_{\alpha} V_\alpha \Delta x^\alpha"), S1, evaluate=False))
print("number of terms:", len(S1.args))
```

## (b) $\sum_{i=1}^{3}(\Delta x^i)^2$

**Two conventions to read correctly.**
1. **Latin index $\Rightarrow$ spatial only.** Schutz uses Greek indices ($\alpha, \beta, \dots$) for $0..3$ and Latin indices ($i, j, \dots$) for $1..3$. Here the sum runs $i = 1, 2, 3$ explicitly, so $\Delta x^0 = \Delta t$ does **not** appear.
2. **The outer exponent is a genuine square.** $(\Delta x^i)^2$ means "the $i$-th coordinate difference, squared". The parentheses separate the label $i$ (a superscript inside) from the power 2 (outside).

**Step 1 — one term per value of $i$.**
$$
\begin{aligned}
i = 1:&\quad (\Delta x^1)^2\\
i = 2:&\quad (\Delta x^2)^2\\
i = 3:&\quad (\Delta x^3)^2
\end{aligned}
$$

**Step 2 — add and substitute names** ($x^1 = x$, $x^2 = y$, $x^3 = z$):
$$
\boxed{\sum_{i=1}^{3}(\Delta x^i)^2 = (\Delta x)^2 + (\Delta y)^2 + (\Delta z)^2}
$$

**What it is.** This is the squared Euclidean distance $(\Delta r)^2$ between the two events' spatial positions (Pythagoras in 3D). It's exactly the spatial part of the interval, Eq. (1.1):
$$
\Delta s^2 = -(\Delta t)^2 + \sum_{i=1}^{3}(\Delta x^i)^2 = -(\Delta t)^2 + (\Delta r)^2 ,
$$
and the light-ray condition of §1.6 is $\Delta t = \Delta r = \big[\sum_i(\Delta x^i)^2\big]^{1/2}$. Three terms; it's a sum of squares, so never negative.

```python
S2 = sum(DX[i]**2 for i in (1, 2, 3))          # i = 1..3 only: Δt is excluded
display(sp.Eq(sp.Symbol(r"\sum_{i=1}^{3} (\Delta x^i)^2"), S2, evaluate=False))
print("number of terms:", len(S2.args), "  contains Δt?", S2.has(dt))
```

## Numerical sanity check with the summation convention

With Einstein's convention (Chapter 2) the $\sum$ in (a) is dropped: $V_\alpha\Delta x^\alpha$. `numpy.einsum` uses exactly this index notation, so it checks the term-by-term expansion.

```python
rng = np.random.default_rng(1)
Vn, Dn = rng.normal(size=4), rng.normal(size=4)
print("(a) einsum:", np.einsum("a,a->", Vn, Dn), "  term by term:", Vn[0]*Dn[0] + Vn[1]*Dn[1] + Vn[2]*Dn[2] + Vn[3]*Dn[3])
print("(b) einsum over i = 1..3:", np.einsum("i,i->", Dn[1:], Dn[1:]), "  term by term:", Dn[1]**2 + Dn[2]**2 + Dn[3]**2)
```

## Summary

$$
\sum_{\alpha=0}^{3} V_\alpha\,\Delta x^\alpha = V_0\,\Delta t + V_1\,\Delta x + V_2\,\Delta y + V_3\,\Delta z
\qquad\qquad
\sum_{i=1}^{3}(\Delta x^i)^2 = (\Delta x)^2 + (\Delta y)^2 + (\Delta z)^2
$$

(a) has four terms, a general linear function of all four coordinate differences. (b) has three terms, spatial only: the squared distance $(\Delta r)^2$ that appears in the interval.
